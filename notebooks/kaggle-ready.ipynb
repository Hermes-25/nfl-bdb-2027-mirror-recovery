# Relative Mirror Recovery and Rookie Protection: Measurement, Validation and Uncertainty

**Abhishek Das · NFL Big Data Bowl 2027 · Open Track**



## Abstract

We test whether Combine mirror-drill kinematics improve rookie offensive linemen's charted pressure forecasts beyond conventional Combine measurements. A speed-normalized reversal-loss profile represents movement during the turn while retaining incoming speed and excursion as task descriptors. The original frozen 2025 evaluation contains 19 rookies, 7,801 charted observations and 668 pressures. Adding mirror descriptors worsened equal-player Brier loss from 0.083684 to 0.084587; a constant forecast performed best. A subsequent, explicitly exploratory extension evaluates functional curves, nested selection, count-aware models and uncertainty. None establishes a dependable predictive advantage. The practical contribution is an auditable film-review measurement, with task-support checks and explicit limits on forecasting and player ranking.



## 1. Introduction

The mirror drill asks a lineman to reverse laterally before exiting forward. Tracking measures movement, but a player grade additionally requires reliability, comparable tasks and transfer to football outcomes. Incoming speed, excursion and instructions can affect movement loss [1].

We ask whether a pre-draft mirror profile improves rookie charted pressure forecasts beyond conventional Combine measurements. We distinguish the original frozen evaluation from extensions conducted after its outcomes were inspected. Those extensions diagnose failure; they cannot provide another untouched test.

### 2.2 Selection and measurement scope

Only 57 of 121 prospects support the original endpoint. Standardized differences are +0.536 and −0.453: retained-minus-excluded means divided by the corresponding all-development-prospect standard deviation (population divisor). Availability modeling describes selection, without causal correction.

Trajectories contain positions and timestamps, but no opponent trajectory, body orientation or stimulus timestamp; reaction latency and technique are unidentified. All 83 development linemen have one attempt. Multiple turns are movement events, not repeated tests.



## 3. Methods

### 3.1 Speed-relative reversal loss

We derive velocity from coordinates and timestamps after five-frame quadratic smoothing, bridge near-zero signs, and crop the forward exit beyond one yard from its initial reference. Usable reversals require opposite pre/post directions with median speeds above 0.5 yards/second; attempts require two usable reversals.

Let $v_y(t)$ denote lateral velocity and $\tilde v_{\mathrm{pre}},\tilde v_{\mathrm{post}}$ the pre/post-window median velocities. Set $s_{\mathrm{in}}=|\tilde v_{\mathrm{pre}}|$ and $u(t)=\operatorname{sign}(\tilde v_{\mathrm{post}})v_y(t)$. For a reversal at time zero,

$$
L=\frac{1}{s_{\mathrm{in}}}\int_{0}^{0.5}\max\{s_{\mathrm{in}}-u(t),0\}\,dt.\tag{1}
$$

With velocity in yards/second and time in seconds, $L$ measures effective speed-relative loss seconds, not stopwatch recovery. The profile contains the 75th percentile of losses, median incoming speed and lateral excursion (90th minus 10th position percentile). Excursion describes task demand.

The exploratory functional representation retains outgoing-signed velocity from −0.4 to +0.6 seconds, normalized by incoming speed. Dense-grid two-component PCA, inspired by functional-data analysis [2], is fitted inside training folds, retaining task descriptors separately; it does not implement sparse-data PACE or identify measurement noise. Training-only imputation and a missingness indicator retain all 38 players when one lacks a complete curve.

![Figure 1](https://www.googleapis.com/download/storage/v1/b/kaggle-user-content/o/inbox%2F36098817%2F19a7add8b4d79fa6dcf70bcc4b6b9422%2Ffigure1_film_card.png?generation=1791509179631163&alt=media)

**Figure 1. Mirror-drill measurement.** Paris Johnson's earliest usable 2023 turn was selected without NFL outcomes. (a) The grey +1-yard boundary marks forward-exit exclusion, distinct from the reversal. (b) The 0–0.5-second shortfall is 0.10369 yards; dividing by 1.47857 yards/second gives 0.07013 effective seconds. One attempt cannot establish test–retest reliability.

### 3.2 Original frozen evaluation

Conventional ridge uses height, weight, forty time and ten-yard split; mirror adds the three descriptors. Unverified pre-draft roster position was removed. Separate 2023 player folds selected penalties 10 and 1 with fold-specific preprocessing; 2024 was repeatedly inspected development data.

Before inspecting 2025 outcomes, we fixed the configuration/code hash, fitted both models on 38 development players and serialized their states. The original endpoint, models and result remain unchanged; later extensions are exploratory.

For $N$ players, $r_i=k_i/n_i$ is the rate of $k_i$ pressures in $n_i$ charted observations; $p_i$ is its forecast. Equal-player Brier loss is

$$
B=\frac{1}{N}\sum_{i=1}^{N}\{r_i(1-p_i)^2+(1-r_i)p_i^2\}.\tag{2}
$$

Players receive equal weight. Improvement $\Delta=B_{\mathrm{conventional}}-B_{\mathrm{mirror}}$ is positive when mirror wins.

### 3.3 Exploratory validation and uncertainty

Exact cohort-balanced outer five-fold and inner three-fold player assignments are fixed for the new comparisons. Players stay grouped; medians, scaling, PCA, penalties and eligible family selection are learned inside training folds [3]. Scalar/curve penalties are 1/10/100; the count-model benchmark adds 0.1 for its ridge and count fits. Known-cohort validation assesses retrospective stability.

Beta-binomial modeling [4] uses counts/exposure and training-only dispersion; new-player forecasts omit that player's count. A studentized jackknife+-style adaptation [5] omits each training player, repeats preprocessing and tuning, and scales leave-one-out errors by modeled count variability. Lower/upper finite-sample order statistics use alpha 0.05. Its empirical performance is reported without assuming that the original theorem's exchangeability and symmetry conditions hold here.



## 4. Results

### 4.1 The original forecast did not improve

**Table 2. Original frozen 2025 equal-player Brier loss.** Lower is better. The constant uses the development mean.

| Forecast | Brier loss |
|---|---:|
| Development-mean constant | **0.083513** |
| Conventional Combine | 0.083684 |
| Conventional + mirror | 0.084587 |

Mirror improvement was **−0.000902**. A 5,000-draw fixed-state 95% player-bootstrap interval was [−0.001734, −0.000179]; 250 pooled training/evaluation-player refits gave a 95% interval of [−0.002077, +0.000160]. Refitting exposes training instability. Neither interval includes earlier research choices, detector uncertainty or shared-game dependence.

The mirror forecast averaged 11.37 pressures per 100 observations versus an observed equal-player rate of 9.20; conventional averaged 9.25. Low prevalence makes absolute Brier scores small. Existing frozen-output rate MAE was 3.60 pressures per 100 observations for mirror, versus 2.35 conventional and 2.14 constant.

(c) The mirror mean forecast exceeds the observed rate. Neither bootstrap scope repeats the historical human model search.

### 4.2 Representation and model-selection instability

On the same 38 development players, nested conventional Brier loss was **0.08280965**, scalar mirror **0.08285783**, and functional curve **0.08285280**. The functional gain interval [−0.00016970, +0.00007652] crosses zero; all nine smoothing/crop variants lost to matched conventional baselines on the complete-curve subset.

Across 500 player resamples, duplicate identities stay grouped and preprocessing, penalty and family selection repeat. The conventional-minus-mirror interval is **[−0.00019798, +0.00027095]**. Eighteen of 20 alternate splits favor mirror; the fixed primary split does not. One deletion changes some mirror forecasts by 3.28 percentage points. Favorable split selection would obscure instability.

(c) The primary split, 20 alternate splits, 38 deletions and 500 fully retuned resamples remain visible. (d) Original 2025 intervals provide context. Sensitivities are dependent; player resampling omits shared-game dependence.

### 4.3 Count awareness does not establish forecasting value

Full-profile beta-binomial loss, **0.08280901**, essentially ties nested conventional ridge: its fixed-prediction difference interval spans zero and log loss slightly worsens versus constant.

Conditional central 90% count intervals cover **27/38** observed player rates, versus **34/38** for the count-model constant. The jackknife+-style adaptation covers **37/38 for both**, with median full-model width 12.12 pressures per 100 versus 9.95 for the constant. These constructions are not an equal-nominal-level contest. Coverage is empirical and conditional on exposure, without demonstrated future-NFL guarantees.

## 5. Discussion: A Defensible Scouting Use

### 5.1 Film-review workflow

The tool links turn timestamps, speed, excursion and trajectory to film, then retrieves five nearby task profiles using reference-scaled speed/excursion. A leave-prospect-out convex-hull check flags **11/83** unsupported tasks without NFL outcomes or a calibrated talent threshold. For these, show the clip and abstain from relative-efficiency ordering. Within support, retain match distances and ask whether the shortfall recurs under standardized instructions and incoming speed. Repeated standardized attempts are needed. The tool's usefulness to scouts is untested; no automatic pressure-risk or roster adjustment is justified.

### 5.2 Limitations and reproducibility

The 38 development and 19 historical test players are selected by opportunity. OL test–retest reliability, stimulus response and technique are unidentified. Known outcomes limit later validation. Fixed partitions, cohort drift and shared games leave exchangeability/symmetry unestablished; player resampling omits shared-game dependence. Instructions and stimuli were unrecorded.

The notebook preserves eight original frozen code cells and separates exploratory extensions, with hashes, exact folds, attempted comparisons and executable audits. Future deployment requires a later cohort, prospectively fixed protocol and standardized repetitions.



## 6. Conclusion

Mirror tracking makes reversal movement measurable, but this study does not establish improved rookie pressure forecasts. Its practical contribution is transparent film comparison, accompanied by population-selection, validation and uncertainty audits. Movement measurement and football prediction require separate evidence.

### A.1. Opportunity-selection and nested-stability module

The module below is readable source. Only three literal assignments became verified file reads; every function/class AST is unchanged.


````python
%%writefile paper_repro_assets/transport_readable.py
"""Self-contained retrospective coverage, validation and film-support appendix.

Only 2023/24 official records are processed. The original 2025 result is an
immutable archived reference, not recomputed or claimed as a new holdout.
Run with --raw-dir <five-official-CSV-directory> --output-dir <output-directory>.
Outputs only aggregate JSON, figures and a source-code ZIP. No player-row CSVs.
"""
from __future__ import annotations
import matplotlib
matplotlib.use('Agg')
import copy,zipfile,base64,zlib
from scipy.spatial import ConvexHull,QhullError
import csv, io, os, json, hashlib, datetime
from pathlib import Path
import numpy as np
import pandas as pd
import scipy
from scipy.signal import savgol_filter
from scipy.integrate import trapezoid
import matplotlib
import matplotlib.pyplot as plt

def locate_data():
    explicit = os.environ.get('NFL_DATA_DIR')
    if explicit:
        p = Path(explicit)
        if not all(((p / f).is_file() for f in REQUIRED)):
            raise FileNotFoundError('NFL_DATA_DIR must contain the five official CSV files.')
        return p
    verified = Path('/kaggle/input/competitions/nfl-big-data-bowl-2027/nfl-big-data-bowl-2027')
    if all(((verified / f).is_file() for f in REQUIRED)):
        return verified
    candidates = [p.parent for p in Path('/kaggle/input').glob('**/players.csv') if all(((p.parent / f).is_file() for f in REQUIRED))]
    if len(candidates) != 1:
        raise FileNotFoundError('Set NFL_DATA_DIR explicitly; input directory is ambiguous or absent.')
    return candidates[0]

def read_selected(path, predicate):
    buf = io.StringIO()
    with path.open(newline='') as f:
        reader = csv.reader(f)
        header = next(reader)
        index = {k: i for i, k in enumerate(header)}
        writer = csv.writer(buf)
        writer.writerow(header)
        for row in reader:
            if len(row) != len(header):
                raise ValueError('Malformed CSV record in ' + path.name)
            if predicate(row, index):
                writer.writerow(row)
    buf.seek(0)
    return pd.read_csv(buf, low_memory=False)

def load_inputs(data_dir, years=(2023, 2024), final_enabled=False):
    years = set((int(y) for y in years))
    if not years <= {2023, 2024, 2025}:
        raise ValueError('Unexpected cohort year.')
    if 2025 in years and (not final_enabled):
        raise PermissionError('2025 loader blocked BEFORE any file is opened.')
    allowed = {str(y) for y in years}
    d = Path(data_dir)
    players = read_selected(d / 'players.csv', lambda r, k: r[k['draft_year']] in allowed)
    results = read_selected(d / 'combine_results.csv', lambda r, k: r[k['draft_year']] in allowed)
    if not players.nfl_id.is_unique or results.duplicated(['draft_year', 'nfl_id']).any():
        raise ValueError('Non-unique player/Combine linkage key.')
    ol = results[results.combine_position.eq('OL')].copy()
    ids = {str(int(x)) for x in ol.nfl_id}
    draft = dict(zip(players.nfl_id.astype(str), players.draft_year.astype(int)))
    games = read_selected(d / 'games.csv', lambda r, k: r[k['season']] in allowed)
    if not games.game_id.is_unique:
        raise ValueError('Duplicate game identifiers.')
    reg = games[games.season_type.eq('REG')]
    gameyear = dict(zip(reg.game_id.astype(str), reg.season.astype(int)))
    tracking = read_selected(d / 'combine_tracking.csv', lambda r, k: r[k['draft_year']] in allowed and r[k['nfl_id']] in ids and (r[k['entity_type']] == 'PLAYER') and (r[k['drill_name']].replace('-', '_') == 'PASS_PRO_MIRROR_DRILL'))

    def rookie_record(r, k):
        game = r[k['game_id']]
        nid = r[k['nfl_id']]
        return game in gameyear and nid in ids and (gameyear[game] == draft.get(nid))
    player_play = read_selected(d / 'player_play.csv', rookie_record)
    assert set(players.draft_year) <= years and set(results.draft_year) <= years
    assert set(tracking.draft_year) <= years and set(games.season) <= years
    for a in years:
        for b in years:
            if a < b:
                assert not set(players[players.draft_year.eq(a)].nfl_id) & set(players[players.draft_year.eq(b)].nfl_id)
    if player_play.duplicated(['game_id', 'play_id', 'nfl_id']).any():
        raise ValueError('Duplicate player-play key.')
    return {'players': players, 'results': results, 'games': games, 'tracking': tracking, 'player_play': player_play, 'audit': {'years': sorted(years), 'accepted_rows': {'players': len(players), 'combine_results': len(results), 'combine_tracking_mirror': len(tracking), 'games': len(games), 'rookie_REG_player_play': len(player_play)}, 'drill_aliases': sorted(tracking.drill_name.unique().tolist()), 'initial_OL_cohort_source': 'combine_results.combine_position == OL', 'outcomes_selected_by_keys_before_typing': True, 'career_file_opened': False}}

def extract(t, window=5, exit_threshold=1.0):
    rows = []
    events = []
    examples = {}
    for key, g in t.groupby(['draft_year', 'event_id', 'nfl_id', 'attempt'], sort=True):
        g = g.sort_values('time').copy()
        tm = pd.to_datetime(g.time)
        sec = (tm - tm.iloc[0]).dt.total_seconds().to_numpy()
        x = g.x.to_numpy()
        y = g.y.to_numpy()
        start = np.median(x[:min(10, len(x))])
        exits = np.flatnonzero((np.abs(x - start) > exit_threshold) & (sec > 0.5))
        stop = int(exits[0]) if len(exits) else len(x)
        quality = 'ok'
        if stop < 15 or np.any(np.abs(np.diff(sec[:stop]) - 0.1) > 0.015):
            quality = 'short_or_non_10hz'
        if np.isnan(x).any() or np.isnan(y).any():
            quality = 'missing_coordinates'
        rev = []
        deficits = []
        if quality == 'ok':
            sy = savgol_filter(y[:stop], window, 2)
            vel = np.gradient(sy, sec[:stop])
            signs = np.where(np.abs(vel) > 0.05, np.sign(vel), 0)
            nz = np.flatnonzero(signs)
            changes = [int(round((i + j) / 2)) for i, j in zip(nz[:-1], nz[1:]) if signs[i] * signs[j] < 0 and j - i <= 10]
            last = -100
            for i in changes:
                if i < 5 or i + 5 >= stop or i - last < 5:
                    continue
                pre = np.median(vel[i - 5:i - 1])
                post = np.median(vel[i + 1:i + 5])
                if abs(pre) < 0.5 or abs(post) < 0.5 or pre * post >= 0:
                    continue
                new = np.sign(post) * vel[i:i + 6]
                deficit = float(trapezoid(np.maximum(0, abs(pre) - new), sec[i:i + 6] - sec[i]))
                rev.append(i)
                deficits.append(deficit)
                last = i
                events.append({'draft_year': key[0], 'event_id': key[1], 'nfl_id': key[2], 'attempt': key[3], 'reversal_time_seconds': float(sec[i]), 'incoming_speed_yards_s': float(abs(pre)), 'recovery_deficit_yards': deficit})
        if len(deficits) < 2 and quality == 'ok':
            quality = 'fewer_than_two_usable_reversals'
        rows.append({'draft_year': key[0], 'event_id': key[1], 'nfl_id': key[2], 'attempt': key[3], 'frames': len(g), 'mirror_frames': stop, 'usable_reversals': len(deficits), 'quality': quality, 'recovery_deficit': float(np.percentile(deficits, 75)) if quality == 'ok' else np.nan, 'x_start': float(start), 'x_range': float(np.ptp(x)), 'y_range': float(np.ptp(y))})
        examples[key] = {'sec': sec, 'x': x, 'y': y, 'stop': stop, 'rev': rev, 'quality': quality}
    return (pd.DataFrame(rows), pd.DataFrame(events), examples)

````

````python
%%writefile --append paper_repro_assets/transport_readable.py
def extract_mirror(t):
    keys = ['draft_year', 'event_id', 'nfl_id', 'attempt', 'time']
    if t.duplicated(keys).any():
        raise ValueError('Duplicate tracking timestamp key; extraction refused.')
    if t.empty:
        raise ValueError('No mirror trajectories; metric is undefined, not zero.')
    if not t.x.between(0, 120).all() or not t.y.between(0, 53.5).all():
        raise ValueError('Coordinate bounds incompatible with documented field-yard units.')
    return extract(t)

def build_features(t):
    attempts, events, examples = extract_mirror(t)
    keys = ['draft_year', 'event_id', 'nfl_id', 'attempt']
    good = attempts[attempts.quality.eq('ok')]
    if good.empty or events.empty:
        raise ValueError('No usable reversals; metric is undefined.')
    e = events.merge(good[keys], on=keys, validate='many_to_one')
    e['normalized_deficit'] = e.recovery_deficit_yards / e.incoming_speed_yards_s
    a = e.groupby(keys, as_index=False).agg(normalized_deficit=('normalized_deficit', lambda z: np.percentile(z, 75)), incoming_speed=('incoming_speed_yards_s', 'median'))
    task = []
    for _, row in good.iterrows():
        key = (int(row.draft_year), row.event_id, int(row.nfl_id), int(row.attempt))
        ex = examples[key]
        yy = ex['y'][:ex['stop']]
        task.append(dict(zip(keys, key), lateral_amplitude=float(np.percentile(yy, 90) - np.percentile(yy, 10)), phase_seconds=float(ex['sec'][ex['stop'] - 1])))
    z = good.merge(a, on=keys, validate='one_to_one').merge(pd.DataFrame(task), on=keys, validate='one_to_one')
    features = z.groupby(['draft_year', 'nfl_id'], as_index=False).agg(normalized_deficit=('normalized_deficit', 'median'), incoming_speed=('incoming_speed', 'median'), lateral_amplitude=('lateral_amplitude', 'median'), raw_deficit=('recovery_deficit', 'median'), usable_reversals=('usable_reversals', 'sum'), usable_attempts=('attempt', 'size'), phase_seconds=('phase_seconds', 'median'))
    return (features, attempts, events, examples)
import argparse, ast, hashlib, json, warnings
from pathlib import Path
import numpy as np
import pandas as pd
from scipy.optimize import minimize
from sklearn.model_selection import StratifiedKFold, StratifiedGroupKFold
from sklearn.metrics import roc_auc_score, average_precision_score
import matplotlib
import matplotlib.pyplot as plt
SEED = 87007
BASE = ['combine_height', 'combine_weight', 'forty', 'ten_yd_split']
MIRROR = ['normalized_deficit', 'incoming_speed', 'lateral_amplitude']
THRESHOLDS = [1, 25, 50, 100, 250]
CANDIDATES = [('constant', None)] + [(f, a) for f in ('combine', 'combine_mirror') for a in (1.0, 10.0, 100.0)]

def build_all_ol(raw: Path, years: list[int], source=None):
    ns = globals()
    inputs = ns['load_inputs'](raw, years, final_enabled=2025 in years)
    features, attempts, events, _ = ns['build_features'](inputs['tracking'])
    c = inputs['results'].query("combine_position == 'OL'").copy()
    p = inputs['players']
    g = inputs['games']
    base = c.merge(p[['draft_year', 'nfl_id', 'nfl_position']], on=['draft_year', 'nfl_id'], how='left', validate='one_to_one')
    base = base.merge(features, on=['draft_year', 'nfl_id'], how='left', validate='one_to_one')
    q = inputs['player_play'].merge(p[['draft_year', 'nfl_id']], on='nfl_id', validate='many_to_one').merge(g[['game_id', 'season', 'season_type']], on='game_id', validate='many_to_one')
    q = q[q.season.eq(q.draft_year) & q.season_type.eq('REG')].copy()
    eligible_rows = q.lined_up_position.isin(['C', 'G', 'T']) & q.team_abbr.eq(q.possession_team) & q.play_nullified_by_penalty.eq('N')
    q = q[eligible_rows].copy()
    observed = q[q.pressure_allowed.notna()].copy()
    labels = observed.pressure_allowed.astype(str).str.lower()
    if not set(labels) <= {'true', 'false'}:
        raise ValueError('Unknown explicit pressure label.')
    observed['pressure'] = labels.eq('true').astype(int)
    outcome = observed.groupby(['draft_year', 'nfl_id'], as_index=False).agg(n=('pressure', 'size'), k=('pressure', 'sum'))
    outcome['rate'] = outcome.k / outcome.n
    base = base.merge(outcome, on=['draft_year', 'nfl_id'], how='left', validate='one_to_one')
    anyreg = inputs['player_play'].groupby('nfl_id').size().rename('rookie_REG_rows')
    base = base.merge(anyreg, on='nfl_id', how='left', validate='one_to_one')
    present = set(attempts.nfl_id)
    base['mirror_present'] = base.nfl_id.isin(present)
    base['mirror_usable'] = base[MIRROR].notna().all(axis=1)
    base['observed_endpoint'] = base.n.notna()
    base['min50'] = base.n.ge(50) & base.mirror_usable
    audit = {'source_sha256': ORIGINAL_SOURCE_SHA256, 'input_audit': inputs['audit'], 'endpoint': 'rookie regular-season, offensive C/G/T, possession team, non-null explicit pressure_allowed, non-nullified', 'missing_pressure_rows_after_football_filter': int(q.pressure_allowed.isna().sum()), 'observed_zero_rusher_rows': int(observed.pass_rushers_encountered.eq(0).sum()), 'unconditional_talent_transport_identified': False}
    return (base, audit)

def loss(y, p):
    return y * (1 - p) ** 2 + (1 - y) * p * p

def preprocess(train, test, cols):
    x = train[cols].to_numpy(float)
    xt = test[cols].to_numpy(float)
    with warnings.catch_warnings():
        warnings.simplefilter('ignore', RuntimeWarning)
        med = np.nanmedian(x, axis=0)
    med = np.where(np.isfinite(med), med, 0)
    x = np.where(np.isfinite(x), x, med)
    xt = np.where(np.isfinite(xt), xt, med)
    mean = x.mean(0)
    std = x.std(0)
    std = np.where(std > 1e-12, std, 1.0)
    return ((x - mean) / std, (xt - mean) / std)

def predict(train, test, candidate):
    family, alpha = candidate
    y = train.rate.to_numpy(float)
    if family == 'constant':
        return np.full(len(test), y.mean())
    cols = BASE + (MIRROR if family == 'combine_mirror' else [])
    x, xt = preprocess(train, test, cols)
    coef = np.linalg.solve(x.T @ x + alpha * np.eye(x.shape[1]), x.T @ (y - y.mean()))
    return np.clip(y.mean() + xt @ coef, 0.001, 0.999)

def folds(frame, k, seed, grouped=False):
    strata = frame.draft_year.to_numpy()
    groups = frame.nfl_id.to_numpy()
    if grouped or len(np.unique(groups)) < len(groups):
        splitter = StratifiedGroupKFold(n_splits=k, shuffle=True, random_state=seed)
        return list(splitter.split(np.zeros(len(frame)), strata, groups))
    return list(StratifiedKFold(n_splits=k, shuffle=True, random_state=seed).split(np.zeros(len(frame)), strata))

def select_candidate(train, inner_folds, family=None):
    candidates = [c for c in CANDIDATES if family is None or c[0] == family]
    scores = []
    for c in candidates:
        p = np.full(len(train), np.nan)
        for ix, jx in inner_folds:
            p[jx] = predict(train.iloc[ix], train.iloc[jx], c)
        scores.append(float(loss(train.rate.to_numpy(), p).mean()))
    best = min(range(len(candidates)), key=lambda i: (scores[i], 0 if candidates[i][0] == 'constant' else 1 if candidates[i][0] == 'combine' else 2, -(candidates[i][1] or 0)))
    return (candidates[best], dict(((f'{c[0]}:{c[1]}', s) for c, s in zip(candidates, scores))))

def id_folds(frame, records):
    """Map immutable player-ID train/test assignments onto arbitrary row order."""
    ids = frame.nfl_id.to_numpy()
    seen = []
    result = []
    for r in records:
        train_ids = set((int(v) for v in r.get('train_ids', r.get('train_player_ids', []))))
        test_ids = set((int(v) for v in r.get('test_ids', r.get('test_player_ids', []))))
        assert train_ids and test_ids and (not train_ids & test_ids)
        assert train_ids | test_ids == set((int(v) for v in ids)), 'Common fold universe differs from eligible cohort.'
        tr = np.flatnonzero(np.isin(ids, list(train_ids)))
        te = np.flatnonzero(np.isin(ids, list(test_ids)))
        result.append((tr, te))
        seen.extend((int(v) for v in ids[te]))
    assert sorted(seen) == sorted((int(v) for v in ids)), 'Every player must be tested once in common fold map.'
    return result

````

````python
%%writefile --append paper_repro_assets/transport_readable.py
def map_records(mapping):
    if isinstance(mapping, list):
        return mapping
    for key in ['outer_folds', 'folds', 'outer']:
        if key in mapping:
            return mapping[key]
    raise ValueError('Common fold map must contain outer_folds with player-ID assignments.')

def nested_predictions(frame, seed=SEED, common_folds=None):
    out = pd.DataFrame({'draft_year': frame.draft_year, 'nfl_id': frame.nfl_id, 'rate': frame.rate, 'n': frame.n})
    for name in ['constant', 'combine', 'combine_mirror', 'selected_procedure']:
        out[name] = np.nan
    records = []
    outer_records = map_records(common_folds) if common_folds is not None else None
    outer = id_folds(frame, outer_records) if outer_records is not None else folds(frame, 5, seed)
    for fi, (tr, te) in enumerate(outer):
        train = frame.iloc[tr].reset_index(drop=True)
        test = frame.iloc[te].reset_index(drop=True)
        if outer_records is not None:
            inner_records = outer_records[fi].get('inner_folds', outer_records[fi].get('inner'))
            if inner_records is None:
                raise ValueError('Common outer map must include immutable inner_folds.')
            inner = id_folds(train, inner_records)
        else:
            inner = folds(train, 3, seed + 100 + fi)
        assert not set(train.nfl_id) & set(test.nfl_id), 'Player leaked outer fold.'
        for ii, jj in inner:
            assert not set(train.iloc[ii].nfl_id) & set(train.iloc[jj].nfl_id), 'Player leaked inner fold.'
        rec = {'fold': fi, 'train_ids': [int(x) for x in train.nfl_id], 'test_ids': [int(x) for x in test.nfl_id], 'inner_grid': {}}
        out.loc[out.index[te], 'constant'] = predict(train, test, ('constant', None))
        for family in ['combine', 'combine_mirror']:
            c, grid = select_candidate(train, inner, family)
            out.loc[out.index[te], family] = predict(train, test, c)
            rec[family] = {'chosen': c, 'scores': grid}
        c, grid = select_candidate(train, inner)
        out.loc[out.index[te], 'selected_procedure'] = predict(train, test, c)
        rec['selected_procedure'] = {'chosen': c, 'scores': grid}
        records.append(rec)
    assert out[['constant', 'combine', 'combine_mirror', 'selected_procedure']].notna().all().all()
    return (out, records)

def availability_predict(train, test, columns):
    x, xt = preprocess(train, test, columns)
    y = train.min50.astype(int).to_numpy()
    prevalence = np.clip(y.mean(), 1e-06, 1 - 1e-06)
    if y.min() == y.max():
        return np.full(len(test), prevalence)
    z = np.column_stack([np.ones(len(x)), x])
    zt = np.column_stack([np.ones(len(xt)), xt])
    from scipy.special import expit

    def obj(b):
        linear = z @ b
        val = (np.logaddexp(0, linear) - y * linear).sum() + 5.0 * np.sum(b[1:] ** 2)
        grad = z.T @ (expit(linear) - y)
        grad[1:] += 10.0 * b[1:]
        return (val, grad)
    fitted = minimize(obj, np.r_[np.log(prevalence / (1 - prevalence)), np.zeros(x.shape[1])], jac=True, method='L-BFGS-B')
    if not fitted.success:
        raise RuntimeError('Availability logistic fit failed: ' + fitted.message)
    return expit(zt @ fitted.x)

def coverage_audit(all_players):
    funnel = []
    for year, g in all_players.groupby('draft_year'):
        funnel.append({'draft_year': int(year), 'Combine_OL': len(g), 'mirror_present': int(g.mirror_present.sum()), 'usable_mirror': int(g.mirror_usable.sum()), 'any_rookie_REG': int(g.rookie_REG_rows.notna().sum()), 'observed_endpoint': int(g.observed_endpoint.sum()), 'eligible_min50_and_mirror': int(g.min50.sum())})
    rows = []
    for threshold in THRESHOLDS:
        keep = all_players.mirror_usable & all_players.n.ge(threshold)
        for year, g in all_players.groupby('draft_year'):
            select = g.mirror_usable & g.n.ge(threshold)
            rows.append({'threshold': threshold, 'draft_year': int(year), 'retained': int(select.sum()), 'total': len(g), 'charted_observations': int(g.loc[select, 'n'].sum()), 'pressures': int(g.loc[select, 'k'].sum())})
    smd = []
    selected = all_players.min50
    for c in BASE + MIRROR:
        values = pd.to_numeric(all_players[c], errors='coerce')
        scale = values.std(ddof=0)
        a = values[selected].dropna()
        b = values[~selected].dropna()
        smd.append({'feature': c, 'retained_mean': float(a.mean()), 'excluded_mean': float(b.mean()), 'standardized_mean_difference': float((a.mean() - b.mean()) / scale), 'retained_missing': int(values[selected].isna().sum()), 'excluded_missing': int(values[~selected].isna().sum())})
    pred = np.full(len(all_players), np.nan)
    for tr, te in folds(all_players, 5, SEED):
        pred[te] = availability_predict(all_players.iloc[tr], all_players.iloc[te], BASE + MIRROR)
    y = all_players.min50.astype(int).to_numpy()
    obs = {'purpose': 'Cross-fitted endpoint availability diagnostic, not ability prediction or causal correction.', 'fixed_logistic_L2_penalty': 10.0, 'player_oof_auc': float(roc_auc_score(y, pred)), 'player_oof_average_precision': float(average_precision_score(y, pred)), 'prevalence': float(y.mean()), 'player_oof_brier': float(np.mean((y - pred) ** 2)), 'predicted_min': float(pred.min()), 'predicted_max': float(pred.max()), 'eligible_propensity_range': [float(pred[y == 1].min()), float(pred[y == 1].max())], 'excluded_propensity_range': [float(pred[y == 0].min()), float(pred[y == 0].max())]}
    return (funnel, rows, smd, obs, pred)

def bootstrap_comparison(pred, draws=5000, seed=SEED):
    rng = np.random.default_rng(seed)
    y = pred.rate.to_numpy()
    diff = loss(y, pred['combine'].to_numpy()) - loss(y, pred['combine_mirror'].to_numpy())
    groups = [np.flatnonzero(pred.draft_year.to_numpy() == yr) for yr in sorted(pred.draft_year.unique())]
    sampled = np.concatenate([rng.choice(ix, size=(draws, len(ix)), replace=True) for ix in groups], axis=1)
    b = diff[sampled].mean(1)
    return {'baseline_minus_mirror_brier': float(diff.mean()), 'fixed_outer_prediction_player_bootstrap_95': np.quantile(b, [0.025, 0.975]).tolist(), 'resampling_unit': 'player within cohort', 'selection_repeated_in_this_interval': False, 'interval_is_not_independent_generalization_guarantee': True}

def retuned_bootstrap(frame, draws):
    rng = np.random.default_rng(SEED + 31)
    records = []
    groups = [np.flatnonzero(frame.draft_year.to_numpy() == yr) for yr in sorted(frame.draft_year.unique())]
    for b in range(draws):
        sampled = np.concatenate([rng.choice(ix, size=len(ix), replace=True) for ix in groups])
        boot = frame.iloc[sampled].reset_index(drop=True)
        try:
            p, r = nested_predictions(boot, seed=SEED + 1000 + b)
        except ValueError as e:
            raise RuntimeError(f'Bootstrap {b} lacked grouped-fold feasibility.') from e
        y = p.rate.to_numpy()
        records.append({'baseline_minus_mirror': float((loss(y, p['combine'].to_numpy()) - loss(y, p['combine_mirror'].to_numpy())).mean()), 'constant_minus_selected': float((loss(y, p['constant'].to_numpy()) - loss(y, p['selected_procedure'].to_numpy())).mean()), 'mirror_selected_folds': sum((q['selected_procedure']['chosen'][0] == 'combine_mirror' for q in r))})
    return records

def cohort_transfer(frame):
    result = []
    for yr in sorted(frame.draft_year.unique()):
        train = frame[frame.draft_year.ne(yr)].reset_index(drop=True)
        test = frame[frame.draft_year.eq(yr)].reset_index(drop=True)
        if len(train) < 10 or len(test) < 5:
            continue
        inner = folds(train, 3, SEED + int(yr))
        rec = {'test_year': int(yr), 'train_years': sorted((int(x) for x in train.draft_year.unique())), 'interpretation': 'Retrospective cross-cohort transfer; backward-facing folds are not forecasting.'}
        y = test.rate.to_numpy()
        for fam in ['constant', 'combine', 'combine_mirror']:
            c, grid = select_candidate(train, inner, fam)
            p = predict(train, test, c)
            rec[fam] = {'brier': float(loss(y, p).mean()), 'selected': c}
        result.append(rec)
    return result

def figures(output, all_players, threshold_rows, smd, pred, availability):
    fig, ax = plt.subplots(1, 2, figsize=(11, 4.5))
    th = pd.DataFrame(threshold_rows)
    for yr, g in th.groupby('draft_year'):
        ax[0].plot(g.threshold, g.retained / g.total, 'o-', label=str(yr))
    ax[0].set(xlabel='Minimum charted rookie observations', ylabel='Share of Combine OL retained', ylim=(0, 1))
    ax[0].legend(title='Draft cohort')
    d = pd.DataFrame(smd)
    ax[1].barh(d.feature, d.standardized_mean_difference, color='#367786')
    ax[1].axvline(0, color='black', linewidth=0.8)
    ax[1].set(xlabel='Retained − excluded mean / all-prospect SD')
    fig.suptitle('NFL outcome coverage changes the prospect population')
    fig.tight_layout()
    fig.savefig(output / 'coverage_selection.png', dpi=180)
    plt.close(fig)
    fig, ax = plt.subplots(1, 2, figsize=(11, 4.5))
    for flag, label in [(False, 'Excluded'), (True, 'Retained ≥50')]:
        v = availability[all_players.min50.to_numpy() == flag]
        ax[0].hist(v, bins=np.linspace(0, 1, 12), alpha=0.6, label=label)
    ax[0].set(xlabel='Cross-fitted probability of endpoint eligibility', ylabel='Prospects')
    ax[0].legend()
    y = pred.rate.to_numpy()
    values = [loss(y, pred[c].to_numpy()).mean() for c in ['constant', 'combine', 'combine_mirror', 'selected_procedure']]
    ax[1].bar(['Constant', 'Combine', 'Combine +\nmirror', 'Nested selection'], values, color=['#808080', '#367786', '#b86b38', '#745889'])
    ax[1].set(ylabel='Nested outer-fold equal-player Brier loss')
    ax[1].tick_params(axis='x', labelsize=9)
    fig.suptitle('Observability and fully specified selection are distinct questions')
    fig.tight_layout()
    fig.savefig(output / 'nested_validation.png', dpi=180)
    plt.close(fig)
````

````python
%%writefile --append paper_repro_assets/transport_readable.py
REQUIRED=['players.csv', 'combine_results.csv', 'combine_tracking.csv', 'games.csv', 'player_play.csv']
ORIGINAL_SOURCE_SHA256='9869aa86b0bf6b87a832b984f1efdfb1fac13d9ec6fbe1567fdefa4d6cbb56f6'
COMMON_FOLD_SHA256='4f5ad7a2fef3dbccb0abd23e627546106d90056eace24776f4e673e58a397b34'
COMMON_FOLDS=json.loads(Path(__file__).with_name('common_fold_manifest.json').read_text())
ARCHIVED_REFERENCE=json.loads(Path(__file__).with_name('archived_reference.json').read_text())

FOLD_PLAN_SHA256='28ab0fd1fb3fe896514fbf1aeb9356e218ba0e51d992159f0d1c54ec63945c5e'
FROZEN_SENSITIVITY_PLANS=json.loads(Path(__file__).with_name('sensitivity_plan.json').read_text())
SPLIT_PLANS=FROZEN_SENSITIVITY_PLANS["splits"]
BOOTSTRAP_PLANS=FROZEN_SENSITIVITY_PLANS["bootstrap"]


def retuned_bootstrap(frame,draws):
    if draws>len(BOOTSTRAP_PLANS):raise ValueError('At most 500 frozen bootstrap plans are bundled.')
    rng=np.random.default_rng(SEED+31);records=[]
    groups=[np.flatnonzero(frame.draft_year.to_numpy()==yr) for yr in sorted(frame.draft_year.unique())]
    for b in range(draws):
        sampled=np.concatenate([rng.choice(ix,size=len(ix),replace=True) for ix in groups])
        boot=frame.iloc[sampled].reset_index(drop=True)
        p,r=nested_predictions(boot,common_folds=BOOTSTRAP_PLANS[b])
        y=p.rate.to_numpy()
        records.append({'baseline_minus_mirror':float((loss(y,p['combine'].to_numpy())-loss(y,p['combine_mirror'].to_numpy())).mean()),'constant_minus_selected':float((loss(y,p['constant'].to_numpy())-loss(y,p['selected_procedure'].to_numpy())).mean()),'mirror_selected_folds':sum(q['selected_procedure']['chosen'][0]=='combine_mirror' for q in r)})
    return records

def trim_map(mapping, remove):
    result = copy.deepcopy(mapping)
    for fold in result['outer_folds']:
        for key in ['train_ids', 'test_ids']:
            fold[key] = [x for x in fold[key] if x != remove]
        for inner in fold['inner_folds']:
            for key in ['train_ids', 'test_ids']:
                inner[key] = [x for x in inner[key] if x != remove]
    return result

def summary(pred):
    y = pred.rate.to_numpy()
    scores = {name: float(loss(y, pred[name].to_numpy()).mean()) for name in ['constant', 'combine', 'combine_mirror', 'selected_procedure']}
    scores['combine_minus_mirror'] = scores['combine'] - scores['combine_mirror']
    scores['constant_minus_selected'] = scores['constant'] - scores['selected_procedure']
    return scores
def film_support(frame):
    TASK=['incoming_speed','lateral_amplitude']
    valid=frame.mirror_usable&frame[TASK].notna().all(axis=1)
    ref=frame[valid].reset_index(drop=True)
    x=ref[TASK].to_numpy(float);mean=x.mean(0);scale=x.std(0);scale=np.where(scale>1e-12,scale,1)
    z=(x-mean)/scale
    rows=[]
    for i,row in ref.iterrows():
     other=np.arange(len(ref))!=i;train=z[other];target=z[i]
     try:
      hull=ConvexHull(train)
      inside=bool(np.all(hull.equations[:,:-1]@target+hull.equations[:,-1]<=1e-10))
     except QhullError:inside=False
     distances=np.sqrt(np.sum((train-target)**2,axis=1));indices=np.flatnonzero(other);closest=indices[np.argsort(distances)[:5]]
     rows.append({'draft_year':int(row.draft_year),'nfl_id':int(row.nfl_id),
                  'feature_only_joint_task_support':inside,
                  'film_comparison_status':'Within measured joint task support; review comparable clips descriptively' if inside else 'Abstain from relative movement ranking: outside measured joint task support',
                  'automatic_pressure_forecast_adjustment':'ABSTAIN: original frozen 2025 test did not confirm improvement',
                  'physiological_repeatability':'UNESTABLISHED: one recorded attempt per player',
                  'median_incoming_speed_yards_s':float(row.incoming_speed),'attempt_excursion_yards':float(row.lateral_amplitude),
                  'normalized_deficit_seconds':float(row.normalized_deficit),'charted_endpoint_eligible_min50':bool(row.min50),
                  'nearest_reference_player_ids':';'.join(str(int(v)) for v in ref.iloc[closest].nfl_id),
                  'nearest_task_distance_standardized':float(distances.min()),
                  'fifth_nearest_task_distance_standardized':float(np.sort(distances)[4])})
    summary={'status':'executed_feature_only_reporting_rule','round':9,'reference_cohorts':[2023,2024],'measured_prospects':len(ref),
             'joint_task_supported':sum(r['feature_only_joint_task_support'] for r in rows),'outside_joint_task_support':sum(not r['feature_only_joint_task_support'] for r in rows),
             'automatic_forecast_adjustments_allowed':0,'covariate_input_sha256':hashlib.sha256(frame[['draft_year','nfl_id']+TASK].to_csv(index=False).encode()).hexdigest(),
             'algorithm':'Leave-the-prospect-out convex hull in incoming-speed × excursion space; nearest five reference IDs and distances for descriptive film comparison. No NFL outcome value enters this rule.',
             'limits':['Geometric support is not a confidence interval, physiological reliability, or equal-opportunity football comparison.','The reference is only the 2023/24 Combine measured OL sample.','Inside-hull comparison does not identify athlete talent or a pressure-risk effect.','No validated numerical threshold for scouting decisions is claimed.'],
             'film_rule':['Require usable trajectory, known turn timestamps and at least two valid reversals.','If incoming speed and excursion fall outside joint measured reference support, show the clip but abstain from relative efficiency ranking.','Within support, compare nearest reference clips at documented task ranges and formulate a coaching question; retain the distances.','Do not translate one recorded attempt into a reliability claim.','Make no automatic pressure forecast or roster adjustment, given the failed original frozen 2025 result and retrospective selection sensitivity.']}
    return summary

def publication_plot(d,output):
    fig,ax=plt.subplots(2,2,figsize=(12.5,9.4));blue='#367786';orange='#b86b38';purple='#745889'
    # A: exact roster coverage, including original reserved cohort.
    rows=d['archive']['cohort_coverage'];x=np.arange(3);w=.34
    ax[0,0].bar(x-w/2,[r['combine_OL'] for r in rows],w,color='#bec5ca',label='Combine OL')
    ax[0,0].bar(x+w/2,[r['eligible_min50'] for r in rows],w,color=blue,label='Usable mirror + ≥50 charted observations')
    for i,r in enumerate(rows):ax[0,0].text(i+w/2,r['eligible_min50']+.8,f"{r['eligible_min50']}/{r['combine_OL']}",ha='center',fontsize=9)
    ax[0,0].set(xticks=x,xticklabels=['2023','2024','2025 original test'],ylabel='Independent prospects',ylim=(0,57),title='A  Only 57 of 121 prospects support the original endpoint')
    ax[0,0].legend(fontsize=8,loc='upper left')
    # B: full fixed feature list, no selective imbalance reporting.
    labels={'combine_height':'Height','combine_weight':'Weight','forty':'40-yard time','ten_yd_split':'10-yard split','normalized_deficit':'Relative recovery cost','incoming_speed':'Incoming lateral speed','lateral_amplitude':'Lateral excursion'}
    s=d['r7']['selection_smd'];y=np.arange(len(s));vals=[r['standardized_mean_difference'] for r in s]
    ax[0,1].barh(y,vals,color=[blue if abs(v)>.4 else '#99adb5' for v in vals]);ax[0,1].axvline(0,color='black',linewidth=.8)
    ax[0,1].set(yticks=y,yticklabels=[labels[r['feature']] for r in s],xlabel='Retained − excluded mean / all-prospect SD',title='B  Selection changes the measured OL population')
    ax[0,1].invert_yaxis()
    # C: all sensitivity directions retained, no favorable split promotion.
    rng=np.random.default_rng(989)
    p=d['r8']['primary']['combine_minus_mirror'];ax[1,0].scatter([p*1e4],[0],marker='D',color='black',s=60,zorder=3)
    for yy,key,col in [(1,'splits',purple),(2,'delete',blue)]:
     v=np.array([r['combine_minus_mirror'] for r in d[key]])
     ax[1,0].scatter(v*1e4,np.full(len(v),yy)+rng.uniform(-.09,.09,len(v)),s=20,alpha=.8,color=col)
    b=np.array([r['baseline_minus_mirror'] for r in d['boot']]);lo,hi=np.quantile(b,[.025,.975]);median=np.median(b)
    ax[1,0].errorbar(median*1e4,3,xerr=1e4*np.array([[median-lo],[hi-median]]),fmt='o',color=orange,capsize=5)
    ax[1,0].axvline(0,color='black',linestyle='--',linewidth=.8)
    ax[1,0].set(yticks=[0,1,2,3],yticklabels=['Shared primary split','20 split repeats','38 prospect deletions','500 nested resamples'],xlabel='Brier improvement (×10⁻⁴); positive favors mirror',ylim=(-.4,3.5),title='C  Mirror evidence depends on resampling')
    ax[1,0].invert_yaxis()
    # D: retain both uncertainty scopes for the original frozen result.
    reserved=d['archive']['signed_comparisons'][1];delta=reserved['baseline_minus_mirror']
    for yy,key,col in [(0,'fixed_prediction95',orange),(1,'refit95',blue)]:
     lo,hi=reserved[key];ax[1,1].errorbar(delta*1e3,yy,xerr=1e3*np.array([[delta-lo],[hi-delta]]),fmt='o',color=col,capsize=6)
    ax[1,1].axvline(0,color='black',linestyle='--',linewidth=.8)
    ax[1,1].set(yticks=[0,1],yticklabels=['Fixed fitted states','Refit train + test players'],xlabel='Brier improvement (×10⁻³); positive favors mirror',ylim=(-.5,1.55),title='D  Original frozen 2025 gain was not confirmed')
    ax[1,1].invert_yaxis()
    fig.suptitle('Coverage and validation limits for mirror-based scouting',fontsize=15)
    fig.tight_layout(rect=(0,0,1,.96));fig.savefig(output/'coverage_selection_stability.png',dpi=200);plt.close(fig)

def run_appendix(raw,output,bootstrap_draws=500,split_repeats=20):
    output.mkdir(parents=True,exist_ok=True)
    all_players,source_audit=build_all_ol(raw,[2023,2024])
    assert len(all_players)==83
    frame=all_players[all_players.min50].sort_values(['draft_year','nfl_id']).reset_index(drop=True)
    assert len(frame)==38 and frame.groupby('draft_year').size().to_dict()=={2023:18,2024:20}
    assert int(frame.n.sum())==16165 and int(frame.k.sum())==1444
    funnel,thresholds,smd,observability,availability=coverage_audit(all_players)
    primary,records=nested_predictions(frame,common_folds=COMMON_FOLDS)
    expected=[0.08281911311702472,0.08280964600983351,0.08285783299123362,0.08287065760981599]
    actual=summary(primary)
    for name,value in zip(['constant','combine','combine_mirror','selected_procedure'],expected):assert abs(actual[name]-value)<1e-12
    result7={'status':'executed_retrospective_analysis','years':[2023,2024],'eligible_players':len(frame),'eligible_observations':int(frame.n.sum()),'eligible_pressures':int(frame.k.sum()),'source_audit':source_audit,'funnel':funnel,'opportunity_thresholds':thresholds,'selection_smd':smd,'availability_diagnostic':observability,'metrics':actual,'comparison':bootstrap_comparison(primary),'common_fold_sha256':COMMON_FOLD_SHA256,'no_new_holdout':True}
    print('R7 reproduced on official data: 83 Combine OL, 38 eligible players; exact shared-fold scores match.',flush=True)
    indexed=primary.set_index('nfl_id');deletions=[]
    for remove in frame.nfl_id:
        reduced=frame[frame.nfl_id.ne(remove)].reset_index(drop=True)
        pred,rec=nested_predictions(reduced,common_folds=trim_map(COMMON_FOLDS,int(remove)))
        aligned=indexed.loc[pred.nfl_id]
        row={'removed_nfl_id':int(remove),**summary(pred)}
        for name in ['combine','combine_mirror','selected_procedure']:row[name+'_max_prediction_shift']=float(np.max(np.abs(pred[name].to_numpy()-aligned[name].to_numpy())))
        deletions.append(row)
    splits=[]
    for i in range(split_repeats):
        seed=SEED+7000+i;pred,rec=nested_predictions(frame,common_folds=SPLIT_PLANS[i])
        splits.append({'seed':seed,**summary(pred)})
    print('All 38 deletion analyses and 20 nested split repeats reproduced.',flush=True)
    boot=retuned_bootstrap(frame,bootstrap_draws)
    result8={'status':'executed_retrospective_robustness','primary':actual,'deletion_count':len(deletions),'deletion_delta_range':[min(r['combine_minus_mirror'] for r in deletions),max(r['combine_minus_mirror'] for r in deletions)],'deletion_mirror_favorable_count':sum(r['combine_minus_mirror']>0 for r in deletions),'max_deleted_player_mirror_prediction_shift':max(r['combine_mirror_max_prediction_shift'] for r in deletions),'split_repeats':len(splits),'split_mirror_favorable_count':sum(r['combine_minus_mirror']>0 for r in splits),'split_delta_mean':float(np.mean([r['combine_minus_mirror'] for r in splits])),'retuned_bootstrap_draws':len(boot),'retuned_bootstrap_95':np.quantile([r['baseline_minus_mirror'] for r in boot],[.025,.975]).tolist(),'retuned_constant_minus_selected_95':np.quantile([r['constant_minus_selected'] for r in boot],[.025,.975]).tolist(),'selected_mirror_bootstrap_frequency':float(sum(r['mirror_selected_folds'] for r in boot)/(5*len(boot)))}
    film=film_support(all_players)
    assert film['measured_prospects']==83 and film['joint_task_supported']==72 and film['outside_joint_task_support']==11
    assert result8['deletion_mirror_favorable_count']==4 and result8['split_mirror_favorable_count']==18
    assert abs(result8['max_deleted_player_mirror_prediction_shift']-.03278123247008892)<1e-12
    if bootstrap_draws==500:
        assert np.allclose(result8['retuned_bootstrap_95'],[-.00019797616882475408,.0002709501818602032],atol=1e-12,rtol=0)
    result={'status':'all_self_contained_appendix_checks_passed','scope':'Only 2023/24 official records; all new work retrospective after original 2025 inspection.','original_2025_archived_not_recomputed':ARCHIVED_REFERENCE,'R7':result7,'R8':result8,'R9_film_support':film,'outputs_exclude_player_row_CSVs':True,'original_frozen_result_changed':False,'new_holdout_claimed':False,'new_predictive_gain_established':False,'common_fold_sha256':COMMON_FOLD_SHA256,'frozen_sensitivity_plan_sha256':FOLD_PLAN_SHA256,'sensitivity_fold_memberships_embedded_to_avoid_sklearn_version_changes':True,'limits':['Specified nested procedure does not capture prior human research choices.','Player resampling excludes shared-team/game dependence.','Geometry support is not physiological repeatability, a causal effect, or a calibrated talent threshold.','No automatic pressure-risk or roster adjustment.']}
    publication_plot({'archive':ARCHIVED_REFERENCE,'r7':result7,'r8':result8,'delete':deletions,'splits':splits,'boot':boot},output)
    # Build ordinary coverage/nested figure only; no per-player predictions are written.
    figures(output,all_players,thresholds,smd,primary,availability)
    result['figure_sha256']={p.name:hashlib.sha256(p.read_bytes()).hexdigest() for p in output.glob('*.png')}
    (output/'TRANSPORT_APPENDIX_AGGREGATE.json').write_text(json.dumps(result,indent=2))
    with zipfile.ZipFile(output/'TRANSPORT_APPENDIX_SOURCE.zip','w',compression=zipfile.ZIP_DEFLATED) as zipped:
        zipped.write(Path(__file__).resolve(),arcname='Notebook_Appendix.py')
    print(json.dumps({'status':result['status'],'primary':actual,'retuned_bootstrap_95':result8['retuned_bootstrap_95'],'support_outside':film['outside_joint_task_support'],'aggregate':str(output/'TRANSPORT_APPENDIX_AGGREGATE.json')},indent=2),flush=True)
    return result

````

### A.2. functional_eb.py


````python
%%writefile functional_code/functional_eb.py
"""Dense-grid functional conditioning for repeated kinematic turn curves.

This is a simple measurement-aware finite-grid analogue of functional conditional
estimation, NOT a reproduction of the sparse-grid PACE algorithm. Every fit must
occur inside the training split. Turns from one attempt are dependent: aggregate
them into one attempt curve before using this estimator.
"""
from __future__ import annotations

import numpy as np
import pandas as pd


class FunctionalEB:
    """Estimate repeat-attempt noise, then condition two latent curve components.

    DataFrame columns: player_id, attempt_id, c00..c14 (or another fixed grid).
    All subjects contribute one equally weighted player-average curve. The
    within-player covariance is estimated only from between-attempt deviations.
    Single-attempt athletes are conditioned with pooled attempt noise, rather
    than assigned certainty. Repeats within one session are not test-retest data.
    """
    def __init__(self, curve_columns: list[str], n_components: int = 2):
        self.curve_columns = curve_columns
        self.n_components = n_components

    @staticmethod
    def _psd(matrix):
        values, vectors = np.linalg.eigh((matrix + matrix.T) / 2)
        return (vectors * np.maximum(values, 0)) @ vectors.T

    def fit(self, attempts: pd.DataFrame):
        cols = self.curve_columns
        clean = attempts.dropna(subset=cols)
        groups = list(clean.groupby("player_id", sort=True))
        if len(groups) < 4:
            raise ValueError("At least four training players are required")
        means = np.array([g[cols].to_numpy(float).mean(0) for _, g in groups])
        counts = np.array([len(g) for _, g in groups])
        self.mean_ = means.mean(0)
        # Covariance uses distinct attempts, not turns. Same-session attempts can
        # still share coach, instruction and sequence effects.
        within = np.zeros((len(cols), len(cols)))
        dof = 0
        repeated_players = 0
        for _, group in groups:
            values = group[cols].to_numpy(float)
            if len(values) > 1:
                residual = values - values.mean(0)
                within += residual.T @ residual
                dof += len(values) - 1
                repeated_players += 1
        self.repeated_players_ = repeated_players
        self.noise_estimable_ = repeated_players >= 5 and dof >= 5
        self.fallback_ = None if self.noise_estimable_ else "ordinary_pca_noise_unestimable"
        if self.noise_estimable_:
            raw_noise = within / dof
            # Fixed shrinkage tempers unstable high-dimensional cross-time noise.
            self.noise_ = 0.5 * raw_noise + 0.5 * np.diag(np.diag(raw_noise))
        else:
            self.noise_ = np.zeros_like(within)
        observed = np.cov(means, rowvar=False, ddof=1)
        signal = self._psd(observed - self.noise_ * np.mean(1 / counts))
        values, vectors = np.linalg.eigh(signal)
        order = np.argsort(values)[::-1][:self.n_components]
        self.eigenvalues_ = np.maximum(values[order], 1e-10)
        self.components_ = vectors[:, order]
        # Eigenvector signs are deterministic for reproducible figures.
        for k in range(self.components_.shape[1]):
            j = np.argmax(np.abs(self.components_[:, k]))
            self.components_[:, k] *= np.sign(self.components_[j, k]) or 1
        self.signal_ = (self.components_ * self.eigenvalues_) @ self.components_.T
        self.training_player_ids_ = [str(pid) for pid, _ in groups]
        return self

    def transform(self, attempts: pd.DataFrame):
        rows = []
        for pid, group in attempts.groupby("player_id", sort=True):
            valid = group.dropna(subset=self.curve_columns)
            count = len(valid)
            if count == 0:
                continue
            mean = valid[self.curve_columns].to_numpy(float).mean(0)
            noise = self.noise_ / count
            if self.noise_estimable_:
                inverse = np.linalg.pinv(self.signal_ + noise + np.eye(len(mean)) * 1e-8)
                h = self.eigenvalues_[:, None] * self.components_.T
                scores = h @ inverse @ (mean - self.mean_)
                uncertainty = np.diag(self.eigenvalues_) - h @ inverse @ h.T
            else:
                scores = self.components_.T @ (mean - self.mean_)
                uncertainty = np.full((len(scores), len(scores)), np.nan)
            row = {"player_id": pid, "n_attempts": count,
                   "noise_estimable": self.noise_estimable_, "fallback": self.fallback_}
            for k in range(len(scores)):
                row[f"functional_eb_pc{k+1}"] = float(scores[k])
                variance = uncertainty[k, k]
                row[f"functional_eb_se{k+1}"] = float(np.sqrt(max(0, variance))) if np.isfinite(variance) else np.nan
            rows.append(row)
        return pd.DataFrame(rows)


def balanced_repeatability(attempts: pd.DataFrame, value_column: str):
    """ANOVA repeatability using the first two attempts of repeated athletes.

    These repeats share session/instruction/coach effects; reported as same-session
    repeatability, never reliability of a stable trait or NFL outcome validity.
    """
    pairs = []
    for _, group in attempts.groupby("player_id", sort=True):
        sort_columns = [c for c in ["event_id", "attempt"] if c in group]
        sort_columns = sort_columns or ["attempt_id"]
        values = group.sort_values(sort_columns)[value_column].dropna().to_numpy(float)
        if len(values) >= 2:
            pairs.append(values[:2])
    values = np.array(pairs)
    if len(values) < 5:
        return {"n_repeated_players": len(values), "estimable": False}
    player_means = values.mean(axis=1)
    grand_mean = values.mean()
    between = 2 * np.sum((player_means - grand_mean) ** 2) / (len(values) - 1)
    within = np.sum((values - player_means[:, None]) ** 2) / len(values)
    if between + within <= 1e-12:
        return {"n_repeated_players":len(values), "estimable":False, "reason":"Degenerate between/within variance"}
    return {"n_repeated_players": len(values), "estimable": True,
            "icc1_1": float((between - within) / (between + within)),
            "typical_error": float(np.std(values[:, 1] - values[:, 0], ddof=1) / np.sqrt(2)),
            "mean_trial_2_minus_1": float(np.mean(values[:, 1] - values[:, 0]))}
````

### A.3. run_functional.py


````python
%%writefile functional_code/run_functional.py
"""OL mirror curve extraction and nested player-level exploratory comparison.

Usage: python run_functional.py --raw-dir /path/to/data --output-dir /path/to/out
Portable Python 3; requires numpy/pandas/scipy/scikit-learn/matplotlib.
Raw observations are never written to outputs. New results are development
evidence; the original entry already inspected 2025.
"""
from __future__ import annotations
import argparse, hashlib, json,io,os
from pathlib import Path
import numpy as np
import pandas as pd
from scipy.signal import savgol_filter
from scipy.stats import spearmanr
from sklearn.model_selection import StratifiedKFold
from sklearn.decomposition import PCA
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from functional_eb import FunctionalEB, balanced_repeatability

BASE = ["combine_height", "combine_weight", "forty", "ten_yd_split"]
FAMILY = ["normalized_deficit", "incoming_speed", "lateral_amplitude"]
TIME_GRID = np.round(np.linspace(-0.4, 0.6, 11), 3)
CURVE = [f"c{i:02d}" for i in range(len(TIME_GRID))]
SEED = 87007
ALPHAS = [1.0, 10.0, 100.0]


def safe_savefig(figure,path,dpi=160):
    """Render completely in memory, verify PNG, then replace output atomically."""
    from PIL import Image
    buffer=io.BytesIO();figure.savefig(buffer,format='png',dpi=dpi)
    data=buffer.getvalue()
    Image.open(io.BytesIO(data)).verify()
    temporary=path.with_name(path.name+f'.{os.getpid()}.pending')
    temporary.write_bytes(data);temporary.replace(path)


def load_data(raw_dir, years):
    d = Path(raw_dir)
    players = pd.read_csv(d / "players.csv")
    players = players[players.draft_year.isin(years)]
    results = pd.read_csv(d / "combine_results.csv")
    results = results[results.draft_year.isin(years) & results.combine_position.eq("OL")]
    ids = set(results.nfl_id)
    if results.duplicated(["draft_year", "nfl_id"]).any():
        raise ValueError("Duplicate Combine participant key")
    games = pd.read_csv(d / "games.csv")
    games = games[games.season.isin(years) & games.season_type.eq("REG")]
    tracking_parts = []
    for q in pd.read_csv(d / "combine_tracking.csv", chunksize=200000, low_memory=False):
        q = q[q.draft_year.isin(years) & q.nfl_id.isin(ids) & q.entity_type.eq("PLAYER") &
              q.drill_name.str.replace("-", "_", regex=False).eq("PASS_PRO_MIRROR_DRILL")]
        if len(q): tracking_parts.append(q)
    tracking = pd.concat(tracking_parts, ignore_index=True)
    observed_parts = []
    for q in pd.read_csv(d / "player_play.csv", chunksize=100000, low_memory=False):
        q = q[q.nfl_id.isin(ids) & q.game_id.isin(set(games.game_id))]
        q = q.merge(players[["nfl_id", "draft_year"]], on="nfl_id", validate="many_to_one").merge(
            games[["game_id", "season"]], on="game_id", validate="many_to_one")
        q = q[q.season.eq(q.draft_year) & q.pressure_allowed.notna() &
              q.lined_up_position.isin(["C", "G", "T"]) &
              q.team_abbr.eq(q.possession_team) & q.play_nullified_by_penalty.eq("N")]
        labels = q.pressure_allowed.astype(str).str.lower()
        if not set(labels) <= {"true", "false"}: raise ValueError("Unexpected pressure labels")
        q = q.assign(pressure=labels.eq("true").astype(int))
        if len(q): observed_parts.append(q[["draft_year", "nfl_id", "game_id", "play_id", "pressure"]])
    observed = pd.concat(observed_parts, ignore_index=True)
    if observed.duplicated(["game_id", "play_id", "nfl_id"]).any():
        raise ValueError("Duplicate eligible player-play key")
    outcomes = observed.groupby(["draft_year", "nfl_id"], as_index=False).agg(
        n=("pressure", "size"), k=("pressure", "sum"))
    outcomes["rate"] = outcomes.k / outcomes.n
    return results, tracking, outcomes


def extract_curves(tracking, smoothing_window=5, exit_threshold=1.0):
    key_columns = ["draft_year", "event_id", "nfl_id", "attempt"]
    if tracking.duplicated(key_columns + ["time"]).any():
        raise ValueError("Duplicate tracking timestamp")
    if not tracking.x.between(0,120).all() or not tracking.y.between(0,53.5).all():
        raise ValueError("Coordinates outside documented football-field yard bounds")
    attempts, events, audit = [], [], []
    for key, group in tracking.groupby(key_columns, sort=True):
        group = group.sort_values("time")
        sec = (pd.to_datetime(group.time) - pd.to_datetime(group.time.iloc[0])).dt.total_seconds().to_numpy()
        x, y = group.x.to_numpy(float), group.y.to_numpy(float)
        quality = "ok"
        exits = np.flatnonzero((np.abs(x - np.median(x[:min(10, len(x))])) > exit_threshold) & (sec > 0.5))
        stop = int(exits[0]) if len(exits) else len(x)
        if stop < 15 or np.any(np.abs(np.diff(sec[:stop]) - 0.1) > 0.015): quality = "short_or_non_10hz"
        if not (np.isfinite(x).all() and np.isfinite(y).all()): quality = "nonfinite_coordinates"
        attempt_id = "|".join(map(str, key))
        player_id = str(int(key[2]))
        local = []
        if quality == "ok":
            smooth = savgol_filter(y[:stop], smoothing_window, 2)
            velocity = np.gradient(smooth, sec[:stop])
            signs = np.where(np.abs(velocity) > 0.05, np.sign(velocity), 0)
            nz = np.flatnonzero(signs)
            changes = [int(round((i+j)/2)) for i,j in zip(nz[:-1], nz[1:])
                       if signs[i]*signs[j] < 0 and j-i <= 10]
            last = -100
            for index in changes:
                # Do not extrapolate truncated pre/post windows into valid curves.
                if index < 5 or index+6 >= stop or index-last < 5: continue
                pre = float(np.median(velocity[index-5:index-1]))
                post = float(np.median(velocity[index+1:index+5]))
                if abs(pre) < 0.5 or abs(post) < 0.5 or pre*post >= 0: continue
                incoming = abs(pre)
                direction = np.sign(post)
                curve = np.interp(sec[index] + TIME_GRID, sec[:stop], direction*velocity) / incoming
                deficit = float(np.trapezoid(np.maximum(0, incoming - direction*velocity[index:index+6]),
                                             sec[index:index+6]-sec[index]) / incoming)
                row = dict(zip(key_columns, key), player_id=player_id, attempt_id=attempt_id,
                           reversal_time_seconds=float(sec[index]), incoming_speed=incoming,
                           normalized_deficit=deficit, direction=int(direction))
                row.update(dict(zip(CURVE, map(float, curve))))
                local.append(row); events.append(row); last = index
        if len(local) < 2 and quality == "ok": quality = "fewer_than_two_complete_curve_turns"
        audit.append(dict(zip(key_columns, key), quality=quality, frames=len(group),
                          mirror_frames=stop, complete_curve_turns=len(local)))
        if quality != "ok": continue
        event_frame = pd.DataFrame(local)
        row = dict(zip(key_columns, key), player_id=player_id, attempt_id=attempt_id,
                   n_turns=len(local), incoming_speed=float(event_frame.incoming_speed.median()),
                   normalized_deficit=float(event_frame.normalized_deficit.quantile(0.75)),
                   lateral_amplitude=float(np.percentile(y[:stop], 90) - np.percentile(y[:stop], 10)))
        row.update(dict(zip(CURVE, event_frame[CURVE].median().to_numpy(float))))
        attempts.append(row)
    return pd.DataFrame(attempts), pd.DataFrame(events), pd.DataFrame(audit)


````

````python
%%writefile --append functional_code/run_functional.py
def brier(rate, predictions):
    return rate*(1-predictions)**2 + (1-rate)*predictions**2


def ridge(train_x, train_y, test_x, alpha):
    median = np.nanmedian(train_x, axis=0)
    median = np.where(np.isfinite(median), median, 0)
    train_x = np.where(np.isnan(train_x), median, train_x)
    test_x = np.where(np.isnan(test_x), median, test_x)
    mean, sd = train_x.mean(0), train_x.std(0)
    sd = np.where(sd > 1e-12, sd, 1)
    xx, xt = (train_x-mean)/sd, (test_x-mean)/sd
    ymean = train_y.mean()
    coefficient = np.linalg.solve(xx.T @ xx + alpha*np.eye(xx.shape[1]), xx.T @ (train_y-ymean))
    return np.clip(ymean + xt @ coefficient, 0.001, 0.999)


def feature_pair(train, test, attempts, mode):
    columns = BASE + (FAMILY if mode == "scalar_family_complete_windows" else [])
    if mode in ("task_demands", "curve_pca2", "curve_eb2"):
        columns = BASE + FAMILY[1:]
    train_x, test_x = train[columns].to_numpy(float), test[columns].to_numpy(float)
    if mode in ("curve_pca2", "curve_eb2"):
        train_attempts = attempts[attempts.player_id.isin(set(train.player_id))]
        test_attempts = attempts[attempts.player_id.isin(set(test.player_id))]
        if mode == "curve_eb2":
            estimator = FunctionalEB(CURVE, 2).fit(train_attempts)
            ft, fv = estimator.transform(train_attempts), estimator.transform(test_attempts)
            pc = ["functional_eb_pc1", "functional_eb_pc2"]
            tx = train[["player_id"]].merge(ft, on="player_id", how="left")[pc].to_numpy(float)
            vx = test[["player_id"]].merge(fv, on="player_id", how="left")[pc].to_numpy(float)
        else:
            ft = train_attempts.groupby("player_id")[CURVE].mean()
            fv = test_attempts.groupby("player_id")[CURVE].mean()
            tx = ft.reindex(train.player_id).to_numpy(float)
            vx = fv.reindex(test.player_id).to_numpy(float)
            median = np.nanmedian(tx, axis=0)
            tx, vx = np.where(np.isnan(tx), median, tx), np.where(np.isnan(vx), median, vx)
            estimator = PCA(n_components=2, svd_solver="full").fit(tx)
            tx, vx = estimator.transform(tx), estimator.transform(vx)
        train_x, test_x = np.column_stack([train_x, tx]), np.column_stack([test_x, vx])
    return train_x, test_x


def id_folds(frame, records):
    ids=frame.nfl_id.to_numpy(int)
    universe=set(ids);seen=[];folds=[]
    for record in records:
        train_ids=set(map(int,record.get("train_ids",record.get("train_player_ids",[]))))
        test_ids=set(map(int,record.get("test_ids",record.get("test_player_ids",[]))))
        if not train_ids or not test_ids or train_ids & test_ids:
            raise ValueError("Empty or overlapping common player fold")
        if train_ids | test_ids != universe:
            raise ValueError("Common fold universe differs from complete-curve eligible cohort")
        train=np.flatnonzero(np.isin(ids,list(train_ids)));test=np.flatnonzero(np.isin(ids,list(test_ids)))
        folds.append((train,test));seen.extend(ids[test].tolist())
    if sorted(seen) != sorted(ids.tolist()):
        raise ValueError("Every eligible player must be tested exactly once")
    return folds


def nested_evaluate(cohort, attempts, fold_manifest=None):
    cohort = cohort.sort_values(["draft_year", "nfl_id"]).reset_index(drop=True)
    if not cohort.nfl_id.is_unique: raise ValueError("Player-level cohort has repeated identifiers")
    outer_records=None
    if fold_manifest is None:
        outer=list(StratifiedKFold(5,shuffle=True,random_state=SEED).split(cohort,cohort.draft_year))
    else:
        if isinstance(fold_manifest,list): outer_records=fold_manifest
        else:
            outer_records=next((fold_manifest[k] for k in ["outer_folds","folds","outer"] if k in fold_manifest),None)
        if outer_records is None: raise ValueError("Common map needs explicit outer and inner train/test player-ID records")
        outer=id_folds(cohort,outer_records)
        if len(outer)!=5: raise ValueError("Five outer folds required")
    fold_id=np.full(len(cohort),-1,int)
    modes = ["constant", "conventional", "task_demands", "scalar_family_complete_windows", "curve_pca2", "curve_eb2"]
    predictions = {mode: np.full(len(cohort), np.nan) for mode in modes}
    selections = []
    for f,(tr,te) in enumerate(outer):
        fold_id[te]=f
        train,test=cohort.iloc[tr].reset_index(drop=True),cohort.iloc[te].reset_index(drop=True)
        predictions["constant"][te]=train.rate.mean()
        if outer_records is not None:
            inner_records=outer_records[f].get("inner_folds",outer_records[f].get("inner"))
            if inner_records is None: raise ValueError("Immutable inner player folds required for matched comparison")
            inner=id_folds(train,inner_records)
            if len(inner)!=3: raise ValueError("Three inner folds required")
        else:
            inner=list(StratifiedKFold(3,shuffle=True,random_state=SEED+100+f).split(train,train.draft_year))
        for mode in modes[1:]:
            losses = []
            for alpha in ALPHAS:
                pp = np.full(len(train), np.nan)
                for ii, jj in inner:
                    a, b = train.iloc[ii], train.iloc[jj]
                    xx, xt = feature_pair(a, b, attempts, mode)
                    pp[jj] = ridge(xx, a.rate.to_numpy(float), xt, alpha)
                losses.append(float(brier(train.rate.to_numpy(float), pp).mean()))
            best = sorted(zip(losses, ALPHAS), key=lambda z:(z[0], -z[1]))[0][1]
            xx, xt = feature_pair(train, test, attempts, mode)
            predictions[mode][te] = ridge(xx, train.rate.to_numpy(float), xt, best)
            selections.append({"outer_fold":f,"mode":mode,"alpha":best,"inner_brier":losses,
                               "train_ids":train.nfl_id.astype(int).tolist(),"test_ids":test.nfl_id.astype(int).tolist(),
                               "inner_folds":[{"train_ids":train.iloc[ii].nfl_id.astype(int).tolist(),
                                               "test_ids":train.iloc[jj].nfl_id.astype(int).tolist()} for ii,jj in inner]})
    y = cohort.rate.to_numpy(float)
    metrics = {mode:{"equal_player_brier":float(brier(y, pp).mean()),
                     "equal_player_mae":float(np.abs(y-pp).mean()),
                     "opportunity_weighted_brier":float(np.average(brier(y, pp), weights=cohort.n))}
               for mode,pp in predictions.items()}
    return cohort, fold_id, predictions, metrics, selections


def bootstrap_difference(y, reference, alternative, seed=SEED, n_boot=5000):
    differences = brier(y, reference) - brier(y, alternative)
    rng = np.random.default_rng(seed)
    boot = differences[rng.integers(0, len(y), size=(n_boot, len(y)))].mean(1)
    return {"reference_minus_alternative":float(differences.mean()),
            "conditional_fixed_predictions_player_bootstrap_95":np.quantile(boot,[0.025,0.975]).tolist(),
            "limitations":"Does not refit folds, account for shared games or correct model search; exploratory diagnostic."}


def repeatability_report(attempts, value_column, n_boot=1000):
    report=balanced_repeatability(attempts,value_column)
    if not report.get("estimable"): return report
    pairs=[]
    for _,group in attempts.groupby("player_id",sort=True):
        sort_columns=[c for c in ["event_id","attempt"] if c in group] or ["attempt_id"]
        values=group.sort_values(sort_columns)[value_column].dropna().to_numpy(float)
        if len(values)>=2: pairs.append(values[:2])
    values=np.array(pairs);rng=np.random.default_rng(SEED)
    draws=values[rng.integers(0,len(values),size=(n_boot,len(values)))]
    means=draws.mean(2);grand=draws.mean((1,2))
    between=2*((means-grand[:,None])**2).sum(1)/(len(values)-1)
    within=((draws-means[:,:,None])**2).sum((1,2))/len(values)
    valid=between+within>1e-12
    icc=(between[valid]-within[valid])/(between[valid]+within[valid])
    te=np.std(draws[:,:,1]-draws[:,:,0],axis=1,ddof=1)/np.sqrt(2)
    report.update({"paired_player_bootstrap_icc_95":np.quantile(icc,[.025,.975]).tolist(),
                   "paired_player_bootstrap_typical_error_95":np.quantile(te,[.025,.975]).tolist(),
                   "bootstrap_replicates":n_boot,
                   "interpretation":"Within-session attempt consistency; shared instructions, coaches, sequence, and tracking error are not disentangled."})
    return report


````

````python
%%writefile --append functional_code/run_functional.py
def make_figures(attempts, events, cohort, out):
    usable = events[events.attempt_id.isin(set(attempts.attempt_id))]
    # Pick most repeated athlete, breaking ties by numeric ID rather than NFL outcome.
    counts = attempts.groupby("player_id").size()
    example = sorted(counts.index, key=lambda p:(-counts[p], int(p)))[0]
    figure, axes = plt.subplots(1,2, figsize=(11,4.2), constrained_layout=True)
    curves = attempts[attempts.player_id.eq(example)]
    selected_events=usable[usable.player_id.eq(example)]
    for _, row in selected_events.iterrows():
        axes[0].plot(TIME_GRID,row[CURVE].to_numpy(float),color="#9aabba",alpha=.65,lw=1)
    for _,row in curves.iterrows():
        axes[0].plot(TIME_GRID,row[CURVE].to_numpy(float),color="#126d9a",lw=2.5,
                     label=f"Attempt {row['attempt']} median; {int(row['n_turns'])} turns")
    axes[0].axvline(0,color="black",lw=0.8); axes[0].axhline(0,color="gray",lw=0.8)
    axes[0].set(xlabel="Seconds from kinematic reversal", ylabel="Outgoing-signed velocity / entry speed",
                title=f"One OL: individual turns and attempt median\nNFL ID {example}; selection ignores NFL outcome")
    axes[0].legend(fontsize=8)
    pairs = []
    for pid, group in attempts.groupby("player_id"):
        if len(group) >= 2:
            values = group.sort_values("attempt_id").normalized_deficit.iloc[:2].to_numpy()
            pairs.append(values)
    if pairs:
        pairs=np.array(pairs); axes[1].scatter(pairs[:,0],pairs[:,1],alpha=0.7)
        lim=[float(pairs.min()),float(pairs.max())]; axes[1].plot(lim,lim,color="gray",ls="--")
        axes[1].set(xlabel="First attempt recovery cost (seconds)",ylabel="Second attempt recovery cost (seconds)",
                    title="Same-session repeatability, not stable talent")
    else:
        points=axes[1].scatter(selected_events.reversal_time_seconds,selected_events.normalized_deficit,
                               c=selected_events.incoming_speed,cmap="viridis",s=80,edgecolor="black",lw=.5)
        figure.colorbar(points,ax=axes[1],label="Entry speed (yards/second)")
        axes[1].set(xlabel="Reversal time within attempt (seconds)",ylabel="Turn recovery cost (seconds)",
                    title="Turn-to-turn task variation\nOnly one attempt: reliability is unidentifiable")
    safe_savefig(figure,out/"functional_measurement.png",dpi=160);plt.close(figure)
    quantiles=np.quantile(attempts[CURVE].to_numpy(float),[.1,.5,.9],axis=0)
    figure,ax=plt.subplots(figsize=(7,4),constrained_layout=True)
    ax.fill_between(TIME_GRID,quantiles[0],quantiles[2],alpha=.25,label="Attempt 10th–90th percentiles")
    ax.plot(TIME_GRID,quantiles[1],label="Attempt median")
    ax.axvline(0,color="black",lw=.8);ax.axhline(0,color="gray",lw=.8)
    ax.set(xlabel="Seconds from kinematic reversal",ylabel="Outgoing-signed velocity / entry speed",
           title="OL mirror movement curves: braking and recovery")
    ax.legend();safe_savefig(figure,out/"population_curve.png",dpi=160);plt.close(figure)


def main():
    parser=argparse.ArgumentParser()
    parser.add_argument("--raw-dir",type=Path,required=True)
    parser.add_argument("--output-dir",type=Path,default=Path(__file__).parent/"round7")
    parser.add_argument("--years",nargs="+",type=int,default=[2023,2024])
    parser.add_argument("--fold-manifest","--fold-map",type=Path)
    args=parser.parse_args();args.output_dir.mkdir(parents=True,exist_ok=True)
    results,tracking,outcomes=load_data(args.raw_dir,args.years)
    attempts,events,audit=extract_curves(tracking)
    scalar=attempts.groupby(["draft_year","nfl_id"],as_index=False)[FAMILY].median()
    cohort=results.merge(scalar,on=["draft_year","nfl_id"],validate="one_to_one").merge(
        outcomes,on=["draft_year","nfl_id"],validate="one_to_one")
    cohort=cohort[cohort.n.ge(50)].copy();cohort["player_id"]=cohort.nfl_id.astype(int).astype(str)
    manifest=json.loads(args.fold_manifest.read_text()) if args.fold_manifest else None
    cohort,folds,predictions,metrics,selections=nested_evaluate(cohort,attempts,manifest)
    noise=FunctionalEB(CURVE,2).fit(attempts[attempts.player_id.isin(set(cohort.player_id))])
    sensitivity={}
    for window in [7,9]:
        aa,_,qq=extract_curves(tracking,smoothing_window=window)
        base=attempts.groupby("player_id").normalized_deficit.median()
        alt=aa.groupby("player_id").normalized_deficit.median()
        match=pd.concat([base,alt],axis=1,join="inner").dropna()
        sensitivity[str(window)]={"players_matched":len(match),"rank_correlation":float(spearmanr(match.iloc[:,0],match.iloc[:,1]).statistic),
                                  "median_absolute_change_seconds":float(np.median(np.abs(match.iloc[:,0]-match.iloc[:,1]))),
                                  "attempt_quality_counts":qq.quality.value_counts().to_dict()}
    output={"status":"exploratory_development_new_methods;2025_previously_observed",
            "years":args.years,"eligible_players":len(cohort),"charted_opportunities":int(cohort.n.sum()),
            "charted_pressures":int(cohort.k.sum()),"tracking_frames":len(tracking),
            "attempts":len(attempts),"turn_events":len(events),"quality_counts":audit.quality.value_counts().to_dict(),
            "metrics":metrics,"alpha_selections":selections,
            "nested_scheme":{"outer":5,"inner":3,"seed":SEED,"stratified_by":"draft_year","grouped_by":"nfl_id",
                             "fold_assignment_source":"immutable_root_player_ID_map" if args.fold_manifest else "candidate_local_generated_not_yet_matched",
                             "fold_map_sha256":hashlib.sha256(args.fold_manifest.read_bytes()).hexdigest() if args.fold_manifest else None},
            "same_session_repeatability":repeatability_report(attempts,"normalized_deficit"),
            "measurement_noise_estimable":noise.noise_estimable_,"training_repeated_players":noise.repeated_players_,
            "curve_eigenvalues":noise.eigenvalues_.tolist(),"detector_sensitivity":sensitivity,
            "comparisons":{m:bootstrap_difference(cohort.rate.to_numpy(float),predictions["conventional"],predictions[m])
                           for m in ["scalar_family_complete_windows","curve_pca2","curve_eb2"]},
            "increment_over_task_demands":{m:bootstrap_difference(cohort.rate.to_numpy(float),predictions["task_demands"],predictions[m])
                                           for m in ["curve_pca2","curve_eb2"]},
            "target":"Explicit non-null charted rookie REG pressures; non-nullified possession C/G/T rows; minimum50; equal-player assessment.",
            "limitations":["Turn events from one attempt are dependent; one median curve per attempt.",
                           "Scalar family follows the prior recipe but is measured on complete -.4..+.6s turn windows; it is not an exact rerun of the original scalar detector.",
                           "Repeated attempts share a session and unknown coach/sequence demands.",
                           "Player folds do not remove shared game dependence or draft-cohort shift.",
                           "No stimulus timing, foot contacts or force plates; kinematics cannot identify reaction time or braking impulse.",
                           "Fixed normalization windows and EB assumptions need sensitivity; this is not exact PACE.",
                           "No confirmatory claim after prior cohort use; bootstrap is conditional on fitted OOF predictions."]}
    (args.output_dir/"ROUND_RESULT.json").write_text(json.dumps(output,indent=2))
    prediction_rows=[]
    for i,row in cohort.iterrows():
        prediction_rows.append({"player_id":row.player_id,"draft_year":int(row.draft_year),"outer_fold":int(folds[i]),
                                "n":int(row.n),"k":int(row.k),"rate":float(row.rate),
                                **{m:float(pp[i]) for m,pp in predictions.items()}})
    (args.output_dir/"OOF_PREDICTIONS.json").write_text(json.dumps(prediction_rows,indent=2))
    (args.output_dir/"EXTRACTION_AUDIT.json").write_text(audit.to_json(orient="records",indent=2))
    make_figures(attempts,events,cohort,args.output_dir)
    print(json.dumps(output,indent=2))


if __name__=="__main__": main()
````

### A.4. legacy_scalar.py


````python
%%writefile functional_code/legacy_scalar.py
"""Exact original scalar extraction functions, copied without original-file mutation.
Source: agents/ol/round5/ol_candidate.py; no outcomes or experiment execution copied.
"""
import numpy as np
import pandas as pd
from scipy.signal import savgol_filter
from scipy.integrate import trapezoid

def extract(t, window=5, exit_threshold=1.0):
    rows=[]; events=[]; examples={}
    for key,g in t.groupby(['draft_year','event_id','nfl_id','attempt'],sort=True):
        g=g.sort_values('time').copy()
        tm=pd.to_datetime(g.time); sec=(tm-tm.iloc[0]).dt.total_seconds().to_numpy()
        x=g.x.to_numpy(); y=g.y.to_numpy()
        start=np.median(x[:min(10,len(x))])
        exits=np.flatnonzero((np.abs(x-start)>exit_threshold)&(sec>0.5))
        stop=int(exits[0]) if len(exits) else len(x)
        quality='ok'
        if stop<15 or np.any(np.abs(np.diff(sec[:stop])-.1)>.015): quality='short_or_non_10hz'
        if np.isnan(x).any() or np.isnan(y).any(): quality='missing_coordinates'
        rev=[]; deficits=[]
        if quality=='ok':
            sy=savgol_filter(y[:stop],window,2)
            vel=np.gradient(sy,sec[:stop])
            # Zero-speed frames are real at turns: bridge near-zero intervals,
            # rather than requiring adjacent nonzero signs (which misses minima).
            signs=np.where(np.abs(vel)>.05,np.sign(vel),0)
            nz=np.flatnonzero(signs)
            changes=[int(round((i+j)/2)) for i,j in zip(nz[:-1],nz[1:])
                     if signs[i]*signs[j]<0 and j-i<=10]
            last=-100
            for i in changes:
                if i<5 or i+5>=stop or i-last<5: continue
                pre=np.median(vel[i-5:i-1]); post=np.median(vel[i+1:i+5])
                if abs(pre)<.5 or abs(post)<.5 or pre*post>=0: continue
                new=np.sign(post)*vel[i:i+6]
                deficit=float(trapezoid(np.maximum(0,abs(pre)-new),sec[i:i+6]-sec[i]))
                rev.append(i); deficits.append(deficit); last=i
                events.append({'draft_year':key[0],'event_id':key[1],'nfl_id':key[2],'attempt':key[3],
                               'reversal_time_seconds':float(sec[i]),'incoming_speed_yards_s':float(abs(pre)),
                               'recovery_deficit_yards':deficit})
        if len(deficits)<2 and quality=='ok': quality='fewer_than_two_usable_reversals'
        rows.append({'draft_year':key[0],'event_id':key[1],'nfl_id':key[2],'attempt':key[3],
                     'frames':len(g),'mirror_frames':stop,'usable_reversals':len(deficits),
                     'quality':quality,'recovery_deficit':float(np.percentile(deficits,75)) if quality=='ok' else np.nan,
                     'x_start':float(start),'x_range':float(np.ptp(x)),'y_range':float(np.ptp(y))})
        examples[key]={'sec':sec,'x':x,'y':y,'stop':stop,'rev':rev,'quality':quality}
    return pd.DataFrame(rows),pd.DataFrame(events),examples

def extract_mirror(t):
    keys=['draft_year','event_id','nfl_id','attempt','time']
    if t.duplicated(keys).any():raise ValueError('Duplicate tracking timestamp key; extraction refused.')
    if t.empty:raise ValueError('No mirror trajectories; metric is undefined, not zero.')
    if not t.x.between(0,120).all() or not t.y.between(0,53.5).all():
        raise ValueError('Coordinate bounds incompatible with documented field-yard units.')
    return extract(t)

def build_features(t):
    attempts,events,examples=extract_mirror(t)
    keys=['draft_year','event_id','nfl_id','attempt']
    good=attempts[attempts.quality.eq('ok')]
    if good.empty or events.empty:raise ValueError('No usable reversals; metric is undefined.')
    e=events.merge(good[keys],on=keys,validate='many_to_one')
    e['normalized_deficit']=e.recovery_deficit_yards/e.incoming_speed_yards_s
    a=e.groupby(keys,as_index=False).agg(normalized_deficit=('normalized_deficit',lambda z:np.percentile(z,75)),
        incoming_speed=('incoming_speed_yards_s','median'))
    task=[]
    for _,row in good.iterrows():
        key=(int(row.draft_year),row.event_id,int(row.nfl_id),int(row.attempt));ex=examples[key]
        yy=ex['y'][:ex['stop']]
        task.append(dict(zip(keys,key),lateral_amplitude=float(np.percentile(yy,90)-np.percentile(yy,10)),
                         phase_seconds=float(ex['sec'][ex['stop']-1])))
    z=good.merge(a,on=keys,validate='one_to_one').merge(pd.DataFrame(task),on=keys,validate='one_to_one')
    features=z.groupby(['draft_year','nfl_id'],as_index=False).agg(
        normalized_deficit=('normalized_deficit','median'),incoming_speed=('incoming_speed','median'),
        lateral_amplitude=('lateral_amplitude','median'),raw_deficit=('recovery_deficit','median'),
        usable_reversals=('usable_reversals','sum'),usable_attempts=('attempt','size'),phase_seconds=('phase_seconds','median'))
    return features,attempts,events,examples
````

### A.5. run_canonical38.py


````python
%%writefile functional_code/run_canonical38.py
"""R8 supplement: exact canonical OL population with explicit missing-curve gate.

Reconstructs the original scalar recipe from preserved functions, then compares
on the immutable38-player common folds. Missing curve features remain missing
until fold-training imputation; missing NFL pressure labels are never zeros.
"""
from pathlib import Path
import argparse,json,hashlib
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
from run_functional import (load_data,extract_curves,BASE,FAMILY,CURVE,SEED,ALPHAS,
                            id_folds,ridge,brier,bootstrap_difference,make_figures,safe_savefig)
from legacy_scalar import build_features

MODES=['constant','conventional','original_scalar_family','missingness_gate','task_demands_missing','curve2_missing']


def feature_pair38(train,test,attempts,mode):
    columns=BASE[:]
    if mode=='original_scalar_family':columns+=['original_'+x for x in FAMILY]
    if mode=='missingness_gate':columns+=['curve_missing']
    if mode in ['task_demands_missing','curve2_missing']:columns+=FAMILY[1:]+['curve_missing']
    xx,xt=train[columns].to_numpy(float),test[columns].to_numpy(float)
    if mode=='curve2_missing':
        means=attempts.groupby('player_id')[CURVE].mean()
        curves=means.reindex(train.player_id).to_numpy(float)
        held=means.reindex(test.player_id).to_numpy(float)
        med=np.nanmedian(curves,axis=0);med=np.where(np.isfinite(med),med,0)
        curves,held=np.where(np.isnan(curves),med,curves),np.where(np.isnan(held),med,held)
        pca=PCA(n_components=2,svd_solver='full').fit(curves)
        xx=np.column_stack([xx,pca.transform(curves)]);xt=np.column_stack([xt,pca.transform(held)])
    return xx,xt


def canonical_nested(cohort,attempts,foldmap):
    cohort=cohort.sort_values(['draft_year','nfl_id']).reset_index(drop=True)
    outer=id_folds(cohort,foldmap['outer_folds'])
    prediction={m:np.full(len(cohort),np.nan) for m in MODES};records=[];fold_id=np.full(len(cohort),-1,int)
    for fold,(ii,jj) in enumerate(outer):
        train,test=cohort.iloc[ii].reset_index(drop=True),cohort.iloc[jj].reset_index(drop=True)
        inner=id_folds(train,foldmap['outer_folds'][fold]['inner_folds'])
        fold_id[jj]=fold;prediction['constant'][jj]=train.rate.mean()
        record={'outer_fold':fold,'train_ids':train.nfl_id.astype(int).tolist(),'test_ids':test.nfl_id.astype(int).tolist(),
                'inner_folds':foldmap['outer_folds'][fold]['inner_folds'],'selection':{}}
        for mode in MODES[1:]:
            losses=[]
            for alpha in ALPHAS:
                pp=np.full(len(train),np.nan)
                for aa,bb in inner:
                    a,b=train.iloc[aa],train.iloc[bb]
                    xx,xt=feature_pair38(a,b,attempts,mode)
                    pp[bb]=ridge(xx,a.rate.to_numpy(float),xt,alpha)
                losses.append(float(brier(train.rate.to_numpy(float),pp).mean()))
            alpha=sorted(zip(losses,ALPHAS),key=lambda z:(z[0],-z[1]))[0][1]
            xx,xt=feature_pair38(train,test,attempts,mode)
            prediction[mode][jj]=ridge(xx,train.rate.to_numpy(float),xt,alpha)
            record['selection'][mode]={'alpha':alpha,'inner_brier':losses}
        records.append(record)
    y=cohort.rate.to_numpy(float)
    metrics={m:{'equal_player_brier':float(brier(y,pp).mean()),'equal_player_mae':float(np.abs(y-pp).mean()),
                'opportunity_weighted_brier':float(np.average(brier(y,pp),weights=cohort.n))} for m,pp in prediction.items()}
    return cohort,fold_id,prediction,metrics,records


def main():
    parser=argparse.ArgumentParser()
    parser.add_argument('--raw-dir',type=Path,required=True)
    parser.add_argument('--fold-map',type=Path,required=True)
    parser.add_argument('--output-dir',type=Path,required=True)
    args=parser.parse_args();out=args.output_dir;out.mkdir(parents=True,exist_ok=True)
    results,tracking,outcomes=load_data(args.raw_dir,[2023,2024])
    legacy,_,_,_=build_features(tracking)
    legacy=legacy[['draft_year','nfl_id']+FAMILY].rename(columns={c:'original_'+c for c in FAMILY})
    attempts,events,audit=extract_curves(tracking)
    new=attempts.groupby(['draft_year','nfl_id'],as_index=False)[FAMILY].median()
    cohort=results.merge(legacy,on=['draft_year','nfl_id'],validate='one_to_one').merge(
        outcomes,on=['draft_year','nfl_id'],validate='one_to_one')
    cohort=cohort[cohort.n.ge(50)].merge(new,on=['draft_year','nfl_id'],how='left',validate='one_to_one')
    cohort['player_id']=cohort.nfl_id.astype(int).astype(str)
    cohort['curve_missing']=cohort.normalized_deficit.isna().astype(float)
    foldmap=json.loads(args.fold_map.read_text())
    cc,fold_id,prediction,metrics,records=canonical_nested(cohort,attempts,foldmap)
    assert len(cc)==38
    result={'status':'executed_exploratory_R8_canonical38_supplement','years':[2023,2024],
            'eligible_players':len(cc),'charted_observations':int(cc.n.sum()),'pressures':int(cc.k.sum()),
            'curve_available_players':int(cc.curve_missing.eq(0).sum()),'curve_missing_ids':cc.loc[cc.curve_missing.eq(1),'nfl_id'].astype(int).tolist(),
            'fold_map_sha256':hashlib.sha256(args.fold_map.read_bytes()).hexdigest(),
            'metrics':metrics,'comparisons_vs_conventional':{m:bootstrap_difference(cc.rate.to_numpy(float),prediction['conventional'],prediction[m]) for m in MODES[2:]},
            'curve_increment_over_task_and_missing_gate':bootstrap_difference(cc.rate.to_numpy(float),prediction['task_demands_missing'],prediction['curve2_missing']),
            'same_session_repeatability':'Not estimable: every official player has only one mirror attempt; no EB forecast advertised.',
            'missing_policy':'One missing curve retains player in validation: explicit availabilityindicator; imputation and PCA fit in inner/outer training only.',
            'interpretation':'Development diagnostic only; original2025 negative result preserved and no new holdout claim.',
            'original_scalar_recipe':'Exact prior extraction functions retained in legacy_scalar.py, without original source mutation.',
            'limits':['Shared game/draft-cohort dependence remains.','Fixed-prediction bootstrap omits full refits, research selection and detector uncertainty.',
                      'One missingcurve example cannot train a generalizable availability effect.','Normalized curves reveal execution/task variation, not reaction time, forces or causal ability.']}
    (out/'CANONICAL38_RESULT.json').write_text(json.dumps(result,indent=2))
    (out/'FOLD_RECORDS.json').write_text(json.dumps(records,indent=2))
    rows=[]
    for i,row in cc.iterrows():
        rows.append({'player_id':row.player_id,'draft_year':int(row.draft_year),'outer_fold':int(fold_id[i]),
                     'n':int(row.n),'k':int(row.k),'rate':float(row.rate),'curve_missing':int(row.curve_missing),
                     **{m:float(pp[i]) for m,pp in prediction.items()}})
    (out/'OOF_PREDICTIONS.json').write_text(json.dumps(rows,indent=2))
    make_figures(attempts,events,cc,out)
    fig,ax=plt.subplots(figsize=(7.5,4.3),constrained_layout=True)
    vals=[metrics[m]['equal_player_brier'] for m in MODES]
    ax.scatter(vals,np.arange(len(MODES)),color='#126d9a',s=65)
    ax.set_yticks(np.arange(len(MODES)),['Constant','Conventional','Original scalar family','Missing-curve indicator','Task demands +indicator','Two curve components +controls'])
    ax.set(xlabel='Nested equal-player Brier loss (lower is better)',title='All38rookie OL: missing curves stay in evaluation')
    ax.invert_yaxis();safe_savefig(fig,out/'canonical38_baselines.png',dpi=170);plt.close(fig)
    print(json.dumps(result,indent=2))


if __name__=='__main__':main()
````

### A.6. robustness.py


````python
%%writefile functional_code/robustness.py
"""Executed R8: fixed37-player universe; detector sensitivity and cohort transfer.

Variants never change the target population. If a detector loses a player's
features, that player remains in validation with fold-training median imputation.
This is exploratory;2025 is historically observed and excluded from this analysis.
"""
from pathlib import Path
import argparse,json,hashlib
import numpy as np
import pandas as pd
from sklearn.model_selection import KFold
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from run_functional import (load_data,extract_curves,BASE,FAMILY,CURVE,SEED,ALPHAS,
                            nested_evaluate,feature_pair,ridge,brier,bootstrap_difference,make_figures,safe_savefig)


def cross_cohort(cohort,attempts):
    modes=["constant","conventional","task_demands","scalar_family_complete_windows","curve_pca2"]
    output=[]
    for year in sorted(cohort.draft_year.unique()):
        train=cohort[cohort.draft_year.ne(year)].sort_values("nfl_id").reset_index(drop=True)
        test=cohort[cohort.draft_year.eq(year)].sort_values("nfl_id").reset_index(drop=True)
        inner=list(KFold(3,shuffle=True,random_state=SEED+int(year)).split(train))
        metric={"constant":{"brier":float(brier(test.rate.to_numpy(float),np.full(len(test),train.rate.mean())).mean())}}
        for mode in modes[1:]:
            scores=[]
            for alpha in ALPHAS:
                prediction=np.full(len(train),np.nan)
                for ii,jj in inner:
                    a,b=train.iloc[ii],train.iloc[jj]
                    xx,xt=feature_pair(a,b,attempts,mode)
                    prediction[jj]=ridge(xx,a.rate.to_numpy(float),xt,alpha)
                scores.append(float(brier(train.rate.to_numpy(float),prediction).mean()))
            alpha=sorted(zip(scores,ALPHAS),key=lambda z:(z[0],-z[1]))[0][1]
            xx,xt=feature_pair(train,test,attempts,mode)
            prediction=ridge(xx,train.rate.to_numpy(float),xt,alpha)
            metric[mode]={"brier":float(brier(test.rate.to_numpy(float),prediction).mean()),"alpha":alpha}
        output.append({"train_years":sorted(map(int,train.draft_year.unique())),"test_year":int(year),
                       "train_players":len(train),"test_players":len(test),"metrics":metric,
                       "interpretation":"Retrospective cross-cohort diagnostic; backward2024->2023 direction is not a forecast."})
    return output


def main():
    parser=argparse.ArgumentParser()
    parser.add_argument("--raw-dir",type=Path,required=True)
    parser.add_argument("--fold-map",type=Path,required=True)
    parser.add_argument("--output-dir",type=Path,required=True)
    args=parser.parse_args();out=args.output_dir;out.mkdir(parents=True,exist_ok=True)
    foldmap=json.loads(args.fold_map.read_text())
    ids=set(map(int,foldmap['outer_folds'][0]['train_ids']+foldmap['outer_folds'][0]['test_ids']))
    results,tracking,outcomes=load_data(args.raw_dir,[2023,2024])
    base=results.merge(outcomes,on=['draft_year','nfl_id'],validate='one_to_one')
    base=base[base.nfl_id.isin(ids)].copy();base['player_id']=base.nfl_id.astype(int).astype(str)
    assert len(base)==len(ids) and base.n.ge(50).all()
    records=[];primary=None
    for window in [5,7,9]:
        for crop in [.75,1.0,1.25]:
            attempts,events,quality=extract_curves(tracking,smoothing_window=window,exit_threshold=crop)
            scalar=attempts.groupby(['draft_year','nfl_id'],as_index=False)[FAMILY].median()
            cohort=base.merge(scalar,on=['draft_year','nfl_id'],how='left',validate='one_to_one')
            cc,folds,predictions,metrics,selections=nested_evaluate(cohort,attempts,foldmap)
            record={'smoothing_window_frames':window,'forward_crop_yards':crop,'fixed_evaluation_players':len(cc),
                    'players_with_usable_curve':int(cc.normalized_deficit.notna().sum()),
                    'feature_missing_players':cc.loc[cc.normalized_deficit.isna(),'nfl_id'].astype(int).tolist(),
                    'usable_attempts_all_OL':len(attempts),'quality_counts':quality.quality.value_counts().to_dict(),
                    'metrics':metrics,'curve_vs_conventional':bootstrap_difference(cc.rate.to_numpy(float),predictions['conventional'],predictions['curve_pca2']),
                    'curve_vs_task_demands':bootstrap_difference(cc.rate.to_numpy(float),predictions['task_demands'],predictions['curve_pca2']),
                    'prediction_imputation':'Missing detector features retained; all imputation fit inside inner/outer training folds'}
            records.append(record)
            print(json.dumps({k:record[k] for k in ['smoothing_window_frames','forward_crop_yards','players_with_usable_curve','curve_vs_conventional']}),flush=True)
            if window==5 and crop==1.0:
                primary=(cc,attempts,events,metrics,predictions)
    cc,attempts,events,metrics,predictions=primary
    make_figures(attempts,events,cc,out)
    transport=cross_cohort(cc,attempts)
    summary={'status':'executed_exploratory_detector_and_cross_cohort_audit','years':[2023,2024],
             'evaluation_players':len(cc),'charted_observations':int(cc.n.sum()),'pressures':int(cc.k.sum()),
             'fold_map_sha256':hashlib.sha256(args.fold_map.read_bytes()).hexdigest(),
             'variants':records,'cross_cohort':transport,
             'reliability':'Zero repeated attempts perplayer; between-attempt ICC/noise unidentifiable; EB equals ordinaryPCA fallback',
             'claim':'No validated forecast gain, stable-trait reliability or causal talent grade is established.'}
    (out/'ROUND8_RESULT.json').write_text(json.dumps(summary,indent=2))
    fig,axes=plt.subplots(1,2,figsize=(11,4.4),constrained_layout=True)
    gains=np.array([record['curve_vs_conventional']['reference_minus_alternative'] for record in records])
    labels=[f"{r['smoothing_window_frames']}frames /{r['forward_crop_yards']}yd" for r in records]
    axes[0].barh(range(len(records)),gains,color=np.where(gains>=0,'#17846b','#a54837'))
    axes[0].axvline(0,color='black',lw=.8);axes[0].set_yticks(range(len(records)),labels)
    axes[0].set(xlabel='Conventional Brier minus curve Brier (higher favors curves)',title='Same37players across detector variants')
    order=['constant','conventional','task_demands','scalar_family_complete_windows','curve_pca2']
    vals=[metrics[m]['equal_player_brier'] for m in order]
    axes[1].scatter(vals,range(len(order)),color='#126d9a',s=60)
    axes[1].set_yticks(range(len(order)),['Constant','Conventional','+Task demands','+Scalar family','+Curve2components'])
    axes[1].set(xlabel='Nested equal-player Brier (lower is better)',title='Exploratory prediction: complexity does not win')
    safe_savefig(fig,out/'robustness_and_baselines.png',dpi=160);plt.close(fig)
    print(json.dumps({'cross_cohort':transport,'curve_gain_range':[float(gains.min()),float(gains.max())]},indent=2),flush=True)


if __name__=='__main__':main()
````

### A.7. build_scouting_cards.py


````python
%%writefile functional_code/build_scouting_cards.py
"""Inspectable turn timestamps, execution intensity and explicit measurement gate.

The selection and cards use Combine tracking only. No NFL pressure label enters
a card, ranking or example choice. Cards are film-review aids, not talent grades.
"""
from pathlib import Path
import argparse,json
import numpy as np
import pandas as pd
from scipy.signal import savgol_filter
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
from run_functional import extract_curves,CURVE,TIME_GRID,safe_savefig


def draw_film_card(tracking,card,output):
    """XY geometry and exact half-second recovery-area construction."""
    turn=card['turns'][0]  # Earliest retained turn; no NFL outcome used.
    group=tracking[tracking.nfl_id.eq(card['nfl_id']) & tracking.event_id.astype(str).eq(turn['event_id']) &
                   tracking.attempt.eq(turn['attempt'])].sort_values('time')
    sec=(pd.to_datetime(group.time)-pd.to_datetime(group.time.iloc[0])).dt.total_seconds().to_numpy()
    x,y=group.x.to_numpy(float),group.y.to_numpy(float)
    origin_x=np.median(x[:min(10,len(x))]);origin_y=y[0]
    exits=np.flatnonzero((np.abs(x-origin_x)>1)&(sec>.5));stop=int(exits[0]) if len(exits) else len(x)
    smooth=savgol_filter(y[:stop],5,2);velocity=np.gradient(smooth,sec[:stop])
    index=int(np.argmin(np.abs(sec-turn['reversal_time_from_attempt_start_seconds'])))
    direction=turn['outgoing_y_direction'];incoming=turn['incoming_lateral_speed_yards_per_second']
    selected=slice(max(0,index-6),min(stop,index+9));relative=sec[selected]-sec[index]
    oriented=direction*velocity[selected]
    window_time=sec[index:index+6]-sec[index];outgoing=direction*velocity[index:index+6]
    shortfall=np.maximum(0,incoming-outgoing)
    area=float(np.trapezoid(shortfall,window_time));cost=area/incoming
    assert np.isclose(cost,turn['normalized_half_second_recovery_cost_seconds'],atol=1e-12)
    fig,axes=plt.subplots(1,2,figsize=(11.2,5),constrained_layout=True)
    ax=axes[0]
    ax.plot(x[:stop]-origin_x,y[:stop]-origin_y,color='#126d9a',lw=2,label='Mirror phase')
    display_end=min(len(x),stop+12)
    if stop<len(x):
        ax.plot(x[stop-1:display_end]-origin_x,y[stop-1:display_end]-origin_y,color='#c86b2d',lw=1.8,ls='--',label='Forward run: excluded')
        ax.scatter(x[stop]-origin_x,y[stop]-origin_y,s=65,marker='x',color='#c86b2d',zorder=5)
        j=min(stop+5,len(x)-1)
        ax.annotate('Forward run',xy=(x[j]-origin_x,y[j]-origin_y),xytext=(8,15),textcoords='offset points',
                    arrowprops={'arrowstyle':'->','color':'#c86b2d'},fontsize=9,color='#8b4216')
        boundary=np.sign(x[stop]-origin_x)
        ax.axvline(boundary,color='#c86b2d',ls=':',lw=1)
    ax.scatter(x[0]-origin_x,y[0]-origin_y,color='#17846b',s=75,zorder=6)
    ax.annotate('Start',xy=(x[0]-origin_x,y[0]-origin_y),xytext=(-30,-15),textcoords='offset points',
                arrowprops={'arrowstyle':'->'},fontsize=9)
    ax.scatter(x[index]-origin_x,y[index]-origin_y,color='#a54837',marker='s',s=65,zorder=7)
    ax.annotate(f"Selected reversal: {sec[index]:.1f}s",xy=(x[index]-origin_x,y[index]-origin_y),
                xytext=(18,12),textcoords='offset points',arrowprops={'arrowstyle':'->','color':'#a54837'},fontsize=9,color='#a54837')
    ax.set_aspect('equal',adjustable='datalim');ax.grid(alpha=.2)
    ax.set(xlabel='Forward displacement from initial median (yards)',ylabel='Lateral displacement from start (yards)',
           title='Tracking geometry: equal yard scales')
    ax.legend(fontsize=8,loc='lower right')
    ax=axes[1]
    ax.plot(relative,oriented,color='#126d9a',lw=2.5,label='Outgoing-signed lateral velocity')
    ax.axhline(incoming,color='#7a5099',ls='--',lw=1.5,label=f'Entry speed: {incoming:.2f}yd/s')
    ax.axhline(0,color='gray',lw=.8);ax.axvline(0,color='#a54837',lw=1)
    ax.fill_between(window_time,incoming-shortfall,incoming,color='#d6a857',alpha=.65,
                    label='Sampled positive shortfall, 0–0.5s')
    ax.axvspan(0,.5,color='#d6a857',alpha=.1)
    ax.text(.03,.94,f'Area {area:.3f}yd ÷ entry speed {incoming:.2f}yd/s\n= recovery cost {cost:.3f}s',
            transform=ax.transAxes,va='top',fontsize=10,bbox={'facecolor':'white','edgecolor':'#dddddd','alpha':.95})
    ax.set(xlabel='Seconds from sampled kinematic reversal',ylabel='Outgoing-signed lateral velocity (yards/second)',
           title='Half-second deficit: movement cost, not reaction time')
    ax.legend(fontsize=8,loc='lower right');ax.grid(alpha=.2)
    fig.suptitle(f"{card['display_name']} · {card['draft_year']} OL mirror · attempt {turn['attempt']}\n"
                 'Example selected by lowest NFL ID, then earliest usable turn; NFL outcomes ignored',fontsize=13)
    safe_savefig(fig,output/'film_review_card.png',dpi=180);plt.close(fig)
    (output/'FILM_CARD_NUMERICAL_CHECK.json').write_text(json.dumps({'nfl_id':card['nfl_id'],'selected_turn_seconds':float(sec[index]),
                                                                  'entry_speed_yards_s':incoming,'positive_shortfall_area_yards':area,
                                                                  'area_divided_by_entry_speed_seconds':cost,'matches_extracted_turn_metric':True,
                                                                  'reliability_established':False},indent=2))


def main():
    parser=argparse.ArgumentParser()
    parser.add_argument('--raw-dir',type=Path,required=True)
    parser.add_argument('--output-dir',type=Path,required=True)
    parser.add_argument('--save-player-cards',action='store_true',help='Local audit only; omit for aggregate public outputs.')
    args=parser.parse_args();args.output_dir.mkdir(parents=True,exist_ok=True)
    players=pd.read_csv(args.raw_dir/'players.csv',usecols=['nfl_id','display_name','draft_year'])
    results=pd.read_csv(args.raw_dir/'combine_results.csv')
    ids=set(results.loc[results.draft_year.isin([2023,2024])&results.combine_position.eq('OL'),'nfl_id'])
    pieces=[]
    for group in pd.read_csv(args.raw_dir/'combine_tracking.csv',chunksize=200000,low_memory=False):
        q=group[group.draft_year.isin([2023,2024])&group.nfl_id.isin(ids)&group.entity_type.eq('PLAYER')&
                group.drill_name.str.replace('-','_',regex=False).eq('PASS_PRO_MIRROR_DRILL')]
        if len(q):pieces.append(q)
    tracking=pd.concat(pieces,ignore_index=True)
    attempts,events,audit=extract_curves(tracking)
    cards=[]
    names=dict(zip(players.nfl_id.astype(int),players.display_name))
    for player_id,qa in audit.groupby('nfl_id',sort=True):
        aa=attempts[attempts.nfl_id.eq(player_id)]
        turns=events[events.attempt_id.isin(set(aa.attempt_id))].sort_values(['event_id','attempt','reversal_time_seconds'])
        card={'nfl_id':int(player_id),'display_name':str(names.get(int(player_id),'Unknown')),
              'draft_year':int(qa.draft_year.iloc[0]),'official_attempt_count':len(qa),
              'complete_curve_attempt_count':len(aa),'complete_curve_turns':len(turns),
              'quality_status':qa.quality.tolist(),
              'repeatability_established':False,
              'measurement_gate':'One official attempt: compare execution context and film; do not interpret as a stable talent grade.',
              'curve_window_seconds':TIME_GRID.tolist(),'turns':[]}
        for _,turn in turns.iterrows():
            card['turns'].append({'event_id':str(turn.event_id),'attempt':int(turn.attempt),
                                  'reversal_time_from_attempt_start_seconds':float(turn.reversal_time_seconds),
                                  'incoming_lateral_speed_yards_per_second':float(turn.incoming_speed),
                                  'normalized_half_second_recovery_cost_seconds':float(turn.normalized_deficit),
                                  'outgoing_y_direction':int(turn.direction),
                                  'outgoing_signed_speed_over_entry_speed_curve':[float(turn[c]) for c in CURVE]})
        if len(turns):
            card['incoming_speed_range_yards_per_second']=[float(turns.incoming_speed.min()),float(turns.incoming_speed.max())]
            card['recovery_cost_range_seconds']=[float(turns.normalized_deficit.min()),float(turns.normalized_deficit.max())]
            card['lateral_excursion_p90_p10_yards']=float(aa.lateral_amplitude.median())
        card['coach_actions']=['Review the listed turns in film and compare entry speed, excursion and cue sequence.',
                               'Collect another standardized attempt before estimating between-attempt consistency or ranking athletes.',
                               'Record coach stimulus timing separately if the intended quantity is reaction latency.']
        cards.append(card)
    output={'status':'executed_Combine_only_film_review_aids','NFL_outcomes_used_for_cards_or_selection':False,
            'players':len(cards),'official_attempts':len(audit),'usable_attempts':len(attempts),
            'repeated_player_attempts':int(audit.groupby('nfl_id').size().ge(2).sum()),
            'example_selection_rule':'Lowest NFL identifier among usable OL mirror players; independent of NFL outcome',
            'example_nfl_id':int(attempts.nfl_id.min()),'cards':cards}
    if args.save_player_cards:
        (args.output_dir/'SCOUTING_CARDS.json').write_text(json.dumps(output,indent=2))
    example=next(c for c in cards if c['nfl_id']==output['example_nfl_id'])
    if args.save_player_cards:
        (args.output_dir/'EXAMPLE_FILM_CARD.json').write_text(json.dumps(example,indent=2))
    draw_film_card(tracking,example,args.output_dir)
    aggregate={k:output[k] for k in output if k!='cards'}
    aggregate['example_display_name']=example['display_name']
    aggregate['example_entry_speed_range_yards_s']=example['incoming_speed_range_yards_per_second']
    aggregate['example_recovery_cost_range_seconds']=example['recovery_cost_range_seconds']
    aggregate['example_lateral_excursion_p90_p10_yards']=example['lateral_excursion_p90_p10_yards']
    (args.output_dir/'MEASUREMENT_AUDIT.json').write_text(json.dumps(aggregate,indent=2))
    print(json.dumps(aggregate,indent=2))


````

````python
%%writefile --append functional_code/build_scouting_cards.py
if __name__=='__main__':main()
````

### A.8. Frozen-result figure module

These figures are newly rendered from the replayed computations. The original 2025 models and result are unchanged.


````python
%%writefile paper_repro_assets/build_frozen_figure.py
"""A new display of the immutable original 2025 evaluation; no model refit."""
from pathlib import Path
import argparse, json, hashlib
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt

def main():
    p=argparse.ArgumentParser()
    p.add_argument('--result',type=Path,required=True)
    p.add_argument('--output',type=Path,required=True)
    a=p.parse_args();r=json.loads(a.result.read_text())
    assert r['evaluation_players']==19 and r['charted_opportunities']==7801
    fig,ax=plt.subplots(1,3,figsize=(14.2,4.6),layout='constrained')
    blue='#236c7a';orange='#b8633c';grey='#6c7680'
    names=['Constant','Conventional','+ mirror']
    scores=[r[k]['player_equal_brier'] for k in ['constant','baseline','enhanced']]
    ys=np.arange(3)
    ax[0].scatter(np.array(scores)*1e3,ys,s=70,color=[grey,blue,orange])
    for y,s in zip(ys,scores):ax[0].annotate(f'{s:.6f}',(s*1e3,y),xytext=(7,7),textcoords='offset points',fontsize=10)
    ax[0].axvline(scores[0]*1e3,color=grey,linestyle=':',alpha=.65)
    ax[0].set(yticks=ys,yticklabels=names,xlabel='Equal-player Brier loss ×1,000\nLower is better',title='A  Constant forecast wins',xlim=(83.35,85.2),ylim=(-.6,2.6))
    ax[0].invert_yaxis()
    delta=r['brier_improvement']
    for y,key,col in [(0,'fixed_state_player_bootstrap95',orange),(1,'pooled_training_and_evaluation_refit_bootstrap95',blue)]:
        lo,hi=r[key]
        ax[1].errorbar(delta*1e3,y,xerr=np.array([[delta-lo],[hi-delta]])*1e3,fmt='o',color=col,capsize=5)
    ax[1].axvline(0,color='black',linestyle='--',linewidth=1)
    ax[1].set(yticks=[0,1],yticklabels=['Fixed fitted states','Train + test player refits'],xlabel='Conventional − mirror Brier ×1,000\nPositive favors mirror',title='B  Keep both uncertainty scopes',ylim=(-.6,1.6))
    ax[1].invert_yaxis()
    means=[r[k]['mean_prediction']*100 for k in ['constant','baseline','enhanced']]
    obs=r['baseline']['observed_mean_rate']*100
    ax[2].scatter(means,ys,s=70,color=[grey,blue,orange])
    for y,m in zip(ys,means):ax[2].annotate(f'{m:.2f}',(m,y),xytext=(7,7),textcoords='offset points',fontsize=10)
    ax[2].axvline(obs,color='black',linestyle='--',linewidth=1,label=f'Observed equal-player rate {obs:.2f}')
    ax[2].legend(fontsize=8,loc='lower right',frameon=False)
    ax[2].set(yticks=ys,yticklabels=names,xlabel='Average pressures per100 charted observations',title='C  Mirror forecast is too high on average',ylim=(-.6,2.6),xlim=(8.7,12.25))
    ax[2].invert_yaxis()
    for x in ax:x.spines[['top','right']].set_visible(False);x.grid(axis='x',alpha=.15)
    fig.suptitle('Original frozen 2025 evaluation | 19 rookies, 7,801 charted observations',fontsize=15)
    a.output.parent.mkdir(parents=True,exist_ok=True)
    fig.savefig(a.output,dpi=180)
    plt.close(fig)
    a.output.with_suffix('.provenance.json').write_text(json.dumps({'source_sha256':hashlib.sha256(a.result.read_bytes()).hexdigest(),'source':a.result.name,'new_fit':False,'new_evaluation':False,'display_only_revision':True},indent=2))

if __name__=='__main__':main()
````

### A.9. Technical-paper figure display module

This module changes plotting scales, typography and labels. It does not fit or select models.


````python
%%writefile paper_repro_assets/render_paper_figures.py
"""Journal-style displays of previously verified results; no models are refitted.

The only computations are plotting-scale changes and previously specified
descriptive summaries. Raw tracking, when available, is used only to redraw the
same deterministically selected film-review example. Captions are external.
"""
from __future__ import annotations

import argparse
import hashlib
import json
from pathlib import Path

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np

BLUE = "#176B86"
ORANGE = "#BB613C"
GREY = "#667078"
PALE = "#CFD5D9"
PURPLE = "#7B6499"
GOLD = "#D0AA5B"


def paper_style():
    plt.rcParams.update({
        "font.family": "DejaVu Sans", "font.size": 9.5,
        "axes.titlesize": 10.5, "axes.titleweight": "normal",
        "axes.labelsize": 9.5, "xtick.labelsize": 9, "ytick.labelsize": 9,
        "legend.fontsize": 8.5, "axes.spines.top": False,
        "axes.spines.right": False, "axes.linewidth": .8,
        "figure.facecolor": "white", "axes.facecolor": "white",
        "savefig.facecolor": "white", "pdf.fonttype": 42,
        "svg.fonttype": "none", "axes.unicode_minus": True,
    })


def axis_style(ax, grid_axis="x"):
    ax.set_axisbelow(True)
    ax.grid(axis=grid_axis, color="#E5E8EA", linewidth=.65)
    ax.tick_params(length=3.5, width=.7)


def panel(ax, letter, title):
    ax.set_title(f"({letter}) {title}", loc="left", pad=10)


def save_figure(fig, output):
    output = Path(output)
    output.parent.mkdir(parents=True, exist_ok=True)
    fig.savefig(output, dpi=220, bbox_inches="tight", pad_inches=.08)
    fig.savefig(output.with_suffix(".pdf"), bbox_inches="tight", pad_inches=.08)
    plt.close(fig)


def draw_selection(transport, sensitivity, frozen, output):
    """All roster totals, seven differences, and every archived sensitivity point."""
    paper_style()
    fig, axes = plt.subplots(2, 2, figsize=(11.8, 8.0), layout="constrained")
    ax = axes[0, 0]
    rows = transport["original_2025_archived_not_recomputed"]["cohort_coverage"]
    x = np.arange(3); width = .34
    ax.bar(x-width/2, [r["combine_OL"] for r in rows], width, color=PALE,
           label="Combine OL roster")
    ax.bar(x+width/2, [r["eligible_min50"] for r in rows], width, color=BLUE,
           label="Usable mirror + ≥50 charted observations")
    for i, r in enumerate(rows):
        ax.annotate(f'{r["eligible_min50"]}/{r["combine_OL"]}',
                    (i+width/2, r["eligible_min50"]), xytext=(0, 4),
                    textcoords="offset points", ha="center", fontsize=8.5)
    ax.set(xticks=x, xticklabels=["2023", "2024", "2025"],
           ylabel="Number of prospects", ylim=(0, 58))
    ax.legend(loc="upper left", frameon=False, fontsize=8)
    panel(ax, "a", "Eligibility by draft cohort")
    axis_style(ax, "y")

    labels = {"combine_height": "Height", "combine_weight": "Weight",
              "forty": "40-yard time", "ten_yd_split": "10-yard split",
              "normalized_deficit": "Relative recovery cost",
              "incoming_speed": "Entry lateral speed",
              "lateral_amplitude": "Lateral excursion"}
    ax = axes[0, 1]
    selection = transport["R7"]["selection_smd"]
    y = np.arange(len(selection))
    values = np.array([r["standardized_mean_difference"] for r in selection])
    ax.barh(y, values, color=BLUE, height=.68)
    ax.axvline(0, color=GREY, linewidth=.8)
    ax.set(yticks=y, yticklabels=[labels[r["feature"]] for r in selection],
           xlabel="(Retained − excluded mean) / all-prospect SD", xlim=(-.61, .67))
    ax.invert_yaxis()
    for yy, vv in zip(y, values):
        ax.annotate(f"{vv:+.2f}", (vv, yy), xytext=(4 if vv >= 0 else -4, 0),
                    textcoords="offset points", va="center",
                    ha="left" if vv >= 0 else "right", fontsize=8)
    panel(ax, "b", "Development-cohort selection")
    axis_style(ax)

    ax = axes[1, 0]
    primary = transport["R8"]["primary"]["combine_minus_mirror"]
    ax.scatter([primary*1000], [0], color=GREY, marker="D", s=35, zorder=4)
    rng = np.random.default_rng(989)
    for yy, name, color in [(1, "splits", PURPLE), (2, "delete", BLUE)]:
        values = np.array([r["combine_minus_mirror"] for r in sensitivity[name]])
        ax.scatter(values*1000, yy+rng.uniform(-.09, .09, len(values)),
                   s=18, alpha=.82, color=color, linewidths=0)
    draws = np.array([r["baseline_minus_mirror"] for r in sensitivity["boot"]])
    lo, hi = np.quantile(draws, [.025, .975]); median = np.median(draws)
    ax.errorbar(median*1000, 3, xerr=np.array([[median-lo], [hi-median]])*1000,
                fmt="o", markersize=4.5, color=ORANGE, capsize=3.5, linewidth=1.25)
    ax.axvline(0, color=GREY, linestyle="--", linewidth=.8)
    ax.set(yticks=[0, 1, 2, 3],
           yticklabels=["Primary split", "20 split repeats", "38 player deletions",
                        "500 nested resamples"], ylim=(-.45, 3.5),
           xlabel="Conventional − mirror Brier (×10⁻³)")
    ax.invert_yaxis(); panel(ax, "c", "Exploratory model-selection stability")
    axis_style(ax)

    ax = axes[1, 1]
    delta = frozen["brier_improvement"]
    for yy, key, color in [(0, "fixed_state_player_bootstrap95", ORANGE),
                           (1, "pooled_training_and_evaluation_refit_bootstrap95", BLUE)]:
        lo, hi = frozen[key]
        ax.errorbar(delta*1000, yy, xerr=np.array([[delta-lo], [hi-delta]])*1000,
                    fmt="o", markersize=4.5, color=color, capsize=3.5, linewidth=1.25)
    ax.axvline(0, color=GREY, linestyle="--", linewidth=.8)
    ax.set(yticks=[0, 1], yticklabels=["Fixed fitted states", "Train + test player refits"],
           ylim=(-.55, 1.55), xlabel="Conventional − mirror Brier (×10⁻³)")
    ax.invert_yaxis(); panel(ax, "d", "Original frozen 2025 comparison")
    axis_style(ax)
    save_figure(fig, output)


````

````python
%%writefile --append paper_repro_assets/render_paper_figures.py
def draw_frozen(frozen, output):
    """Immutable 2025 losses, archived uncertainty scopes, and calibration means."""
    paper_style()
    assert frozen["evaluation_players"] == 19
    assert frozen["charted_opportunities"] == 7801
    fig, axes = plt.subplots(1, 3, figsize=(12.2, 4.25), layout="constrained")
    keys = ["constant", "baseline", "enhanced"]
    names = ["Constant", "Conventional", "+ mirror"]
    colors = [GREY, BLUE, ORANGE]; y = np.arange(3)
    ax = axes[0]
    loss = np.array([frozen[k]["player_equal_brier"] for k in keys])
    ax.scatter(loss*1000, y, s=35, color=colors, zorder=4)
    for yy, vv in zip(y, loss):
        ax.annotate(f"{vv:.6f}", (vv*1000, yy), xytext=(5, 7),
                    textcoords="offset points", fontsize=8.5)
    ax.axvline(loss[0]*1000, color=GREY, linestyle=":", linewidth=1)
    ax.set(yticks=y, yticklabels=names, ylim=(-.6, 2.6),
           xlim=(83.35, 85.15), xlabel="Equal-player Brier loss (×10⁻³)")
    ax.invert_yaxis(); panel(ax, "a", "Forecast loss")
    axis_style(ax)

    ax = axes[1]; delta = frozen["brier_improvement"]
    for yy, key, color in [(0, "fixed_state_player_bootstrap95", ORANGE),
                           (1, "pooled_training_and_evaluation_refit_bootstrap95", BLUE)]:
        lo, hi = frozen[key]
        ax.errorbar(delta*1000, yy, xerr=np.array([[delta-lo], [hi-delta]])*1000,
                    fmt="o", markersize=4.5, color=color, capsize=3.5, linewidth=1.25)
    ax.axvline(0, color=GREY, linestyle="--", linewidth=.8)
    ax.set(yticks=[0, 1], yticklabels=["Fixed fitted states", "Train + test\nplayer refits"],
           ylim=(-.6, 1.6), xlabel="Conventional − mirror Brier (×10⁻³)")
    ax.invert_yaxis(); panel(ax, "b", "Two uncertainty scopes")
    axis_style(ax)

    ax = axes[2]
    means = np.array([frozen[k]["mean_prediction"] for k in keys])*100
    observed = frozen["baseline"]["observed_mean_rate"]*100
    ax.scatter(means, y, s=35, color=colors, zorder=4)
    for yy, vv in zip(y, means):
        ax.annotate(f"{vv:.2f}", (vv, yy), xytext=(5, 7),
                    textcoords="offset points", fontsize=8.5)
    ax.axvline(observed, color=GREY, linestyle="--", linewidth=.8,
               label=f"Observed: {observed:.2f}")
    ax.legend(frameon=False, loc="lower right", fontsize=8)
    ax.set(yticks=y, yticklabels=names, ylim=(-.6, 2.6), xlim=(8.65, 12.15),
           xlabel="Mean pressures per 100 charted observations")
    ax.invert_yaxis(); panel(ax, "c", "Mean forecast and observed rate")
    axis_style(ax)
    save_figure(fig, output)


def draw_count(counts, intervals, output, primary=None, interval_rows=None):
    """Use exact runtime row data when supplied; archived aggregates otherwise.

    Local aggregate-only preview shows point losses, not invented confidence
    intervals. The notebook can supply the already fitted row tables to retain
    the original four fixed-OOF bootstrap intervals exactly.
    """
    paper_style()
    fig, axes = plt.subplots(1, 3, figsize=(12.2, 4.3), layout="constrained")
    models = ["nested_conventional_ridge", "beta_binomial_conventional",
              "beta_binomial_full", "nested_full_ridge"]
    names = ["Conventional ridge", "Conventional\nbeta-binomial",
             "Mirror beta-binomial", "Mirror ridge"]
    ax = axes[0]; y = np.arange(4)
    if primary is not None:
        wide = primary.pivot(index="player_id", columns="model", values="prediction")
        target = primary[primary.model.eq("constant_equal_player")].set_index("player_id").loc[wide.index]
        rate = (target.y/target.n).to_numpy()
        samples = np.random.default_rng(87007).integers(0, len(wide), size=(5000, len(wide)))
        means=[]; limits=[]
        for model in models:
            delta = (rate-wide.constant_equal_player.to_numpy())**2-(rate-wide[model].to_numpy())**2
            means.append(delta.mean()); limits.append(np.quantile(delta[samples].mean(1), [.025, .975]))
        means = np.array(means); limits = np.array(limits)
        ax.errorbar(means*1000, y,
                    xerr=np.vstack([means-limits[:, 0], limits[:, 1]-means])*1000,
                    fmt="o", color=BLUE, capsize=3.5, markersize=4.5, linewidth=1.25)
        ax.axvline(0, color=GREY, linestyle="--", linewidth=.8)
        ax.set(xlabel="Constant − model Brier (×10⁻³)")
        panel(ax, "a", "Fixed-OOF comparison")
    else:
        scores=np.array([counts["metrics"][model]["equal_player_brier"] for model in models])
        ax.scatter(scores*1000, y, s=35, color=BLUE, zorder=4)
        base=counts["metrics"]["constant_equal_player"]["equal_player_brier"]
        ax.axvline(base*1000, color=GREY, linestyle="--", linewidth=.8,
                   label=f"Constant: {base:.6f}")
        for yy, vv in zip(y, scores):
            ax.annotate(f"{vv:.6f}", (vv*1000, yy), xytext=(4, 7),
                        textcoords="offset points", fontsize=8)
        ax.legend(frameon=False, loc="lower right", fontsize=8)
        ax.set(xlim=(82.79, 83.02), xlabel="Equal-player Brier loss (×10⁻³)")
        panel(ax, "a", "Exploratory forecast loss")
    ax.set(yticks=y, yticklabels=names, ylim=(-.55, 3.6))
    ax.invert_yaxis(); axis_style(ax)

    rows=[intervals[k] for k in ["beta_binomial_constant", "beta_binomial_full"]]
    x=np.arange(2); width=.33
    ax=axes[1]
    left=ax.bar(x-width/2, [r["central90_count_coverage"]*100 for r in rows],
                width, color=ORANGE, label="Beta-binomial central 90%")
    right=ax.bar(x+width/2, [r["jackknife_style_alpha05_empirical_coverage"]*100 for r in rows],
                 width, color=BLUE, label="Jackknife+ style, α = 0.05")
    for bars in [left, right]:
        for b in bars:
            ax.annotate(f"{round(b.get_height()/100*38)}/38",
                        (b.get_x()+b.get_width()/2, b.get_height()), xytext=(0, 4),
                        textcoords="offset points", ha="center", fontsize=8.5)
    ax.set(xticks=x, xticklabels=["Constant prior", "Mirror prior"], ylim=(0, 110),
           ylabel="Empirical player coverage (%)")
    ax.legend(loc="lower center", bbox_to_anchor=(.5, -.29), frameon=False, fontsize=8)
    panel(ax, "b", "Empirical interval coverage"); axis_style(ax, "y")

    ax=axes[2]
    for yy, r in enumerate(rows):
        central=r["median_central90_width"]*100
        jack=r["median_jackknife_style_alpha05_width"]*100
        ax.plot([central, jack], [yy, yy], color=PALE, linewidth=1.5)
        ax.scatter([central, jack], [yy, yy], color=[ORANGE, BLUE], s=35, zorder=4)
        for vv, offset in [(central, -15), (jack, 8)]:
            ax.annotate(f"{vv:.2f}", (vv, yy), xytext=(0, offset),
                        textcoords="offset points", ha="center", fontsize=8.5)
    ax.set(yticks=x, yticklabels=["Constant prior", "Mirror prior"], ylim=(-.6, 1.6),
           xlim=(4.9, 13.45), xlabel="Median interval width (percentage points)")
    ax.invert_yaxis(); panel(ax, "c", "Interval width"); axis_style(ax)
    save_figure(fig, output)


def draw_film(tracking, card, output):
    """Redraw the exact first retained turn without changing its detector."""
    from scipy.signal import savgol_filter
    import pandas as pd
    paper_style()
    turn=card["turns"][0]
    group=tracking[tracking.nfl_id.eq(card["nfl_id"]) &
                   tracking.event_id.astype(str).eq(turn["event_id"]) &
                   tracking.attempt.eq(turn["attempt"])].sort_values("time")
    sec=(pd.to_datetime(group.time)-pd.to_datetime(group.time.iloc[0])).dt.total_seconds().to_numpy()
    x,y=group.x.to_numpy(float),group.y.to_numpy(float)
    origin_x=np.median(x[:min(10,len(x))]); origin_y=y[0]
    exits=np.flatnonzero((np.abs(x-origin_x)>1)&(sec>.5))
    stop=int(exits[0]) if len(exits) else len(x)
    velocity=np.gradient(savgol_filter(y[:stop],5,2), sec[:stop])
    index=int(np.argmin(np.abs(sec-turn["reversal_time_from_attempt_start_seconds"])))
    direction=turn["outgoing_y_direction"]
    incoming=turn["incoming_lateral_speed_yards_per_second"]
    selected=slice(max(0,index-6),min(stop,index+9))
    relative=sec[selected]-sec[index]; oriented=direction*velocity[selected]
    window_time=sec[index:index+6]-sec[index]
    shortfall=np.maximum(0,incoming-direction*velocity[index:index+6])
    area=float(np.trapezoid(shortfall,window_time));cost=area/incoming
    assert np.isclose(cost,turn["normalized_half_second_recovery_cost_seconds"],atol=1e-12)
    fig,axes=plt.subplots(1,2,figsize=(11.8,4.55),layout="constrained")
    ax=axes[0]
    ax.plot(x[:stop]-origin_x,y[:stop]-origin_y,color=BLUE,lw=1.7,label="Mirror phase")
    if stop<len(x):
        end=min(len(x),stop+12)
        ax.plot(x[stop-1:end]-origin_x,y[stop-1:end]-origin_y,color=ORANGE,lw=1.3,ls="--",
                label="Excluded forward run")
        ax.scatter(x[stop]-origin_x,y[stop]-origin_y,s=30,marker="x",color=ORANGE,zorder=5)
        boundary=np.sign(x[stop]-origin_x)
        ax.axvline(boundary,color=GREY,ls=":",lw=.8)
        ax.text(boundary+.08,.64,f"Forward-exit boundary\n({boundary:+.0f} yd)",
                transform=ax.get_xaxis_transform(),va="top",ha="left",
                fontsize=8,color=GREY)
    ax.scatter(x[0]-origin_x,y[0]-origin_y,color=GREY,s=30,zorder=6)
    ax.annotate("Start",(x[0]-origin_x,y[0]-origin_y),xytext=(-25,-13),textcoords="offset points",
                arrowprops={"arrowstyle":"->","linewidth":.7},fontsize=8.5)
    ax.scatter(x[index]-origin_x,y[index]-origin_y,color=ORANGE,marker="s",s=30,zorder=7)
    ax.annotate(f"Reversal at {sec[index]:.1f} s",(x[index]-origin_x,y[index]-origin_y),xytext=(18,12),
                textcoords="offset points",arrowprops={"arrowstyle":"->","color":ORANGE,"linewidth":.7},
                fontsize=8.5,color=ORANGE)
    ax.set_aspect("equal",adjustable="datalim")
    ax.set(xlabel="Forward displacement from initial median (yd)",
           ylabel="Lateral displacement from start (yd)")
    ax.legend(frameon=False,loc="lower right",fontsize=8)
    panel(ax,"a","Selected mirror-drill trajectory");axis_style(ax,"both")
    ax=axes[1]
    ax.plot(relative,oriented,color=BLUE,lw=1.7,label="Outgoing-signed velocity")
    ax.axhline(incoming,color=PURPLE,ls="--",lw=1.2,label=f"Entry speed: {incoming:.2f} yd/s")
    ax.axhline(0,color=GREY,lw=.7);ax.axvline(0,color=ORANGE,lw=.8)
    ax.fill_between(window_time,incoming-shortfall,incoming,color=GOLD,alpha=.65,
                    label="Positive velocity shortfall, 0–0.5 s")
    ax.axvspan(0,.5,color=GOLD,alpha=.07)
    ax.text(.035,.96,f"Area: {area:.3f} yd\nArea / entry speed: {cost:.3f} s",transform=ax.transAxes,
            va="top",fontsize=8.5,bbox={"facecolor":"white","edgecolor":"#E1E5E7","alpha":.98})
    ax.set(xlabel="Time from sampled kinematic reversal (s)",
           ylabel="Outgoing-signed lateral velocity (yd/s)")
    ax.legend(frameon=False,loc="lower right",fontsize=8)
    panel(ax,"b","Half-second movement recovery cost");axis_style(ax,"both")
    save_figure(fig,output)
    return {"nfl_id":card["nfl_id"],"turn_seconds":float(sec[index]),
            "entry_speed_yards_s":incoming,"area_yards":area,"cost_seconds":cost,
            "matches_original_metric":True,"reliability_established":False}


````

````python
%%writefile --append paper_repro_assets/render_paper_figures.py
def draw_film_from_file(tracking_path, functional_code_directory, output):
    """Small official-input extraction for the same deterministic film example.

    Only the already selected example is read into memory. No row table is
    written, no NFL outcomes are opened, and the curve detector is unchanged.
    """
    import sys
    import pandas as pd
    sys.path.insert(0,str(functional_code_directory))
    from run_functional import extract_curves
    fragments=[]
    for frame in pd.read_csv(tracking_path,chunksize=200000,low_memory=False):
        mask=(frame.nfl_id.eq(55870)&frame.entity_type.eq("PLAYER")&
              frame.drill_name.str.replace("-","_",regex=False).eq("PASS_PRO_MIRROR_DRILL"))
        if mask.any():
            fragments.append(frame.loc[mask])
    tracking=pd.concat(fragments,ignore_index=True)
    attempts,events,audit=extract_curves(tracking)
    turns=events[events.attempt_id.isin(set(attempts.attempt_id))].sort_values(["event_id","attempt","reversal_time_seconds"])
    turn=turns.iloc[0]
    card={"nfl_id":55870,"turns":[{
        "event_id":str(turn.event_id),"attempt":int(turn.attempt),
        "reversal_time_from_attempt_start_seconds":float(turn.reversal_time_seconds),
        "incoming_lateral_speed_yards_per_second":float(turn.incoming_speed),
        "normalized_half_second_recovery_cost_seconds":float(turn.normalized_deficit),
        "outgoing_y_direction":int(turn.direction),
    }]}
    return draw_film(tracking,card,output)


def main():
    parser=argparse.ArgumentParser()
    parser.add_argument("--source-base",type=Path,required=True)
    parser.add_argument("--output-dir",type=Path,required=True)
    parser.add_argument("--tracking",type=Path)
    parser.add_argument("--predictions",type=Path)
    parser.add_argument("--interval-rows",type=Path)
    args=parser.parse_args();base=args.source_base;out=args.output_dir
    out.mkdir(parents=True,exist_ok=True)
    source_paths={
        "transport":base/"final_revision/verified_run/transport_appendix_outputs/TRANSPORT_APPENDIX_AGGREGATE.json",
        "frozen":base/"final_revision/verified_run/FINAL_2025_RESULT.json",
        "counts":base/"final_revision/verified_run/hierarchical_results/results.json",
        "intervals":base/"final_revision/verified_run/hierarchical_results/interval_coverage.json",
        "splits":base/"transport_validation/round8/split_sensitivity.json",
        "delete":base/"transport_validation/round8/deletion_influence.json",
        "boot":base/"transport_validation/round8/retuned_bootstrap.json",
    }
    d={k:json.loads(p.read_text()) for k,p in source_paths.items()}
    draw_selection(d["transport"],d,d["frozen"],out/"figure2_selection_stability.png")
    draw_frozen(d["frozen"],out/"figure3_frozen_result.png")
    primary=None; interval_rows=None; checks={}
    if args.predictions:
        import pandas as pd
        primary=pd.read_csv(args.predictions)
        n=primary[primary.model.eq("constant_equal_player")]
        assert n.player_id.nunique()==d["counts"]["n_players"]==38
        assert int(n.n.sum())==d["counts"]["observations"]==16165
        assert int(n.y.sum())==d["counts"]["pressures"]==1444
        maximum=0.0
        for model, rows in primary.groupby("model"):
            if model not in d["counts"]["metrics"]:
                continue
            rate=rows.y/rows.n
            brier=float((rate*(1-rows.prediction)**2+(1-rate)*rows.prediction**2).mean())
            diff=abs(brier-d["counts"]["metrics"][model]["equal_player_brier"])
            maximum=max(maximum,diff)
            assert diff<1e-8,(model,diff)
        checks["private_prediction_max_brier_difference_from_verified_aggregate"]=maximum
        source_paths["private_predictions"]=args.predictions
    if args.interval_rows:
        import pandas as pd
        interval_rows=pd.read_csv(args.interval_rows)
        maximum=0.0
        for model,rows in interval_rows.groupby("model"):
            if model not in d["intervals"]:
                continue
            rate=rows.y/rows.n
            values={
                "central90_count_coverage":float(((rate>=rows.conditional90_low)&(rate<=rows.conditional90_high)).mean()),
                "jackknife_style_alpha05_empirical_coverage":float(((rate>=rows.jackknife_plus_low)&(rate<=rows.jackknife_plus_high)).mean()),
                "median_central90_width":float((rows.conditional90_high-rows.conditional90_low).median()),
                "median_jackknife_style_alpha05_width":float((rows.jackknife_plus_high-rows.jackknife_plus_low).median()),
            }
            for key,value in values.items():
                diff=abs(value-d["intervals"][model][key]);maximum=max(maximum,diff)
                assert diff<1e-8,(model,key,diff)
        checks["private_interval_max_difference_from_verified_aggregate"]=maximum
        source_paths["private_intervals"]=args.interval_rows
    draw_count(d["counts"],d["intervals"],out/"figure4_count_uncertainty.png",primary,interval_rows)
    if args.tracking:
        check=draw_film_from_file(args.tracking,base/"final_revision/verified_run/functional_code",out/"figure1_film_card.png")
        original=json.loads((base/"final_revision/verified_run/functional_results/FILM_CARD_NUMERICAL_CHECK.json").read_text())
        for left,right in [("turn_seconds","selected_turn_seconds"),
                           ("entry_speed_yards_s","entry_speed_yards_s"),
                           ("area_yards","positive_shortfall_area_yards"),
                           ("cost_seconds","area_divided_by_entry_speed_seconds")]:
            assert abs(check[left]-original[right])<1e-12,(left,check[left],original[right])
        checks["film_card"]=check
        checks["film_card_matches_verified_run_check_atol"]=1e-12
        source_paths["private_tracking"]=args.tracking
    provenance={"display_only":True,"new_fit":False,"new_evaluation":False,
        "source_sha256":{k:hashlib.sha256(p.read_bytes()).hexdigest() for k,p in source_paths.items()},
        "figure_sha256":{p.name:hashlib.sha256(p.read_bytes()).hexdigest() for p in out.glob("figure*.png")},
        "figure4_local_panel_a":"Exact four original fixed-OOF bootstrap intervals, unchanged draws and seed." if primary is not None else "Verified aggregate point loss; original interval computation available via primary= runtime row table.",
        "figure1":"Regenerated from official private tracking and matched archived numerical check." if args.tracking else "Faithful callable renderer supplied; awaiting private tracking/card source or exact runtime extraction.",
        "checks":checks}
    (out/"VISUAL_PROVENANCE.json").write_text(json.dumps(provenance,indent=2))
    print(json.dumps(provenance,indent=2))


if __name__=="__main__":
    main()
````

## Execution contract and result map

Restart the kernel and run every cell in order. The notebook creates a dedicated
`paper_reproduction_run` directory; generated row-level files are removed only
inside that directory. Official inputs are read-only. Set `NFL_DATA_DIR` when
the five required CSVs are outside Kaggle or input discovery is ambiguous.
The first eight historical cells replay the already inspected 2025 evaluation;
this is reproduction, not another independent test. All subsequent analysis is
exploratory and uses the 2023–24 cohorts. No fitted state or out-of-fold prediction
is supplied as a shortcut to computation.

| Contract | Value |
|---|---|
| Required inputs | `players.csv`, `combine_results.csv`, `combine_tracking.csv`, `games.csv`, `player_play.csv` |
| Primary unit / endpoint | One prospect; explicit charted rookie regular-season pressure conditional on at least 50 eligible observations |
| Loss weighting | Equal-player binary Brier; opportunity-weighted scores are secondary |
| Original alphas / uncertainty | Conventional 10, scalar mirror 1; evaluation seed 62025; 5,000 fixed-state and 250 paired-refit draws |
| Exploratory folds / grids | Exact 38-player outer five / inner three assignments, seed 87007; primary scalar/curve 1/10/100; count benchmark 0.1/1/10/100 |
| Public output | Aggregate summaries, figures and source; official tables and generated player/frame tables excluded |

| Paper claim | Executable stage | Audit output |
|---|---|---|
| Original frozen 2025 failure and calibration | C. Original frozen computation | `FINAL_2025_RESULT.json` |
| Count benchmark and empirical interval coverage | D. Count-aware benchmark | `hierarchical_results/results.json`, `interval_coverage.json` |
| Opportunity selection, nested selection stability, task support | E. Opportunity-selection audit | `transport_appendix_outputs/TRANSPORT_APPENDIX_AGGREGATE.json` |
| Functional PCA and missing-curve gate | F. Functional representation | `functional_results/CANONICAL38_RESULT.json` |
| Four manuscript figures / source identities | G. Reproduction display | `REVISION_REPRODUCTION.json` and figure hashes |
| Aggregate-only publication boundary | H. Cleanup | Final cleanup assertion and `PAPER_REPRODUCTION_SOURCE.zip` |

Static source preservation, fresh local execution, public Kaggle execution,
browser rendering and competition submission are separate verification states.
The displayed manuscript is an attributed audited snapshot. The appendix
recomputes its scientific results; presentation settings do not establish
execution or submission status.

### C.1. Guarded input and configuration


````python
import csv, io, os, json, hashlib, datetime
from pathlib import Path
import numpy as np
import pandas as pd
import scipy
from scipy.signal import savgol_filter
from scipy.integrate import trapezoid
import matplotlib
import matplotlib.pyplot as plt

# Root must explicitly authorize and freeze this branch before opening 2025.
FINAL_EVALUATION_ENABLED = True
FINAL_SELECTION_AUTHORIZATION = 'OL_selected_by_root'
FROZEN_BASELINE_ALPHA = 10
FROZEN_ENHANCED_ALPHA = 1
SEED = 73025
GRID = [1.0, 10.0, 100.0]
FIXED_BOOT = 5000
REFIT_BOOT = 250
BASE = ['combine_height','combine_weight','forty','ten_yd_split']
FAMILY = ['normalized_deficit','incoming_speed','lateral_amplitude']
MODEL_COLUMNS = {'conventional': BASE, 'normalized': BASE+FAMILY[:1],
                 'speed_amplitude': BASE+FAMILY[1:], 'full_family': BASE+FAMILY}
REQUIRED = ['players.csv','combine_results.csv','combine_tracking.csv','games.csv','player_play.csv']
WORK = Path.cwd()
VERSIONS = {'numpy':np.__version__,'pandas':pd.__version__,'scipy':scipy.__version__,
            'matplotlib':matplotlib.__version__}
print('Runtime:', VERSIONS)
print('Reproduce development first. Reserved evaluation enabled:', FINAL_EVALUATION_ENABLED)

def locate_data():
    explicit=os.environ.get('NFL_DATA_DIR')
    if explicit:
        p=Path(explicit)
        if not all((p/f).is_file() for f in REQUIRED):
            raise FileNotFoundError('NFL_DATA_DIR must contain the five official CSV files.')
        return p
    verified=Path('/kaggle/input/competitions/nfl-big-data-bowl-2027/nfl-big-data-bowl-2027')
    if all((verified/f).is_file() for f in REQUIRED):return verified
    candidates=[p.parent for p in Path('/kaggle/input').glob('**/players.csv')
                if all((p.parent/f).is_file() for f in REQUIRED)]
    if len(candidates)!=1:raise FileNotFoundError('Set NFL_DATA_DIR explicitly; input directory is ambiguous or absent.')
    return candidates[0]

def read_selected(path,predicate):
    # Record selection examines only cohort/game/player keys. Discarded NFL
    # outcomes never enter a typed DataFrame, analysis, model, or output.
    buf=io.StringIO()
    with path.open(newline='') as f:
        reader=csv.reader(f);header=next(reader);index={k:i for i,k in enumerate(header)}
        writer=csv.writer(buf);writer.writerow(header)
        for row in reader:
            if len(row)!=len(header):raise ValueError('Malformed CSV record in '+path.name)
            if predicate(row,index):writer.writerow(row)
    buf.seek(0)
    return pd.read_csv(buf,low_memory=False)

def load_inputs(data_dir,years=(2023,2024),final_enabled=False):
    years=set(int(y) for y in years)
    if not years <= {2023,2024,2025}:raise ValueError('Unexpected cohort year.')
    if 2025 in years and not final_enabled:
        raise PermissionError('2025 loader blocked BEFORE any file is opened.')
    allowed={str(y) for y in years};d=Path(data_dir)
    players=read_selected(d/'players.csv',lambda r,k:r[k['draft_year']] in allowed)
    results=read_selected(d/'combine_results.csv',lambda r,k:r[k['draft_year']] in allowed)
    if not players.nfl_id.is_unique or results.duplicated(['draft_year','nfl_id']).any():
        raise ValueError('Non-unique player/Combine linkage key.')
    ol=results[results.combine_position.eq('OL')].copy()
    ids={str(int(x)) for x in ol.nfl_id};draft=dict(zip(players.nfl_id.astype(str),players.draft_year.astype(int)))
    games=read_selected(d/'games.csv',lambda r,k:r[k['season']] in allowed)
    if not games.game_id.is_unique:raise ValueError('Duplicate game identifiers.')
    reg=games[games.season_type.eq('REG')]
    gameyear=dict(zip(reg.game_id.astype(str),reg.season.astype(int)))
    tracking=read_selected(d/'combine_tracking.csv',lambda r,k:
        r[k['draft_year']] in allowed and r[k['nfl_id']] in ids and
        r[k['entity_type']]=='PLAYER' and r[k['drill_name']].replace('-','_')=='PASS_PRO_MIRROR_DRILL')
    def rookie_record(r,k):
        game=r[k['game_id']];nid=r[k['nfl_id']]
        return game in gameyear and nid in ids and gameyear[game]==draft.get(nid)
    player_play=read_selected(d/'player_play.csv',rookie_record)
    assert set(players.draft_year)<=years and set(results.draft_year)<=years
    assert set(tracking.draft_year)<=years and set(games.season)<=years
    for a in years:
        for b in years:
            if a<b:assert not set(players[players.draft_year.eq(a)].nfl_id)&set(players[players.draft_year.eq(b)].nfl_id)
    if player_play.duplicated(['game_id','play_id','nfl_id']).any():raise ValueError('Duplicate player-play key.')
    return {'players':players,'results':results,'games':games,'tracking':tracking,'player_play':player_play,
            'audit':{'years':sorted(years),'accepted_rows':{'players':len(players),'combine_results':len(results),
                     'combine_tracking_mirror':len(tracking),'games':len(games),'rookie_REG_player_play':len(player_play)},
                     'drill_aliases':sorted(tracking.drill_name.unique().tolist()),'initial_OL_cohort_source':'combine_results.combine_position == OL',
                     'outcomes_selected_by_keys_before_typing':True,'career_file_opened':False}}

CODE_FINGERPRINT = '837b80cf6446558b03c6af2c210132af262e668c2932b40f5b425e5be5550b9b'
````

### C.2. Coordinate-derived mirror profile


````python
def extract(t, window=5, exit_threshold=1.0):
    rows=[]; events=[]; examples={}
    for key,g in t.groupby(['draft_year','event_id','nfl_id','attempt'],sort=True):
        g=g.sort_values('time').copy()
        tm=pd.to_datetime(g.time); sec=(tm-tm.iloc[0]).dt.total_seconds().to_numpy()
        x=g.x.to_numpy(); y=g.y.to_numpy()
        start=np.median(x[:min(10,len(x))])
        exits=np.flatnonzero((np.abs(x-start)>exit_threshold)&(sec>0.5))
        stop=int(exits[0]) if len(exits) else len(x)
        quality='ok'
        if stop<15 or np.any(np.abs(np.diff(sec[:stop])-.1)>.015): quality='short_or_non_10hz'
        if np.isnan(x).any() or np.isnan(y).any(): quality='missing_coordinates'
        rev=[]; deficits=[]
        if quality=='ok':
            sy=savgol_filter(y[:stop],window,2)
            vel=np.gradient(sy,sec[:stop])
            # Zero-speed frames are real at turns: bridge near-zero intervals,
            # rather than requiring adjacent nonzero signs (which misses minima).
            signs=np.where(np.abs(vel)>.05,np.sign(vel),0)
            nz=np.flatnonzero(signs)
            changes=[int(round((i+j)/2)) for i,j in zip(nz[:-1],nz[1:])
                     if signs[i]*signs[j]<0 and j-i<=10]
            last=-100
            for i in changes:
                if i<5 or i+5>=stop or i-last<5: continue
                pre=np.median(vel[i-5:i-1]); post=np.median(vel[i+1:i+5])
                if abs(pre)<.5 or abs(post)<.5 or pre*post>=0: continue
                new=np.sign(post)*vel[i:i+6]
                deficit=float(trapezoid(np.maximum(0,abs(pre)-new),sec[i:i+6]-sec[i]))
                rev.append(i); deficits.append(deficit); last=i
                events.append({'draft_year':key[0],'event_id':key[1],'nfl_id':key[2],'attempt':key[3],
                               'reversal_time_seconds':float(sec[i]),'incoming_speed_yards_s':float(abs(pre)),
                               'recovery_deficit_yards':deficit})
        if len(deficits)<2 and quality=='ok': quality='fewer_than_two_usable_reversals'
        rows.append({'draft_year':key[0],'event_id':key[1],'nfl_id':key[2],'attempt':key[3],
                     'frames':len(g),'mirror_frames':stop,'usable_reversals':len(deficits),
                     'quality':quality,'recovery_deficit':float(np.percentile(deficits,75)) if quality=='ok' else np.nan,
                     'x_start':float(start),'x_range':float(np.ptp(x)),'y_range':float(np.ptp(y))})
        examples[key]={'sec':sec,'x':x,'y':y,'stop':stop,'rev':rev,'quality':quality}
    return pd.DataFrame(rows),pd.DataFrame(events),examples

def extract_mirror(t):
    keys=['draft_year','event_id','nfl_id','attempt','time']
    if t.duplicated(keys).any():raise ValueError('Duplicate tracking timestamp key; extraction refused.')
    if t.empty:raise ValueError('No mirror trajectories; metric is undefined, not zero.')
    if not t.x.between(0,120).all() or not t.y.between(0,53.5).all():
        raise ValueError('Coordinate bounds incompatible with documented field-yard units.')
    return extract(t)

def build_features(t):
    attempts,events,examples=extract_mirror(t)
    keys=['draft_year','event_id','nfl_id','attempt']
    good=attempts[attempts.quality.eq('ok')]
    if good.empty or events.empty:raise ValueError('No usable reversals; metric is undefined.')
    e=events.merge(good[keys],on=keys,validate='many_to_one')
    e['normalized_deficit']=e.recovery_deficit_yards/e.incoming_speed_yards_s
    a=e.groupby(keys,as_index=False).agg(normalized_deficit=('normalized_deficit',lambda z:np.percentile(z,75)),
        incoming_speed=('incoming_speed_yards_s','median'))
    task=[]
    for _,row in good.iterrows():
        key=(int(row.draft_year),row.event_id,int(row.nfl_id),int(row.attempt));ex=examples[key]
        yy=ex['y'][:ex['stop']]
        task.append(dict(zip(keys,key),lateral_amplitude=float(np.percentile(yy,90)-np.percentile(yy,10)),
                         phase_seconds=float(ex['sec'][ex['stop']-1])))
    z=good.merge(a,on=keys,validate='one_to_one').merge(pd.DataFrame(task),on=keys,validate='one_to_one')
    features=z.groupby(['draft_year','nfl_id'],as_index=False).agg(
        normalized_deficit=('normalized_deficit','median'),incoming_speed=('incoming_speed','median'),
        lateral_amplitude=('lateral_amplitude','median'),raw_deficit=('recovery_deficit','median'),
        usable_reversals=('usable_reversals','sum'),usable_attempts=('attempt','size'),phase_seconds=('phase_seconds','median'))
    return features,attempts,events,examples

def prepare_cohort(inputs,years=(2023,2024),endpoint='charted_all'):
    features,attempts,events,examples=build_features(inputs['tracking'][inputs['tracking'].draft_year.isin(years)])
    c=inputs['results'];p=inputs['players'];g=inputs['games']
    c=c[c.draft_year.isin(years)&c.combine_position.eq('OL')]
    base=p[p.draft_year.isin(years)].merge(c[['draft_year','nfl_id']+BASE],on=['draft_year','nfl_id'],validate='one_to_one')
    base=base.merge(features,on=['draft_year','nfl_id'],validate='one_to_one')
    # NFL position is retained ONLY for explicitly labeled historical sensitivity.
    base['historical_is_tackle']=base.nfl_position.eq('T').astype(int)
    q=inputs['player_play'].merge(p[['nfl_id','draft_year']],on='nfl_id',validate='many_to_one').merge(
        g[['game_id','season','season_type']],on='game_id',validate='many_to_one')
    q=q[q.draft_year.isin(years)&q.season.eq(q.draft_year)&q.season_type.eq('REG')].copy()
    good=q.pressure_allowed.notna()&q.lined_up_position.isin(['C','G','T'])&q.team_abbr.eq(q.possession_team)&q.play_nullified_by_penalty.eq('N')
    observed=q[good].copy()
    labels=observed.pressure_allowed.map(lambda z:str(z).lower())
    if not set(labels)<= {'true','false'}:raise ValueError('Unknown charted pressure boolean; no coercion to zero allowed.')
    observed['pressure']=labels.eq('true').astype(int)
    if endpoint=='encountered_at_least_one':observed=observed[observed.pass_rushers_encountered.ge(1)]
    elif endpoint=='pass_results_C_I_IN_S':observed=observed[observed.pass_result.isin(['C','I','IN','S'])]
    elif endpoint=='encountered_and_C_I_IN_S':observed=observed[observed.pass_rushers_encountered.ge(1)&observed.pass_result.isin(['C','I','IN','S'])]
    elif endpoint!='charted_all':raise ValueError('Unrecognized endpoint; no silent substitute.')
    outcome=observed.groupby(['draft_year','nfl_id'],as_index=False).agg(n=('pressure','size'),k=('pressure','sum'))
    outcome['rate']=outcome.k/outcome.n
    allplayers=base.merge(outcome,on=['draft_year','nfl_id'],how='left',validate='one_to_one')
    eligible=allplayers[allplayers.n.ge(50)].copy()
    funnel={str(year):{'Combine_OL_participants':int(c.draft_year.eq(year).sum()),
        'mirror_participants':int(attempts[attempts.draft_year.eq(year)].nfl_id.nunique()),
        'usable_feature_players':int(base.draft_year.eq(year).sum()),'any_rookie_REG_players':int(q[q.draft_year.eq(year)].nfl_id.nunique()),
        'observed_endpoint_players':int((allplayers.draft_year.eq(year)&allplayers.n.notna()).sum()),
        'eligible_players_min50':int(eligible.draft_year.eq(year).sum()),
        'eligible_opportunities':int(eligible.loc[eligible.draft_year.eq(year),'n'].sum()),
        'eligible_pressures':int(eligible.loc[eligible.draft_year.eq(year),'k'].sum()),
        'missing_charted_pressure_rows':int((q.draft_year.eq(year)&q.pressure_allowed.isna()).sum()),
        'observed_zero_rusher_rows':int((observed.draft_year.eq(year)&observed.pass_rushers_encountered.eq(0)).sum())}
        for year in years}
    return {'eligible':eligible,'all_players':allplayers,'funnel':funnel,'attempts':attempts,'events':events,
            'examples':examples,'features':features,'rookie_rows':q,'observed_rows':observed}
````

### C.3. Training-fold preprocessing and models


````python
def fit_predict(xx,y,xt,alpha):
    med=np.nanmedian(xx,axis=0);med=np.where(np.isfinite(med),med,0)
    xx=np.where(np.isnan(xx),med,xx);xt=np.where(np.isnan(xt),med,xt)
    mean=xx.mean(0);std=xx.std(0);std=np.where(std>1e-12,std,1)
    xx=(xx-mean)/std;xt=(xt-mean)/std
    ym=y.mean();coef=np.linalg.solve(xx.T@xx+alpha*np.eye(xx.shape[1]),xx.T@(y-ym))
    return np.clip(ym+xt@coef,.001,.999)

def loss(y,p):return y*(1-p)**2+(1-y)*p*p

def logloss(y,p):return -y*np.log(p)-(1-y)*np.log1p(-p)

def cv_select(x,y,ids):
    scores=[];predictions={}
    for alpha in GRID:
        p=np.full(len(y),np.nan)
        for nid in np.unique(ids):
            test=ids==nid;train=~test
            p[test]=fit_predict(x[train],y[train],x[test],alpha)
        assert np.isfinite(p).all()
        scores.append({'alpha':alpha,'player_cv_brier':float(loss(y,p).mean()),'player_cv_log_loss':float(logloss(y,p).mean()),'player_cv_rate_mse':float(((y-p)**2).mean())})
        predictions[str(alpha)]=p.tolist()
    chosen=min(scores,key=lambda z:(z['player_cv_brier'],-z['alpha']))['alpha']
    return chosen,scores,predictions

def scores(y,p,n):
    return {'player_equal_brier':float(loss(y,p).mean()),'player_equal_log_loss':float(logloss(y,p).mean()),
            'player_rate_mae':float(abs(y-p).mean()),'opportunity_weighted_brier':float(np.average(loss(y,p),weights=n)),
            'mean_prediction':float(p.mean()),'observed_mean_rate':float(y.mean())}

def fit_state(train,columns,alpha):
    if len(train)<10:raise ValueError('At least10 eligible training players required.')
    xx=train[columns].to_numpy(float);y=train.rate.to_numpy(float)
    med=np.nanmedian(xx,axis=0);med=np.where(np.isfinite(med),med,0)
    xx=np.where(np.isnan(xx),med,xx);mean=xx.mean(0);std=xx.std(0);std=np.where(std>1e-12,std,1)
    xx=(xx-mean)/std;ym=y.mean()
    coef=np.linalg.solve(xx.T@xx+alpha*np.eye(xx.shape[1]),xx.T@(y-ym))
    return {'feature_columns':columns,'alpha':float(alpha),'median':med.tolist(),'mean':mean.tolist(),
            'std':std.tolist(),'coefficient':coef.tolist(),'intercept':float(ym),'prediction_clip':[.001,.999],
            'fit_player_ids':[int(x) for x in train.nfl_id],'fit_years':sorted(int(x) for x in train.draft_year.unique())}

def apply_state(state,test):
    xx=test[state['feature_columns']].to_numpy(float);med=np.asarray(state['median'])
    xx=np.where(np.isnan(xx),med,xx);xx=(xx-np.asarray(state['mean']))/np.asarray(state['std'])
    return np.clip(state['intercept']+xx@np.asarray(state['coefficient']),*state['prediction_clip'])

def fit_recipe(train,test,baseline_alpha,enhanced_alpha,base_columns=BASE):
    if len(train)<10 or len(test)<10:raise ValueError('At least 10 eligible players per fit/evaluation cohort are required.')
    y=train.rate.to_numpy();yy=test.rate.to_numpy();n=test.n.to_numpy()
    b=fit_predict(train[base_columns].to_numpy(float),y,test[base_columns].to_numpy(float),baseline_alpha)
    f=fit_predict(train[base_columns+FAMILY].to_numpy(float),y,test[base_columns+FAMILY].to_numpy(float),enhanced_alpha)
    return {'constant':scores(yy,np.full(len(test),y.mean()),n),'baseline':scores(yy,b,n),'enhanced':scores(yy,f,n),
            'baseline_predictions':b,'enhanced_predictions':f}

def development_models(cohort,historical_role=False):
    tr=cohort[cohort.draft_year.eq(2023)].reset_index(drop=True);te=cohort[cohort.draft_year.eq(2024)].reset_index(drop=True)
    if len(tr)<10 or len(te)<10:raise ValueError('At least 10 eligible players in each development cohort are required.')
    y=te.rate.to_numpy();yt=tr.rate.to_numpy();n=te.n.to_numpy();ids=tr.nfl_id.to_numpy()
    base=BASE+(['historical_is_tackle'] if historical_role else [])
    columns={'conventional':base,'normalized':base+FAMILY[:1],'speed_amplitude':base+FAMILY[1:],'full_family':base+FAMILY}
    result={'train_players':len(tr),'selection_players':len(te),'train_opportunities':int(tr.n.sum()),'selection_opportunities':int(te.n.sum()),
        'constant':scores(y,np.full(len(te),yt.mean()),n),'models':{},'comparisons':{}}
    pred={};chosen={};arrays={}
    for name,cols in columns.items():
        xx=tr[cols].to_numpy(float);xt=te[cols].to_numpy(float);arrays[name]=(xx,xt)
        alpha,cv,_=cv_select(xx,yt,ids);chosen[name]=alpha;pred[name]=fit_predict(xx,yt,xt,alpha)
        result['models'][name]={'features':cols,'alpha_selected_2023':alpha,'2023_cv_grid':cv,
            '2024_selection_grid':{str(a):scores(y,fit_predict(xx,yt,xt,a),n) for a in GRID},'selected_model_score':scores(y,pred[name],n)}
    # The role-conditioned history is reconstructed, not silently relabeled.
    if historical_role:return result,{'train':tr,'test':te,'predictions':pred,'alphas':chosen}
    rng=np.random.default_rng(SEED);vi=rng.integers(0,len(te),size=(FIXED_BOOT,len(te)))
    rng=np.random.default_rng(SEED+1);bi=rng.integers(0,len(tr),size=(REFIT_BOOT,len(tr)));bj=rng.integers(0,len(te),size=(REFIT_BOOT,len(te)))
    for name in ['normalized','speed_amplitude','full_family']:
        pb=pred['conventional'];pe=pred[name]
        diffs={'brier':loss(y,pb)-loss(y,pe),'log_loss':logloss(y,pb)-logloss(y,pe)}
        boot={k:[] for k in diffs};xb,xbt=arrays['conventional'];xf,xft=arrays[name]
        for ti,j in zip(bi,bj):
            bp=fit_predict(xb[ti],yt[ti],xbt[j],chosen['conventional']);fp=fit_predict(xf[ti],yt[ti],xft[j],chosen[name])
            boot['brier'].append(float((loss(y[j],bp)-loss(y[j],fp)).mean()))
            boot['log_loss'].append(float((logloss(y[j],bp)-logloss(y[j],fp)).mean()))
        influence=[]
        for drop in range(len(tr)):
            keep=np.arange(len(tr))!=drop
            ab,_,_=cv_select(xb[keep],yt[keep],ids[keep]);af,_,_=cv_select(xf[keep],yt[keep],ids[keep])
            bp=fit_predict(xb[keep],yt[keep],xbt,ab);fp=fit_predict(xf[keep],yt[keep],xft,af)
            influence.append({'removed_nfl_id':int(ids[drop]),'baseline_alpha':ab,'enhanced_alpha':af,
                'brier_improvement':float((loss(y,bp)-loss(y,fp)).mean()),'max_prediction_shift':float(abs(fp-pe).max())})
        result['comparisons'][name]={'baseline_score':scores(y,pb,n),'enhanced_score':scores(y,pe,n),
            'uncertainty':{k:{'baseline_minus_enhanced':float(d.mean()),'fixed_model_player_bootstrap95':np.quantile(d[vi].mean(1),[.025,.975]).tolist(),
                             'training_selection_refit_bootstrap95':np.quantile(boot[k],[.025,.975]).tolist()} for k,d in diffs.items()},
            'training_influence_reselecting_alpha':influence,'max_LOO_prediction_shift':max(z['max_prediction_shift'] for z in influence)}
    pp=te[['draft_year','nfl_id','display_name','n','k','rate']].copy()
    for name,val in pred.items():pp['prediction_'+name]=val
    result['baseline_missingness']={str(year):df[BASE].isna().sum().astype(int).to_dict() for year,df in [(2023,tr),(2024,te)]}
    return result,{'train':tr,'test':te,'predictions':pred,'alphas':chosen,'prediction_frame':pp}
````

### C.4. Original development comparison and audits


````python
DATA_DIR=locate_data()
inputs=load_inputs(DATA_DIR,years=(2023,2024),final_enabled=False)
prepared=prepare_cohort(inputs)
primary,bundle=development_models(prepared['eligible'])
print('Primary 2023 training / 2024 selection players:',primary['train_players'],primary['selection_players'])
print('Timing-safe predictors:',BASE+FAMILY)
print('2023 CV chosen alphas:',bundle['alphas'])
print('Constant, conventional, full-family Brier:',primary['constant']['player_equal_brier'],
      primary['models']['conventional']['selected_model_score']['player_equal_brier'],
      primary['models']['full_family']['selected_model_score']['player_equal_brier'])
print('Primary Brier improvement and uncertainty:',primary['comparisons']['full_family']['uncertainty']['brier'])
bundle['prediction_frame'].to_csv(WORK/'development_predictions.csv',index=False)

historical,_=development_models(prepared['eligible'],historical_role=True)
assert abs(historical['models']['conventional']['selected_model_score']['player_equal_brier']-0.08483422126191108)<1e-10
assert abs(historical['models']['full_family']['selected_model_score']['player_equal_brier']-0.08468259293121042)<1e-10
sensitivities={}
for endpoint in ['encountered_at_least_one','pass_results_C_I_IN_S','encountered_and_C_I_IN_S']:
    alt=prepare_cohort(inputs,endpoint=endpoint)
    rr,_=development_models(alt['eligible'])
    sensitivities[endpoint]={'funnel':alt['funnel'],'models':rr}
print('Role-conditioned R4 values reproduced and labeled historical only.')
print('Three denominator sensitivities retained; none promoted by 2024 score.')

# Meaningful adversarial checks. These do not train/test hypothetical outcomes.
audit={'input':inputs['audit'],'funnel':prepared['funnel'],'coordinate_velocity_source':'x/y/time; supplied s/a/dis ignored',
       'attempt_repeatability':'one recorded mirror attempt per player in development; no physiological reliability claim',
       'timing_correction':'NFL roster is_tackle removed from primary. Initial OL cohort now uses documented Combine_position OL.',
       'event_key_repair':'Full event_id+attempt keys retained for multiple-event safety; development metrics unchanged.'}
t=inputs['tracking'];first=t.sort_values(['draft_year','nfl_id','event_id','attempt','time']).iloc[0]
same=t[(t.draft_year==first.draft_year)&(t.nfl_id==first.nfl_id)&(t.event_id==first.event_id)&(t.attempt==first.attempt)].sort_values('time')
checks={}
try:extract_mirror(pd.concat([same,same.iloc[:1]],ignore_index=True))
except ValueError as e:checks['duplicate_timestamp_rejected']=str(e)
short,_,_=extract_mirror(same.iloc[:10])
assert short.quality.ne('ok').all() and short.recovery_deficit.isna().all()
checks['truncated_actual_trajectory']='First10 frames of a real2023 trajectory produce undefined metric, not0.'
bad=same.copy();bad.loc[bad.index[0],'x']=999
try:extract_mirror(bad)
except ValueError as e:checks['coordinate_unit_failure_rejected']=str(e)
try:fit_recipe(bundle['train'].iloc[:0],bundle['test'],10,10)
except ValueError as e:checks['empty_cohort_rejected']=str(e)
try:load_inputs(Path('/must_not_be_opened'),years=(2025,),final_enabled=False)
except PermissionError as e:checks['final_guard_rejected_before_fileaccess']=str(e)
assert len(checks)==5
audit['adversarial_checks']=checks
truth=prepared['rookie_rows'].copy()
truth['pressure_label']=truth.pressure_allowed.map(lambda x:'missing' if pd.isna(x) else str(x).lower())
truth['rusher_bucket']=np.select([truth.pass_rushers_encountered.isna(),truth.pass_rushers_encountered.eq(0),truth.pass_rushers_encountered.gt(0)],['missing','zero','positive'],default='other')
truth_table=truth.groupby(['draft_year','pressure_label','rusher_bucket']).size().rename('rows').reset_index()
audit['pressure_label_truth_table']=truth_table.to_dict('records')
audit['charted_false']='Explicit label, including0-encounter rows; missing is notFalse. Conditional charted protection estimand.'
audit['accepted_attempts_by_year']={str(y):int((prepared['attempts'].draft_year.eq(y)&prepared['attempts'].quality.eq('ok')).sum()) for y in [2023,2024]}
audit['no_role_primary_assertion']=all('tackle' not in c and c!='nfl_position' for c in BASE+FAMILY)
assert audit['no_role_primary_assertion']
state=fit_state(bundle['train'],BASE+FAMILY,bundle['alphas']['full_family'])
state_prediction=apply_state(state,bundle['test'])
assert np.allclose(state_prediction,bundle['predictions']['full_family'],rtol=0,atol=1e-14)
audit['fitted_state_prediction_equivalence_max_abs_difference']=float(abs(state_prediction-bundle['predictions']['full_family']).max())
print('Adversarial checks:',checks)

# Development-only mechanical sensitivity inherited from earlier audit; no NFL
# score is used to choose an exit threshold, smoothing window or family member.
geo_sensitivity=[]
reference=prepared['features'][prepared['features'].draft_year.eq(2023)]
for window,exit_yards in [(3,1.0),(7,1.0),(5,.75),(5,1.25)]:
    aa,ee,ex=extract(t[t.draft_year.eq(2023)],window,exit_yards)
    good=aa[aa.quality.eq('ok')];keys=['draft_year','event_id','nfl_id','attempt']
    e=ee.merge(good[keys],on=keys,validate='many_to_one');e['norm']=e.recovery_deficit_yards/e.incoming_speed_yards_s
    z=e.groupby(['draft_year','nfl_id'],as_index=False).agg(normalized_deficit=('norm',lambda z:np.percentile(z,75)),incoming_speed=('incoming_speed_yards_s','median'))
    amp=[]
    for _,row in good.iterrows():
        key=(int(row.draft_year),row.event_id,int(row.nfl_id),int(row.attempt));v=ex[key]['y'][:ex[key]['stop']]
        amp.append({'draft_year':int(row.draft_year),'nfl_id':int(row.nfl_id),'lateral_amplitude':float(np.percentile(v,90)-np.percentile(v,10))})
    z=z.merge(pd.DataFrame(amp).groupby(['draft_year','nfl_id'],as_index=False).median(),on=['draft_year','nfl_id'])
    paired=z.merge(reference,on=['draft_year','nfl_id'],suffixes=('_alternative','_primary'))
    geo_sensitivity.append({'smoothing_frames':window,'exit_yards':exit_yards,'common_players':len(paired),
        'rank_correlation':{field:float(scipy.stats.spearmanr(paired[field+'_alternative'],paired[field+'_primary']).statistic) for field in FAMILY}})
audit['geometry_sensitivity_2023_only']=geo_sensitivity
audit['dependency_versions']=VERSIONS
````

### C.6. Original input and development provenance


````python
def input_hash(path):
    h=hashlib.sha256()
    with path.open('rb') as f:
        for block in iter(lambda:f.read(1024*1024),b''):h.update(block)
    return h.hexdigest()

input_ids={f:{'name':f,'bytes':(DATA_DIR/f).stat().st_size,'sha256':input_hash(DATA_DIR/f)} for f in REQUIRED}
selected_alpha_base=bundle['alphas']['conventional'];selected_alpha_full=bundle['alphas']['full_family']
config={'development_years':[2023,2024],'final_enabled':bool(FINAL_EVALUATION_ENABLED),'seed':SEED,'ridge_grid':GRID,
        'primary_endpoint':'charted_all','minimum_charted_opportunities':50,'base_features':BASE,'tracking_family':FAMILY,
        'baseline_alpha_selected_2023':selected_alpha_base,'enhanced_alpha_selected_2023':selected_alpha_full}
manifest={'candidate':'OL_speed_relative_mirror_recovery','created_utc':datetime.datetime.now(datetime.timezone.utc).isoformat(),
          'code_fingerprint':CODE_FINGERPRINT,'dependencies':VERSIONS,'official_input_identifiers':input_ids,'config':config,
          'timing':'Only Combine measurements/tracking in primary predictors; NFL lineup role defines target exposure, not a prediction input.',
          'detector':{'drill_aliases':['PASS_PRO_MIRROR_DRILL','PASS_PRO-MIRROR_DRILL'],'coordinate_units':'yards',
                      'sampling':'10Hz; adjacentdtwithin0.015s of0.1s','lateral_axis':'y','forward_exit':'first abs(x-median(first10x))>1yard after0.5s',
                      'smoothing':'Savitzky-Golay5frames,polynomial2','stationary_bridge':'abs lateralvelocity<=.05yd/s;bridge<=1s',
                      'usable_turn':'preceding/following median speed>.5yd/s, opposite signs,0.5s spacing;>=2reversals',
                      'loss_window_seconds':.5,'recovery_summary':'P75 of reversal deficit/incoming speed;medianacross attempts',
                      'speed_summary':'median incoming speed','excursion':'P90(y)-P10(y) withinmirrorphase;medianacross attempts',
                      'event_keys':['draft_year','event_id','nfl_id','attempt'],'repeatability':'Notestablished'},
          'outcome_policy':'Combine_position OL participants; explicitnon-nullpressure_allowed,offensiveC/G/Tlineup,possessionteam,notnullified,rookie season==draft_year,REGonly;min50. Missingneverzero.',
          'metric_policy':'Primary equal-player mean binary Brier; logloss and calibration descriptive; matchedeligibleplayers; no correlation-based headline.',
          'final_recipe':{'enabled':False,'authorization_required':'OL_selected_by_root','fit_years':[2023,2024],'evaluate_year':2025,
                          'baseline_features':BASE,'enhanced_features':BASE+FAMILY,'baseline_alpha':selected_alpha_base,
                          'enhanced_alpha':selected_alpha_full,'preprocessing':'medianimpute/standardize fitted on pooled2023/24eligibleplayers only',
                          'player_weighting':'equal','prediction_clip':[.001,.999],'cohort_policy':'identical frozen endpoint/min50/drillfeature gates',
                          'no_holdout_model_switch':True},
          'uncertainty_scope':'250training+selectionplayerbootstraprefitsfixselectedalpha;5000fixedmodelplayerdraws;LOOalpha reselected. Excludesalpha/familyselection uncertainty and sharedteam/game dependence.'}
packet={'status':'completed','primary':primary,'denominator_sensitivities':sensitivities,'role_conditioned_R4_history':historical,
        'audit':audit,'config':config,'input_identifiers':input_ids,'freeze_manifest':manifest,'final_2025_opened':False,
        'history':'R2geometrybug repair;R3normalization/denominator audits;R4compactfamily/CV;R5drops unverifiedpre-draftNFLrole. Repeated2024selection; not original preregistration.'}
(WORK/'config.json').write_text(json.dumps(config,indent=2))
(WORK/'freeze_manifest.json').write_text(json.dumps(manifest,indent=2))
(WORK/'ROUND5_RESULT.json').write_text(json.dumps(packet,indent=2,allow_nan=False))

u=primary['comparisons']['full_family']['uncertainty']['brier'];ci=u['training_selection_refit_bootstrap95']
bs=primary['models']['conventional']['selected_model_score'];fs=primary['models']['full_family']['selected_model_score']
article=f'''# Relative mirror recovery: a measured question about rookie pass protection

**Question.** Does a prospect's speed-relative recovery during the Combine mirror drill add information about first-year charted NFL pass protection beyond conventional Combine measurements?

**Current evidence.** The timing-safe profile has a Brier-loss improvement of {u['baseline_minus_enhanced']:+.6f} on 20 rookies in the 2024 exploratory selection cohort. Its training-and-selection player-refit interval is [{ci[0]:+.6f}, {ci[1]:+.6f}]. That uncertainty does not establish incremental predictive value. It also does not establish that tracking is useless. We propose a specific measurement for film review, not a talent ranking or automatic grade adjustment.

## A football measurement, not a reaction-time score

An offensive lineman repeatedly changes lateral direction in the mirror drill, then finishes with a forward run. We separate that run-out before measuring reversal cost. The recorded geometry supports lateral y and forward x; we have no body orientation, opponent trajectory, or stimulus timestamp. Consequently, the measure describes the prospect's own movement, not reaction latency, stance quality, or a causal athletic trait.

For each usable reversal we integrate positive movement shortfall over 0.5 seconds: the preceding lateral speed minus velocity toward the new direction, bounded below by zero. Dividing by incoming speed expresses the shortfall in **effective seconds of speed-relative movement loss**. We use the 75th percentile across reversals and the median across attempts. Two companion descriptors characterize the task: median incoming lateral speed in yards/second and the mirror-phase 90th-minus-10th percentile lateral excursion in yards. Excursion can reflect drill instructions; it is not an ability grade.

Positions are smoothed with a five-frame quadratic Savitzky–Golay filter; velocities come from coordinates and timestamps, not supplied speed or distance columns. Near-zero velocities are bridged to avoid missing stationary turns. Usable turns have opposite directions and median incoming/outgoing speeds above 0.5 yards/second; at least two are required. The forward run starts when x moves more than one yard from its initial reference. Full event and attempt identifiers prevent trajectories from being combined accidentally.

![Figure 1](figure1_geometry.png)

**Figure 1.** Geometry and discretized recovery loss for the smallest player identifier in each cohort and the longest 2023 recording. These cases were selected without NFL outcomes. The third case illustrates variable drill duration; recorded attempts are not repeated physiological trials. The full development cohorts contain 37 and 46 mirror participants.

## The NFL question is conditional on protection opportunity

We start with Combine position **OL**, link NFL identifiers, and retain only the prospect's draft-year regular season. An eligible opportunity has an explicit non-null pressure-allowed label, an offensive C/G/T alignment on the possession team, and a play that was not nullified. We require at least 50 charted observations per player. Missing labels are undefined; they are never pressure-free plays. Explicit False labels include some zero-rusher-encounter observations, so this is a charted-protection estimand rather than a universal pass-rush denominator.

The 2023 cohort has 18 eligible players, 8,182 opportunities and 733 charted pressures; 2024 has 20 players, 7,983 opportunities and 711 pressures. Playing opportunity and charting create selection: the results do not describe all 83 prospects' unconditional NFL ability or career success.

![Figure 2](figure2_coverage.png)

**Figure 2.** The participant funnel shows why thousands of plays do not provide thousands of independent prospects. Complete Combine measurement does not guarantee a meaningful rookie role. The effective model/evaluation samples are 18 and 20 players.

## Timing-safe comparison and uncertainty

The conventional baseline uses Combine height, weight, forty time and ten-yard split. The full model adds only the three frozen tracking descriptors. NFL roster position was removed after an information-timing audit: it is not verified as a pre-draft input. The earlier role-conditioned result is reproduced in the executable appendix and explicitly retained as exploratory history.

Each model independently chooses ridge strength from 1, 10 or 100 using 2023 leave-one-player-out cross-validation. Imputation and scaling are fitted inside every fold. The selected baseline alpha is {selected_alpha_base:g}; the full-profile alpha is {selected_alpha_full:g}. A 2023 model-selection score is not independent final accuracy. Every grid score is retained, and 2024 remains repeatedly used exploratory selection data. We do not select the endpoint or model using the most favorable 2024 result.

| 2024 timing-safe model | Equal-player Brier loss |
|---|---:|
| Training-mean constant | {primary['constant']['player_equal_brier']:.6f} |
| Conventional Combine measurements | {bs['player_equal_brier']:.6f} |
| Conventional measurements plus mirror profile | {fs['player_equal_brier']:.6f} |

The interval above resamples players independently within the training and selection cohorts and refits preprocessing and models 250 times while fixing the selected alphas. A separate 5,000-draw player bootstrap fixes the trained models. Neither accounts fully for earlier family/alpha choices or shared teams and games. Leaving out each training player and reselecting alpha changes some full-profile predictions by as much as {100*primary['comparisons']['full_family']['max_LOO_prediction_shift']:.2f} percentage points, emphasizing small-sample fragility.

![Figure 3](figure3_decision.png)

**Figure 3.** Proper-loss improvement and player-level uncertainty on identical 2024 participants. Normalized recovery alone and the speed/excursion pair are explanatory subsets, not replacement headline models. The notebook also retains zero-encounter exclusion, C/I/IN/S-code restriction, and their intersection as distinct conditional endpoint sensitivities. None is promoted for a favorable score.

## What a scouting department can use now

Attach the extracted turn timestamps and trajectory to available film. Review whether a large relative recovery loss recurs at comparable incoming speed and excursion, and use the clip to formulate a technical question. This workflow preserves observable movement rather than compressing it into an unsupported talent grade. These pilot data do not justify automatically adjusting a rookie pressure forecast or recommending a roster decision.

The executable audit checks aliases, linkage uniqueness, coordinate bounds, chart-label missingness, duplicate timestamps, truncated trajectories and empty cohorts. Development-only smoothing and exit-threshold sensitivities are reported, without choosing thresholds by NFL outcome scores. One recorded attempt per player means computational robustness cannot establish physiological reliability.

A final 2025 branch is disabled. Only the team's selected candidate may receive a frozen pooled-2023/2024 fit and a single 2025 confirmation under unchanged features, alphas and eligibility rules. If that confirmation fails, the failure must be reported without selecting a different candidate. This article is an internal candidate artifact; no public notebook or competition submission has been published.
'''
assert len(article.split())<=2000
(WORK/'article.txt').write_text(article)
print('Article words:',len(article.split()),'; figures:3; article tables:1')
print('Output packet, article, config and candidate freeze manifest written.')
````

### C.7. Replay of the original frozen 2025 evaluation


````python
# This branch is deliberately disabled in every Round5 candidate.
# Root alone may authorize the ONE selected candidate, record code/config hashes,
# fill the frozen alphas from freeze_manifest.json, and then execute this branch.
if FINAL_EVALUATION_ENABLED:
    if FINAL_SELECTION_AUTHORIZATION != 'OL_selected_by_root':
        raise PermissionError('Root selection authorization required before any final outcome loading.')
    if FROZEN_BASELINE_ALPHA not in GRID or FROZEN_ENHANCED_ALPHA not in GRID:
        raise ValueError('Both frozen alphas must be explicitly supplied before final outcomes are opened.')
    if FROZEN_BASELINE_ALPHA != selected_alpha_base or FROZEN_ENHANCED_ALPHA != selected_alpha_full:
        raise ValueError('Final alphas differ from frozen development recipe; evaluation stopped before loading2025.')
    # Fit and serialize BOTH states before opening any final outcomes.
    pooled=prepared['eligible'].reset_index(drop=True)
    baseline_state=fit_state(pooled,BASE,FROZEN_BASELINE_ALPHA)
    enhanced_state=fit_state(pooled,BASE+FAMILY,FROZEN_ENHANCED_ALPHA)
    preholdout_state={'created_utc':datetime.datetime.now(datetime.timezone.utc).isoformat(),
        'code_fingerprint':CODE_FINGERPRINT,'baseline':baseline_state,'enhanced':enhanced_state,
        'training_mean_pressure':float(pooled.rate.mean()),'config':config}
    serialized_state=json.dumps(preholdout_state,indent=2)
    (WORK/'PREHOLDOUT_FITTED_STATES.json').write_text(serialized_state)
    preholdout_hash=hashlib.sha256(serialized_state.encode()).hexdigest()
    # Only now may the single selected candidate load its final cohort.
    final_inputs=load_inputs(DATA_DIR,years=(2025,),final_enabled=True)
    final_prepared=prepare_cohort(final_inputs,years=(2025,))
    heldout=final_prepared['eligible'].reset_index(drop=True)
    if len(heldout)<10:raise ValueError('Final cohort fails frozen10-player feasibility gate; no replacement model allowed.')
    yy=heldout.rate.to_numpy();n=heldout.n.to_numpy()
    bp=apply_state(baseline_state,heldout);ep=apply_state(enhanced_state,heldout)
    final_result={'constant':scores(yy,np.full(len(heldout),pooled.rate.mean()),n),'baseline':scores(yy,bp,n),'enhanced':scores(yy,ep,n),
                  'preholdout_fitted_state_sha256':preholdout_hash}
    final_result['funnel']=final_prepared['funnel'];final_result['status']='one_frozen_confirmation_no_model_switch'
    (WORK/'FINAL_2025_RESULT.json').write_text(json.dumps(final_result,indent=2))
    print('Frozen2025confirmation:',final_result)
else:
    print('Final2025branch NOT EXECUTED. No2025 outcomes evaluated.')
````

### C.8. Original player-level uncertainty

## D. Count-aware benchmark and empirical intervals

All predictions omit the evaluated player's outcomes. Exact shared assignments are read from verified fixtures; modeling cells are unchanged.


````python
import os, csv, io, json, hashlib
from pathlib import Path
import numpy as np
import pandas as pd
from scipy.signal import savgol_filter
from scipy.integrate import trapezoid
from scipy.stats import betabinom
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
BASE=['combine_height','combine_weight','forty','ten_yd_split']
FAMILY=['normalized_deficit','incoming_speed','lateral_amplitude']
required=['players.csv','combine_results.csv','combine_tracking.csv','games.csv','player_play.csv']
explicit=os.environ.get('NFL_DATA_DIR')
if explicit: DATA_DIR=Path(explicit)
else:
    candidates=[p.parent for p in Path('/kaggle/input').glob('**/players.csv') if all((p.parent/f).is_file() for f in required)]
    if len(candidates)!=1: raise FileNotFoundError('Set NFL_DATA_DIR to the official CSV directory.')
    DATA_DIR=candidates[0]
if not all((DATA_DIR/f).is_file() for f in required): raise FileNotFoundError('Official inputs incomplete.')
WORK=Path.cwd()/'hierarchical_results';WORK.mkdir(exist_ok=True)
print('Development cohorts2023/2024 only; original2025 result retained as historical context.')
````

````python
"""Exploratory, count-aware OL prediction; no test-player outcome enters a prediction.

Run with standardized CSV columns player_id, cohort, y, n, numerical feature columns,
and a JSON fold map {"outer": {"player_id": fold}, "inner_seed": 72027}.
Outer folds must be shared/frozen before model comparison. Every model uses the same
rows and out-of-player folds. Fits are empirical Bayes/MAP, not full posterior Bayes.
"""

import argparse

import hashlib

import json

from pathlib import Path

import numpy as np

import pandas as pd

from scipy.optimize import minimize

from scipy.special import betaln, digamma, expit, gammaln, logit

def scalar(v):
    if isinstance(v, np.generic):
        return v.item()
    raise TypeError(type(v).__name__)

class FoldPreprocessor:
    def fit(self, frame, columns, add_missing=True):
        self.columns = list(columns)
        raw = frame[self.columns].to_numpy(float) if self.columns else np.zeros((len(frame), 0))
        self.med = np.array([np.nanmedian(x) if np.isfinite(x).any() else 0 for x in raw.T])
        self.missing_columns = np.where(~np.isfinite(raw).all(axis=0))[0] if add_missing else np.array([], dtype=int)
        x = self._impute(raw)
        self.mean = x.mean(axis=0)
        self.std = x.std(axis=0)
        self.std[self.std < 1e-8] = 1
        return self

    def _impute(self, raw):
        miss = ~np.isfinite(raw)
        imp = np.where(miss, self.med[None, :], raw)
        return np.column_stack([imp, miss[:, self.missing_columns].astype(float)])

    def transform(self, frame):
        raw = frame[self.columns].to_numpy(float) if self.columns else np.zeros((len(frame), 0))
        return (self._impute(raw) - self.mean) / self.std

class BetaBinomialRegressor:
    """y_i|p_i~Binomial(n_i,p_i); p_i~Beta(mu_i*phi,(1-mu_i)*phi).

    mu_i=logistic(intercept + standardized covariates beta). Normal ridge priors
    shrink the slopes. log(phi)~N(log(40),2^2) prevents unstable boundary estimates.
    New-player mean is mu_i; training-player posterior summaries are descriptive
    only and are never scored as held-out forecasts.
    """
    def __init__(self, penalty=10.0, family="beta_binomial"):
        self.penalty, self.family = float(penalty), family

    def fit(self, x, y, n):
        x, y, n = np.asarray(x, float), np.asarray(y, float), np.asarray(n, float)
        design = np.column_stack([np.ones(len(y)), x])
        init = np.zeros(design.shape[1] + (self.family == "beta_binomial"))
        init[0] = logit(np.clip(y.sum() / n.sum(), 1e-4, 1 - 1e-4))
        if self.family == "beta_binomial":
            init[-1] = np.log(40.0)
        logchoose = gammaln(n + 1) - gammaln(y + 1) - gammaln(n - y + 1)

        def objective(theta):
            coef = theta[:design.shape[1]]
            mu = np.clip(expit(design @ coef), 1e-10, 1 - 1e-10)
            if self.family == "beta_binomial":
                phi = np.exp(theta[-1])
                a, b = mu * phi, (1 - mu) * phi
                loglike = logchoose + betaln(y + a, n - y + b) - betaln(a, b)
                da = digamma(y + a) - digamma(a)
                db = digamma(n - y + b) - digamma(b)
                d_eta = phi * (da - db) * mu * (1 - mu)
                d_logphi = phi * (mu * da + (1 - mu) * db + digamma(phi) - digamma(n + phi))
                prior = 0.5 * ((theta[-1] - np.log(40)) / 2) ** 2
                gradphi = -d_logphi.sum() + (theta[-1] - np.log(40)) / 4
            elif self.family == "binomial":
                loglike = logchoose + y * np.log(mu) + (n - y) * np.log1p(-mu)
                d_eta, prior = y - n * mu, 0.0
            else:
                raise ValueError(self.family)
            grad = -(design.T @ d_eta)
            grad[1:] += self.penalty * coef[1:]
            val = -loglike.sum() + 0.5 * self.penalty * np.square(coef[1:]).sum() + prior
            if self.family == "beta_binomial":
                grad = np.r_[grad, gradphi]
            return float(val), grad

        bounds = [(-12, 12)] * design.shape[1]
        if self.family == "beta_binomial":
            bounds.append((np.log(0.05), np.log(100000)))
        result = minimize(objective, init, jac=True, method="L-BFGS-B", bounds=bounds,
                          options={"maxiter": 1000, "maxls": 50, "ftol": 1e-9, "gtol": 1e-6})
        if not result.success or not np.isfinite(result.fun):
            raise RuntimeError(f"Fit failed: {result.message}")
        self.coef = result.x[:design.shape[1]]
        self.phi = float(np.exp(result.x[-1])) if self.family == "beta_binomial" else None
        self.convergence = {"success": bool(result.success), "iterations": int(result.nit),
                            "objective": float(result.fun), "message": str(result.message)}
        return self

    def predict(self, x):
        return np.clip(expit(self.coef[0] + np.asarray(x) @ self.coef[1:]), 1e-8, 1 - 1e-8)

def metrics(y, n, p, phi=None):
    y, n, p = map(lambda z: np.asarray(z, float), (y, n, p))
    rate = y / n
    brier = rate * (1 - p) ** 2 + (1 - rate) * p ** 2
    ll = -rate * np.log(p) - (1 - rate) * np.log1p(-p)
    result = {"equal_player_brier": float(brier.mean()),
              "opportunity_weighted_brier": float(np.average(brier, weights=n)),
              "equal_player_logloss": float(ll.mean()),
              "opportunity_weighted_logloss": float(np.average(ll, weights=n)),
              "rate_mse": float(np.square(rate - p).mean()),
              "rate_mae": float(np.abs(rate - p).mean()),
              "predicted_mean_rate": float(p.mean()), "observed_mean_rate": float(rate.mean())}
    if phi is not None:
        a, b = p * phi, (1 - p) * phi
        logchoose = gammaln(n + 1) - gammaln(y + 1) - gammaln(n - y + 1)
        result["equal_player_count_nll"] = float(-(logchoose + betaln(y + a, n - y + b) - betaln(a, b)).mean())
    return result

def inner_partition(frame, nfolds=3, seed=87007):
    """Outcome-independent, cohort-balanced ID partition of training players."""
    folds = {}
    for cohort, rows in frame.groupby("cohort", sort=True):
        ids = rows.player_id.astype(str).tolist()
        ids.sort(key=lambda z: hashlib.sha256(f"{seed}|{cohort}|{z}".encode()).hexdigest())
        folds.update({pid: i % min(nfolds, len(ids)) for i, pid in enumerate(ids)})
    return np.array([folds[str(pid)] for pid in frame.player_id])

def fit_predict(train, test, columns, penalty, family):
    pre = FoldPreprocessor().fit(train, columns, add_missing=family not in ("ridge", "mean"))
    if family in ("ridge", "mean"):
        x, rate = pre.transform(train), (train.y / train.n).to_numpy()
        model = RateRidgeRegressor(penalty).fit(x, rate)
    else:
        model = BetaBinomialRegressor(penalty, family).fit(pre.transform(train), train.y, train.n)
    return model.predict(pre.transform(test)), model, pre

class RateRidgeRegressor:
    """Historical equal-player ridge comparator with train-only median imputation."""
    def __init__(self, penalty):
        self.penalty, self.phi = penalty, None

    def fit(self, x, rate):
        intercept = float(rate.mean())
        slope = np.linalg.solve(x.T @ x + self.penalty * np.eye(x.shape[1]), x.T @ (rate - intercept))
        self.coef = np.r_[intercept, slope]
        self.convergence = {"success": True, "iterations": 1, "message": "closed-form ridge"}
        return self

    def predict(self, x):
        return np.clip(self.coef[0] + x @ self.coef[1:], 0.001, 0.999)

def nested_cv(frame, outer, feature_sets, penalties=(0.1, 1, 10, 100), seed=87007, outer_records=None):
    outputs, fits = [], []
    for fold in sorted(set(outer)):
        train, test = frame.loc[outer != fold].copy(), frame.loc[outer == fold].copy()
        if outer_records is None:
            infold = inner_partition(train, seed=seed)
        else:
            record = next(z for z in outer_records if int(z['fold']) == int(fold))
            train_ids, test_ids = set(train.player_id.astype(str)), set(test.player_id.astype(str))
            if train_ids != {str(z) for z in record['train_ids']} or test_ids != {str(z) for z in record['test_ids']}:
                raise ValueError('Outer records do not match current player population.')
            inner = {}
            for k, rec in enumerate(record['inner_folds']):
                itr, ite = {str(z) for z in rec['train_ids']}, {str(z) for z in rec['test_ids']}
                if itr & ite or itr | ite != train_ids:
                    raise ValueError('Inner train/test split fails player isolation or full coverage.')
                for pid in ite:
                    if pid in inner:
                        raise ValueError('Player appears in multiple inner evaluation folds.')
                    inner[pid] = k
            if set(inner) != train_ids:
                raise ValueError('Inner evaluation folds do not cover each outer training player exactly once.')
            infold = np.array([inner[str(pid)] for pid in train.player_id])
        for label, spec in feature_sets.items():
            columns, family = spec["columns"], spec.get("family", "beta_binomial")
            if "fixed_penalty" in spec:
                selected, curve = float(spec["fixed_penalty"]), []
            elif not columns:
                selected, curve = 0.0, []
            else:
                curve = []
                for penalty in penalties:
                    pred = np.full(len(train), np.nan)
                    for inner in sorted(set(infold)):
                        fitmask = infold != inner
                        p, _, _ = fit_predict(train.loc[fitmask], train.loc[~fitmask], columns, penalty, family)
                        pred[~fitmask] = p
                    loss = metrics(train.y, train.n, pred)["equal_player_brier"]
                    curve.append({"penalty": penalty, "inner_brier": loss})
                best = min(v["inner_brier"] for v in curve)
                # Exact ties prefer stronger shrinkage; no outer outcome guides this.
                selected = max(v["penalty"] for v in curve if v["inner_brier"] <= best + 1e-12)
            p, model, pre = fit_predict(train, test, columns, selected, family)
            for ix, probability in zip(test.index, p):
                outputs.append({"player_id": str(frame.loc[ix, "player_id"]), "cohort": int(frame.loc[ix, "cohort"]),
                                "model": label, "fold": int(fold), "y": int(frame.loc[ix, "y"]),
                                "n": int(frame.loc[ix, "n"]), "prediction": float(probability),
                                "phi": model.phi, "penalty": selected})
            fits.append({"model": label, "fold": int(fold), "train_players": len(train), "test_players": len(test),
                         "train_ids": train.player_id.tolist(), "test_ids": test.player_id.tolist(),
                         "inner_folds": [{"train_ids": train.loc[infold != k, 'player_id'].tolist(),
                                          "test_ids": train.loc[infold == k, 'player_id'].tolist()}
                                         for k in sorted(set(infold))],
                         "selected_penalty": selected, "inner_curve": curve, "phi": model.phi,
                         "coef": model.coef.tolist(), "columns": columns,
                         "missing_indicator_columns": [columns[i] for i in pre.missing_columns],
                         "convergence": model.convergence})
    pred = pd.DataFrame(outputs)
    summary = {name: metrics(rows.y, rows.n, rows.prediction,
                            rows.phi if rows.phi.notna().all() else None)
               for name, rows in pred.groupby("model")}
    return pred, fits, summary
````

````python
def read_selected(path,predicate):
    # Record selection examines only cohort/game/player keys. Discarded NFL
    # outcomes never enter a typed DataFrame, analysis, model, or output.
    buf=io.StringIO()
    with path.open(newline='') as f:
        reader=csv.reader(f);header=next(reader);index={k:i for i,k in enumerate(header)}
        writer=csv.writer(buf);writer.writerow(header)
        for row in reader:
            if len(row)!=len(header):raise ValueError('Malformed CSV record in '+path.name)
            if predicate(row,index):writer.writerow(row)
    buf.seek(0)
    return pd.read_csv(buf,low_memory=False)

def load_inputs(data_dir,years=(2023,2024),final_enabled=False):
    years=set(int(y) for y in years)
    if not years <= {2023,2024,2025}:raise ValueError('Unexpected cohort year.')
    if 2025 in years and not final_enabled:
        raise PermissionError('2025 loader blocked BEFORE any file is opened.')
    allowed={str(y) for y in years};d=Path(data_dir)
    players=read_selected(d/'players.csv',lambda r,k:r[k['draft_year']] in allowed)
    results=read_selected(d/'combine_results.csv',lambda r,k:r[k['draft_year']] in allowed)
    if not players.nfl_id.is_unique or results.duplicated(['draft_year','nfl_id']).any():
        raise ValueError('Non-unique player/Combine linkage key.')
    ol=results[results.combine_position.eq('OL')].copy()
    ids={str(int(x)) for x in ol.nfl_id};draft=dict(zip(players.nfl_id.astype(str),players.draft_year.astype(int)))
    games=read_selected(d/'games.csv',lambda r,k:r[k['season']] in allowed)
    if not games.game_id.is_unique:raise ValueError('Duplicate game identifiers.')
    reg=games[games.season_type.eq('REG')]
    gameyear=dict(zip(reg.game_id.astype(str),reg.season.astype(int)))
    tracking=read_selected(d/'combine_tracking.csv',lambda r,k:
        r[k['draft_year']] in allowed and r[k['nfl_id']] in ids and
        r[k['entity_type']]=='PLAYER' and r[k['drill_name']].replace('-','_')=='PASS_PRO_MIRROR_DRILL')
    def rookie_record(r,k):
        game=r[k['game_id']];nid=r[k['nfl_id']]
        return game in gameyear and nid in ids and gameyear[game]==draft.get(nid)
    player_play=read_selected(d/'player_play.csv',rookie_record)
    assert set(players.draft_year)<=years and set(results.draft_year)<=years
    assert set(tracking.draft_year)<=years and set(games.season)<=years
    for a in years:
        for b in years:
            if a<b:assert not set(players[players.draft_year.eq(a)].nfl_id)&set(players[players.draft_year.eq(b)].nfl_id)
    if player_play.duplicated(['game_id','play_id','nfl_id']).any():raise ValueError('Duplicate player-play key.')
    return {'players':players,'results':results,'games':games,'tracking':tracking,'player_play':player_play,
            'audit':{'years':sorted(years),'accepted_rows':{'players':len(players),'combine_results':len(results),
                     'combine_tracking_mirror':len(tracking),'games':len(games),'rookie_REG_player_play':len(player_play)},
                     'drill_aliases':sorted(tracking.drill_name.unique().tolist()),'initial_OL_cohort_source':'combine_results.combine_position == OL',
                     'outcomes_selected_by_keys_before_typing':True,'career_file_opened':False}}

def extract(t, window=5, exit_threshold=1.0):
    rows=[]; events=[]; examples={}
    for key,g in t.groupby(['draft_year','event_id','nfl_id','attempt'],sort=True):
        g=g.sort_values('time').copy()
        tm=pd.to_datetime(g.time); sec=(tm-tm.iloc[0]).dt.total_seconds().to_numpy()
        x=g.x.to_numpy(); y=g.y.to_numpy()
        start=np.median(x[:min(10,len(x))])
        exits=np.flatnonzero((np.abs(x-start)>exit_threshold)&(sec>0.5))
        stop=int(exits[0]) if len(exits) else len(x)
        quality='ok'
        if stop<15 or np.any(np.abs(np.diff(sec[:stop])-.1)>.015): quality='short_or_non_10hz'
        if np.isnan(x).any() or np.isnan(y).any(): quality='missing_coordinates'
        rev=[]; deficits=[]
        if quality=='ok':
            sy=savgol_filter(y[:stop],window,2)
            vel=np.gradient(sy,sec[:stop])
            # Zero-speed frames are real at turns: bridge near-zero intervals,
            # rather than requiring adjacent nonzero signs (which misses minima).
            signs=np.where(np.abs(vel)>.05,np.sign(vel),0)
            nz=np.flatnonzero(signs)
            changes=[int(round((i+j)/2)) for i,j in zip(nz[:-1],nz[1:])
                     if signs[i]*signs[j]<0 and j-i<=10]
            last=-100
            for i in changes:
                if i<5 or i+5>=stop or i-last<5: continue
                pre=np.median(vel[i-5:i-1]); post=np.median(vel[i+1:i+5])
                if abs(pre)<.5 or abs(post)<.5 or pre*post>=0: continue
                new=np.sign(post)*vel[i:i+6]
                deficit=float(trapezoid(np.maximum(0,abs(pre)-new),sec[i:i+6]-sec[i]))
                rev.append(i); deficits.append(deficit); last=i
                events.append({'draft_year':key[0],'event_id':key[1],'nfl_id':key[2],'attempt':key[3],
                               'reversal_time_seconds':float(sec[i]),'incoming_speed_yards_s':float(abs(pre)),
                               'recovery_deficit_yards':deficit})
        if len(deficits)<2 and quality=='ok': quality='fewer_than_two_usable_reversals'
        rows.append({'draft_year':key[0],'event_id':key[1],'nfl_id':key[2],'attempt':key[3],
                     'frames':len(g),'mirror_frames':stop,'usable_reversals':len(deficits),
                     'quality':quality,'recovery_deficit':float(np.percentile(deficits,75)) if quality=='ok' else np.nan,
                     'x_start':float(start),'x_range':float(np.ptp(x)),'y_range':float(np.ptp(y))})
        examples[key]={'sec':sec,'x':x,'y':y,'stop':stop,'rev':rev,'quality':quality}
    return pd.DataFrame(rows),pd.DataFrame(events),examples

def extract_mirror(t):
    keys=['draft_year','event_id','nfl_id','attempt','time']
    if t.duplicated(keys).any():raise ValueError('Duplicate tracking timestamp key; extraction refused.')
    if t.empty:raise ValueError('No mirror trajectories; metric is undefined, not zero.')
    if not t.x.between(0,120).all() or not t.y.between(0,53.5).all():
        raise ValueError('Coordinate bounds incompatible with documented field-yard units.')
    return extract(t)

def build_features(t):
    attempts,events,examples=extract_mirror(t)
    keys=['draft_year','event_id','nfl_id','attempt']
    good=attempts[attempts.quality.eq('ok')]
    if good.empty or events.empty:raise ValueError('No usable reversals; metric is undefined.')
    e=events.merge(good[keys],on=keys,validate='many_to_one')
    e['normalized_deficit']=e.recovery_deficit_yards/e.incoming_speed_yards_s
    a=e.groupby(keys,as_index=False).agg(normalized_deficit=('normalized_deficit',lambda z:np.percentile(z,75)),
        incoming_speed=('incoming_speed_yards_s','median'))
    task=[]
    for _,row in good.iterrows():
        key=(int(row.draft_year),row.event_id,int(row.nfl_id),int(row.attempt));ex=examples[key]
        yy=ex['y'][:ex['stop']]
        task.append(dict(zip(keys,key),lateral_amplitude=float(np.percentile(yy,90)-np.percentile(yy,10)),
                         phase_seconds=float(ex['sec'][ex['stop']-1])))
    z=good.merge(a,on=keys,validate='one_to_one').merge(pd.DataFrame(task),on=keys,validate='one_to_one')
    features=z.groupby(['draft_year','nfl_id'],as_index=False).agg(
        normalized_deficit=('normalized_deficit','median'),incoming_speed=('incoming_speed','median'),
        lateral_amplitude=('lateral_amplitude','median'),raw_deficit=('recovery_deficit','median'),
        usable_reversals=('usable_reversals','sum'),usable_attempts=('attempt','size'),phase_seconds=('phase_seconds','median'))
    return features,attempts,events,examples

def prepare_cohort(inputs,years=(2023,2024),endpoint='charted_all'):
    features,attempts,events,examples=build_features(inputs['tracking'][inputs['tracking'].draft_year.isin(years)])
    c=inputs['results'];p=inputs['players'];g=inputs['games']
    c=c[c.draft_year.isin(years)&c.combine_position.eq('OL')]
    base=p[p.draft_year.isin(years)].merge(c[['draft_year','nfl_id']+BASE],on=['draft_year','nfl_id'],validate='one_to_one')
    base=base.merge(features,on=['draft_year','nfl_id'],validate='one_to_one')
    # NFL position is retained ONLY for explicitly labeled historical sensitivity.
    base['historical_is_tackle']=base.nfl_position.eq('T').astype(int)
    q=inputs['player_play'].merge(p[['nfl_id','draft_year']],on='nfl_id',validate='many_to_one').merge(
        g[['game_id','season','season_type']],on='game_id',validate='many_to_one')
    q=q[q.draft_year.isin(years)&q.season.eq(q.draft_year)&q.season_type.eq('REG')].copy()
    good=q.pressure_allowed.notna()&q.lined_up_position.isin(['C','G','T'])&q.team_abbr.eq(q.possession_team)&q.play_nullified_by_penalty.eq('N')
    observed=q[good].copy()
    labels=observed.pressure_allowed.map(lambda z:str(z).lower())
    if not set(labels)<= {'true','false'}:raise ValueError('Unknown charted pressure boolean; no coercion to zero allowed.')
    observed['pressure']=labels.eq('true').astype(int)
    if endpoint=='encountered_at_least_one':observed=observed[observed.pass_rushers_encountered.ge(1)]
    elif endpoint=='pass_results_C_I_IN_S':observed=observed[observed.pass_result.isin(['C','I','IN','S'])]
    elif endpoint=='encountered_and_C_I_IN_S':observed=observed[observed.pass_rushers_encountered.ge(1)&observed.pass_result.isin(['C','I','IN','S'])]
    elif endpoint!='charted_all':raise ValueError('Unrecognized endpoint; no silent substitute.')
    outcome=observed.groupby(['draft_year','nfl_id'],as_index=False).agg(n=('pressure','size'),k=('pressure','sum'))
    outcome['rate']=outcome.k/outcome.n
    allplayers=base.merge(outcome,on=['draft_year','nfl_id'],how='left',validate='one_to_one')
    eligible=allplayers[allplayers.n.ge(50)].copy()
    funnel={str(year):{'Combine_OL_participants':int(c.draft_year.eq(year).sum()),
        'mirror_participants':int(attempts[attempts.draft_year.eq(year)].nfl_id.nunique()),
        'usable_feature_players':int(base.draft_year.eq(year).sum()),'any_rookie_REG_players':int(q[q.draft_year.eq(year)].nfl_id.nunique()),
        'observed_endpoint_players':int((allplayers.draft_year.eq(year)&allplayers.n.notna()).sum()),
        'eligible_players_min50':int(eligible.draft_year.eq(year).sum()),
        'eligible_opportunities':int(eligible.loc[eligible.draft_year.eq(year),'n'].sum()),
        'eligible_pressures':int(eligible.loc[eligible.draft_year.eq(year),'k'].sum()),
        'missing_charted_pressure_rows':int((q.draft_year.eq(year)&q.pressure_allowed.isna()).sum()),
        'observed_zero_rusher_rows':int((observed.draft_year.eq(year)&observed.pass_rushers_encountered.eq(0)).sum())}
        for year in years}
    return {'eligible':eligible,'all_players':allplayers,'funnel':funnel,'attempts':attempts,'events':events,
            'examples':examples,'features':features,'rookie_rows':q,'observed_rows':observed}

def default_specs():
    return {
        'constant_equal_player': {'columns': [], 'family': 'mean'},
        'original_conventional_fixed10': {'columns': BASE, 'family': 'ridge', 'fixed_penalty': 10},
        'original_full_fixed1': {'columns': BASE + FAMILY, 'family': 'ridge', 'fixed_penalty': 1},
        'nested_conventional_ridge': {'columns': BASE, 'family': 'ridge'},
        'nested_full_ridge': {'columns': BASE + FAMILY, 'family': 'ridge'},
        'beta_binomial_constant': {'columns': [], 'family': 'beta_binomial'},
        'beta_binomial_conventional': {'columns': BASE, 'family': 'beta_binomial'},
        'beta_binomial_full': {'columns': BASE + FAMILY, 'family': 'beta_binomial'},
        'binomial_full': {'columns': BASE + FAMILY, 'family': 'binomial'},
    }
````

````python
def tune_fit(train, test, record, columns):
    """No omitted-player outcome may influence preprocessing or chosen penalty."""
    ids = set(train.player_id)
    trials = []
    for penalty in [.1, 1, 10, 100]:
        pred = pd.Series(index=train.index, dtype=float)
        for inner in record['inner_folds']:
            itr = {str(z) for z in inner['train_ids']} & ids
            ite = {str(z) for z in inner['test_ids']} & ids
            a, b = train[train.player_id.isin(itr)], train[train.player_id.isin(ite)]
            if not len(a) or not len(b):
                raise ValueError('Empty inner split after jackknife restriction.')
            p, _, _ = fit_predict(a, b, columns, penalty, 'beta_binomial')
            pred.loc[b.index] = p
        if pred.isna().any():
            raise ValueError('Inner predictions incomplete.')
        trials.append((metrics(train.y, train.n, pred)['equal_player_brier'], penalty))
    best = min(t[0] for t in trials)
    penalty = max(t[1] for t in trials if t[0] <= best + 1e-12)
    p, model, pre = fit_predict(train, test, columns, penalty, 'beta_binomial')
    return p, model, pre, penalty

def finite_interval_quantiles(lower, upper, alpha=.05):
    """Paper-inspired finite-sample order statistics with tail level alpha=.05.

    Studentization is our adaptation. Only empirical coverage is claimed here;
    neither the paper's iid guarantee nor its nominal target is asserted for NFL.
    """
    m = lower.shape[0]
    lo = int(np.floor(alpha * (m + 1)))
    hi = int(np.ceil((1 - alpha) * (m + 1)))
    lower_bound = np.full(lower.shape[1], -np.inf) if lo < 1 else np.sort(lower, axis=0)[lo - 1]
    upper_bound = np.full(upper.shape[1], np.inf) if hi > m else np.sort(upper, axis=0)[hi - 1]
    return np.clip(lower_bound, 0, 1), np.clip(upper_bound, 0, 1)

def jackknife_plus(frame, folds, columns, label):
    outputs, audit = [], []
    for record in folds['outer_folds']:
        train = frame[frame.player_id.isin({str(z) for z in record['train_ids']})]
        test = frame[frame.player_id.isin({str(z) for z in record['test_ids']})]
        minus, plus, choices = [], [], []
        for ix, omitted in train.iterrows():
            fit = train.drop(index=ix)
            combined = pd.concat([train.loc[[ix]], test], ignore_index=True)
            if columns:
                p, model, _, penalty = tune_fit(fit, combined, record, columns)
            else:
                p, model, _ = fit_predict(fit, combined, [], 0, 'beta_binomial')
                penalty = 0
            # Count noise scale is conditional on recorded/design denominator,
            # never used as a predictor of mean; known opportunity counts affect interval width.
            variance = p * (1 - p) * (combined.n.to_numpy() + model.phi) / (combined.n.to_numpy() * (1 + model.phi))
            scale = np.sqrt(np.maximum(variance, 1e-12))
            score = abs(omitted.y / omitted.n - p[0]) / scale[0]
            minus.append(p[1:] - score * scale[1:]); plus.append(p[1:] + score * scale[1:])
            choices.append({'omitted_player_id': omitted.player_id, 'penalty': penalty,
                            'phi': model.phi, 'calibration_score': float(score),
                            'training_ids': fit.player_id.tolist()})
        lower, upper = finite_interval_quantiles(np.asarray(minus), np.asarray(plus))
        if columns:
            point, model, _, penalty = tune_fit(train, test, record, columns)
        else:
            point, model, _ = fit_predict(train, test, [], 0, 'beta_binomial'); penalty = 0
        rawlo = betabinom.ppf(.05, test.n, point * model.phi, (1 - point) * model.phi) / test.n
        rawhi = betabinom.ppf(.95, test.n, point * model.phi, (1 - point) * model.phi) / test.n
        for (_, row), p, lo, hi, rl, rh in zip(test.iterrows(), point, lower, upper, rawlo, rawhi):
            outputs.append({'player_id': row.player_id, 'cohort': int(row.cohort), 'fold': record['fold'],
                            'model': label, 'y': int(row.y), 'n': int(row.n), 'prediction': float(p),
                            'jackknife_plus_low': float(lo), 'jackknife_plus_high': float(hi),
                            'conditional90_low': float(rl), 'conditional90_high': float(rh)})
        audit.append({'fold': record['fold'], 'model': label, 'outer_penalty': penalty,
                      'jackknife_fits': choices})
        print(f'Completed {label} outer fold{record["fold"]}; {len(train)} nested jackknife omissions', flush=True)
    return pd.DataFrame(outputs), audit

def plot_count_aware(primary, intervals, output):
    plt.rcParams.update({'font.size':10,'axes.spines.top':False,'axes.spines.right':False})
    wide = primary.pivot(index='player_id', columns='model', values='prediction')
    target = primary[primary.model.eq('constant_equal_player')].set_index('player_id').loc[wide.index]
    rate = (target.y/target.n).to_numpy()
    rng=np.random.default_rng(87007);samples=rng.integers(0,len(wide),size=(5000,len(wide)))
    models=['nested_conventional_ridge','beta_binomial_conventional','beta_binomial_full','nested_full_ridge']
    labels=['Conventional ridge','Conventional beta-binomial','Mirror beta-binomial','Mirror ridge']
    means=[];limits=[]
    for model in models:
        delta=(rate-wide.constant_equal_player.to_numpy())**2-(rate-wide[model].to_numpy())**2
        means.append(delta.mean());limits.append(np.quantile(delta[samples].mean(1),[.025,.975]))
    means=np.array(means);limits=np.array(limits)
    fig,ax=plt.subplots(1,3,figsize=(15,4.5),layout='constrained')
    positions=np.arange(4)
    ax[0].errorbar(means*10000,positions,xerr=np.vstack([(means-limits[:,0])*10000,(limits[:,1]-means)*10000]),
                  fmt='o',color='#176876',capsize=3)
    ax[0].axvline(0,color='#555',linewidth=1)
    ax[0].set_yticks(positions,labels)
    ax[0].set(xlabel='Constant minus model Brier ×10,000',title='A. Point gains remain unresolved')
    ax[0].text(.01,-.21,'95% fixed-OOF player bootstrap; no refitting',transform=ax[0].transAxes,fontsize=8)
    rows=[]
    for label,z in intervals.groupby('model'):
        y=z.y/z.n
        rows.append({'model':label,'raw':((y>=z.conditional90_low)&(y<=z.conditional90_high)).mean(),
                     'resampling':((y>=z.jackknife_plus_low)&(y<=z.jackknife_plus_high)).mean(),
                     'raw_width':(z.conditional90_high-z.conditional90_low).median()*100,
                     'resampling_width':(z.jackknife_plus_high-z.jackknife_plus_low).median()*100})
    rows=pd.DataFrame(rows).set_index('model').loc[['beta_binomial_constant','beta_binomial_full']]
    x=np.arange(2);w=.34
    bars1=ax[1].bar(x-w/2,rows.raw*100,w,color='#BD603B',label='Beta-binomial central90%')
    bars2=ax[1].bar(x+w/2,rows.resampling*100,w,color='#176876',label='Jackknife+ style, α=0.05')
    for bars in [bars1,bars2]:
        for b in bars:
            count=int(round(b.get_height()/100*38));ax[1].text(b.get_x()+b.get_width()/2,b.get_height()+1,f'{count}/38',ha='center',fontsize=9)
    ax[1].set_xticks(x,['Constant prior','Mirror prior']);ax[1].set(ylim=(0,112),ylabel='Empirical player coverage (%)',title='B. Interval assumptions need a test')
    ax[1].legend(fontsize=8,loc='lower center',bbox_to_anchor=(.5,-.31),frameon=False)
    for i,(_,r) in enumerate(rows.iterrows()):
        ax[2].plot([r.raw_width,r.resampling_width],[i,i],color='#999',linewidth=2)
        ax[2].scatter(r.raw_width,i,color='#BD603B',s=55);ax[2].scatter(r.resampling_width,i,color='#176876',s=55)
        ax[2].annotate(f'{r.raw_width:.2f}',(r.raw_width,i),xytext=(0,-18),textcoords='offset points',ha='center',fontsize=9)
        ax[2].annotate(f'{r.resampling_width:.2f}',(r.resampling_width,i),xytext=(0,12),textcoords='offset points',ha='center',fontsize=9)
    ax[2].set_yticks(x,['Constant prior','Mirror prior']);ax[2].set(xlim=(4,14),ylim=(-.6,1.6),xlabel='Median interval width: pressures per100',title='C. Wider intervals cost precision')
    fig.suptitle('38 rookies,16,165 charted observations | 2023/24 exploratory development',fontsize=14)
    fig.text(.5,-.11,'Different interval constructions; no equal-nominal-level comparison or NFL coverage guarantee.\n'
             'All703 mirror-interval pairs overlap: abstain from Combine-derived ordering. Prior2025 failure remains unchanged.',
             ha='center',fontsize=9)
    fig.savefig(output,dpi=180,bbox_inches='tight')
    plt.close(fig)
````

````python
intervals=[];jackknife_audit=[]
for label,columns in [('beta_binomial_constant',[]),('beta_binomial_full',BASE+FAMILY)]:
    predictions,audit=jackknife_plus(frame,FOLDS,columns,label)
    intervals.append(predictions);jackknife_audit.extend(audit)
intervals=pd.concat(intervals,ignore_index=True)
intervals.to_csv(WORK/'jackknife_plus_predictions.csv',index=False)
(WORK/'jackknife_fit_audit.json').write_text(json.dumps(jackknife_audit,indent=2,default=scalar))
coverage={}
for label,rows in intervals.groupby('model'):
    rate=rows.y/rows.n
    coverage[label]={'n':len(rows),'central90_count_coverage':float(((rate>=rows.conditional90_low)&(rate<=rows.conditional90_high)).mean()),
                     'jackknife_style_alpha05_empirical_coverage':float(((rate>=rows.jackknife_plus_low)&(rate<=rows.jackknife_plus_high)).mean()),
                     'median_central90_width':float((rows.conditional90_high-rows.conditional90_low).median()),
                     'median_jackknife_style_alpha05_width':float((rows.jackknife_plus_high-rows.jackknife_plus_low).median())}
(WORK/'interval_coverage.json').write_text(json.dumps(coverage,indent=2))
print(json.dumps(coverage,indent=2))
plot_count_aware(primary,intervals,WORK/'figure_count_aware_loss_coverage.png')
try:
    from IPython.display import display,Image
    display(Image(filename=str(WORK/'figure_count_aware_loss_coverage.png')))
except ImportError:
    print('Saved figure:',WORK/'figure_count_aware_loss_coverage.png')
````

````python
functional_required=['players.csv','combine_results.csv','combine_tracking.csv','games.csv','player_play.csv']
if os.environ.get('NFL_DATA_DIR'):
    functional_data=Path(os.environ['NFL_DATA_DIR']).resolve()
    if not all((functional_data/f).is_file() for f in functional_required):
        raise FileNotFoundError('NFL_DATA_DIR lacks one of the five official CSVs.')
else:
    functional_roots=sorted({p.parent for p in Path('/kaggle/input').glob('**/players.csv')
                            if all((p.parent/f).is_file() for f in functional_required)})
    if len(functional_roots)!=1:
        raise RuntimeError(f'Expected exactly one official input root; found {len(functional_roots)}.')
    functional_data=functional_roots[0].resolve()
functional_output=Path('functional_results').resolve()
functional_output.mkdir(exist_ok=True)
functional_commands=[
    [sys.executable,str(functional_code/'run_canonical38.py'),'--raw-dir',str(functional_data),
     '--fold-map',str(functional_code/'common_folds_OL.json'),'--output-dir',str(functional_output)],
    [sys.executable,str(functional_code/'robustness.py'),'--raw-dir',str(functional_data),
     '--fold-map',str(functional_code/'common_folds_OL_37_corrected.json'),
     '--output-dir',str(functional_output/'robustness37')],
    [sys.executable,str(functional_code/'build_scouting_cards.py'),'--raw-dir',str(functional_data),
     '--output-dir',str(functional_output)]
]
for functional_command in functional_commands:
    subprocess.run(functional_command,check=True)
# Public output retains aggregate results, figures and reproducible source.
# No participant-level outcome/prediction or complete tracking-card rows remain.
functional_removed=[]
for functional_name in ['OOF_PREDICTIONS.json','FOLD_RECORDS.json','SCOUTING_CARDS.json','EXAMPLE_FILM_CARD.json']:
    for functional_path in list(functional_output.rglob(functional_name)):
        functional_path.unlink(missing_ok=True)
        if functional_path.exists():
            raise RuntimeError('Participant-level public output cleanup failed.')
        functional_removed.append(str(functional_path.relative_to(functional_output)))
print('Removed participant-level public outputs:',functional_removed)
````

## H. Aggregate-only publication output


````python
# Retain source and aggregates; exclude generated row-level data in this run only.
import zipfile
assert REVISION_WORK.name=='paper_reproduction_run'
assert REVISION_WORK.resolve()==PAPER_ASSETS.parent.resolve()
assert not Path(DATA_DIR).resolve().is_relative_to(REVISION_WORK.resolve())
source_zip=REVISION_WORK/'PAPER_REPRODUCTION_SOURCE.zip'
portable_transport_zip=REVISION_WORK/'transport_appendix_outputs/TRANSPORT_APPENDIX_SOURCE.zip'
with zipfile.ZipFile(portable_transport_zip,'a',zipfile.ZIP_DEFLATED) as archive:
    for name in ['common_fold_manifest.json','archived_reference.json','sensitivity_plan.json']:
        archive.write(PAPER_ASSETS/name,name)
    archive.writestr('README.txt','Run Notebook_Appendix.py with these three adjacent JSON fixtures. The calculations match the exact self-contained reference_transport.py bundled in PAPER_REPRODUCTION_SOURCE.zip. Official data are excluded.\n')
with zipfile.ZipFile(source_zip,'w',zipfile.ZIP_DEFLATED) as archive:
    for directory, prefix in [(PAPER_ASSETS,'reproduction_assets'),(functional_code,'functional')]:
        for p in sorted(directory.rglob('*')):
            if p.is_file() and p.suffix in {'.py','.json','.txt'}:
                archive.write(p,prefix+'/'+str(p.relative_to(directory)))
removed=[]
for p in REVISION_WORK.rglob('*'):
    if not p.is_file():continue
    if p.suffix in {'.csv','.parquet','.pkl','.pyc'} or p.name in {'OOF_PREDICTIONS.json','SCOUTING_CARDS.json','jackknife_fit_audit.json','FOLD_RECORDS.json'}:
        removed.append(str(p.relative_to(REVISION_WORK)));p.unlink()
assert not list(REVISION_WORK.rglob('*.csv'))
print('Aggregate-only output check passed; source and figures retained.')
````
